- Body1,Body2,Body3で連結されたホイストクレーン
- Body1は台車、Body2はワイヤ、Body3は吊り荷である。
- Body1には$\bar{x}_1$方向に制御入力$f$が与えられる。
- Body1は水平方向のみ移動し、$\bar{y}_1, \theta_1$方向には移動しない。
- Body2はBody1と回転が自由な節点1($J_1$)で連結されている。
- Body2はその長さ$l_2$が可変であり、線密度$\rho$であり、その重心位置は$l_2/2$の位置になる。
- Body2の長さ$l_2$を制御し、ここから張力$T$が求まる。
- Body2はBody3から十分な力で引っ張られているため、撓みや曲げが生じないと仮定する。
- Body2はBody3と$l_2$で規制された並進移動と回転が自由な節点$(J_2)$で連結されている。
- Body3は並進と回転が自由である。
- 上線を付けた座標はlocal座標と示し、慣性座標系におけるlocal座標の原点位置を(X,Y)の大文字で示す。

## 運動方程式を求めるステップ

1. Step 0 : Body2を質量ゼロの可変長ワイヤとして導出する。
2. Step 1 : Body2をBody1から無限に巻き出される質量のある可変長ワイヤとして導出する。Body1の質量は変化しない。
3. Step 2 : Body2をBody1にある巻き取りモーターから巻き出される有限量ワイヤとして導出する。Body1の質量も変化する。

また以下では表記の簡略化のため、$S_\theta = \sin(\theta), C_\theta = \cos(\theta)$ とする。

## Step 0 : Body2を質量ゼロの可変長ワイヤとする運動方程式

Body3の重心($X_3,Y_3$)を慣性座標系($x_0, y_0$)で表すと以下となる。

$$
\begin{bmatrix}
X_3 \\ Y_3
\end{bmatrix}
= \begin{bmatrix}
x_1 + l_2 S_{\theta_2} + a_3 S_{\theta_3} \\
-( a_1 + l_2 C_{\theta_2} + a_3 C_{\theta_3})
\end{bmatrix}
$$ 

### 力の関係

節点 $J_1, J_2$に生じる張力は以下の図となる。<br>
節点 $J_1$においてワイヤ長$l_2(t)$が制御される。質量ゼロの理想ワイヤを仮定するため、ワイヤ両端には同じ大きさの張力$T$が作用する。張力$T$は指定された$l_2(t)$を実現するために必要は拘束力として運動方程式から求められる。

<img src="../images/Tension-Body1-Body3.png" style="width:45%;" />
</p>

Body2はワイヤであるため、次のような特性がある。
- ワイヤの軸方向へは引っ張り力を伝える
- 横方向のせん断力は伝えない
- 曲げモーメントは伝えない
- 張力が負になる場合、ワイヤは圧縮力を伝えられず、弛む

よって、ワイヤであるBody2がBody3へ作用させる力は常にBody2の方向である。

重力$m_3 g$はBody3の重心に作用するため、Body3の重心周りの重力によるモーメントは0である。一方、張力$T$は重心から$a_3$離れた節点$J_3$に作用するため、回転トルクを与える。

よって、以下の図のようにBody1とBody3の力とトルクの関係が決まる。

### Body1 並進方向の運動方程式

Body1の$\bar{y}_1, \theta_1$は拘束されている。よって、$x_0 = x_1 = \bar{x}_1$方向の運動方程式は以下となる。

$$
m_1 \ddot{x}_1 = f + T S_{\theta_2}
$$

### Body3 の並進方向運動方程式

運動方程式を導出するため、慣性座標系のBody3の位置($X_3,Y_3$)の速度、加速度を求める。

$$
\begin{bmatrix}
\dot{X}_3 \\ \dot{Y}_3
\end{bmatrix}
= \begin{bmatrix}
\dot{x}_1 + \dot{l}_2 S_{\theta_2} + l_2 C_{\theta_2} \dot{\theta}_2 + a_3 C_{\theta_3} \dot{\theta}_3\\
 - \dot{l}_2 C_{\theta_2} + l_2 S_{\theta_2} \dot{\theta}_2 + a_3 S_{\theta_3} \dot{\theta}_3
\end{bmatrix}
$$ 

$$
\begin{bmatrix}
\ddot{X}_3 \\ \ddot{Y}_3
\end{bmatrix}
= \begin{bmatrix}
\ddot{x}_1 + \ddot{l}_2 S_{\theta_2} + \dot{l}_2 C_{\theta_2} \dot{\theta}_2 
+ \dot{l}_2 C_{\theta_2} \dot{\theta}_2 - l_2 S_{\theta_2} {\dot{\theta}_2}^2 + l_2 C_{\theta_2} \ddot{\theta}_2
- a_3 S_{\theta_3} {\dot{\theta}_3}^2 + a_3 C_{\theta_3} \ddot{\theta}_3\\
- \ddot{l}_2 C_{\theta_2} + \dot{l}_2 S_{\theta_2} \dot{\theta}_2 
+ \dot{l}_2 S_{\theta_2} \dot{\theta}_2 + l_2 C_{\theta_2} {\dot{\theta}_2}^2 + l_2 S_{\theta_2} \ddot{\theta}_2
+ a_3 C_{\theta_3} {\dot{\theta}_3}^2  + a_3 S_{\theta_3} \ddot{\theta}_3
\end{bmatrix}
$$ 

よって以下の運動方程式となる。

$$
\begin{aligned}
m_3 \ddot{X}_3 &= -T S_{\theta_2} \\
m_3 \ddot{Y}_3 &= T C_{\theta_2} - m_3 g
\end{aligned}
$$

### Body3 の回転方向運動方程式

Body3の重心まわりにトルクを生じさせる張力成分は、張力$T$を$\bar{x}_3$方向へ射影した$T_r = T S_{\theta_2 - \theta_3}$である。この力に、$J_2$から重心までの距離$a_3$を掛け、トルクとすると、以下の回転の運動方程式となる。

$$
I_{3} \ddot{\theta}_3 = a_3 T S_{\theta_2 - \theta_3} 
$$


### 式の整理

制御入力$f(t)$と、既知のワイヤ長軌道$l_2$を与える。この時、$l_2(t), \dot{l}_2(t), \ddot{l}_2(t)$ は既知の量である。
$$
m_1 \ddot{x}_1 - T S_{\theta_2}= f 
$$

$$
 m_3 \ddot{x}_1 
+ m_3 l_2 C_{\theta_2} \ddot{\theta}_2
+ m_3 a_3 C_{\theta_3} \ddot{\theta}_3  + T S_{\theta_2} = - m_3 \ddot{l}_2 S_{\theta_2} - 2 m_3 \dot{l}_2 C_{\theta_2} \dot{\theta}_2 + m_3 l_2 S_{\theta_2} {\dot{\theta}_2}^2 + m_3 a_3 S_{\theta_3} {\dot{\theta}_3}^2 \\
$$
$$
m_3 l_2 S_{\theta_2} \ddot{\theta}_2
 + m_3 a_3 S_{\theta_3} \ddot{\theta}_3 - T C_{\theta_2}= - m_3 g + m_3 \ddot{l}_2 C_{\theta_2} - 2 m_3 \dot{l}_2 S_{\theta_2} \dot{\theta}_2 - m_3 l_2 C_{\theta_2} {\dot{\theta}_2}^2 - m_3 a_3 C_{\theta_3} {\dot{\theta}_3}^2 
$$

$$
I_{3} \ddot{\theta}_3 - a_3 T S_{\theta_2 - \theta_3} = 0  
$$



上式の未知変数を$[\ddot{x}_1, \ddot{\theta}_2, \ddot{\theta}_3, T]$として行列形式で表すと、次のようになる。これは加速度$\ddot{x}_1, \ddot{\theta}_2, \ddot{\theta}_3$と拘束力$T$で表した式となっている。


$$
\begin{bmatrix}
m_1 & 0 & 0 & -S_{\theta_2} \\
m_3 & m_3 l_2 C_{\theta_2} & m_3 a_3 C_{\theta_3} & S_{\theta_2}  \\
0 & m_3 l_2 S_{\theta_2} & m_3 a_3 S_{\theta_3} & -  C_{\theta_2} \\
0 & 0 & I_{3} & - a_3 S_{\theta_2 - \theta_3}
\end{bmatrix}
\begin{bmatrix}
\ddot{x}_1 \\ \ddot{\theta}_2 \\ \ddot{\theta}_3 \\ T
\end{bmatrix} = 
\begin{bmatrix}
f \\
- m_3 \ddot{l}_2 S_{\theta_2} - 2 m_3 \dot{l}_2 C_{\theta_2} \dot{\theta}_2 + m_3 l_2 S_{\theta_2} {\dot{\theta}_2}^2 + m_3 a_3 S_{\theta_3} {\dot{\theta}_3}^2 \\
- m_3 g + m_3 \ddot{l}_2 C_{\theta_2} - 2 m_3 \dot{l}_2 S_{\theta_2} \dot{\theta}_2 - m_3 l_2 C_{\theta_2} {\dot{\theta}_2}^2 - m_3 a_3 C_{\theta_3} {\dot{\theta}_3}^2 \\
0
\end{bmatrix}
$$


## Step 1: Body2をBody1から無限に巻き出される質量のある可変長ワイヤとして導出する。

Step 0 では Body2を質量ゼロとして扱ったが、Step1では質量を考える。そして、Step1ではBody1の質量は変化しないという仮定を置く。
