- Body1,Body2,Body3で連結されたホイストクレーン
- Body1は台車、Body2はワイヤ、Body3は吊り荷である。
- Body1には$\bar{x}_1$方向に制御入力$f$が与えられる。
- Body1は水平方向のみ移動し、$\bar{y}_1, \theta_1$方向には移動しない。
- Body2はBody1と回転が自由な節点1($J_1$)で連結されている。
- Body2はその長さ$l_2$が可変であり、線密度$\rho$であり、その重心位置は$l_2/2$の位置になる。
- Body2はの長さ$l_2$は張力$T$により制御される。
- Body2はBody3から十分な力で引っ張られているため、撓みや曲げが生じないと仮定する。
- Body2はBody3と$l_2$で規制された並進移動と回転が自由な節点$(J_2)$で連結されている。
- Body3は並進と回転が自由である。
- 上線を付けた座標はlocal座標と示し、慣性座標系におけるlocal座標の原点位置を(X,Y)の大文字で示す。

## 運動方程式を求めるステップ

1. Body2を質量ゼロの可変長ワイヤとして導出する。
2. Body2をBody1から無限に供給される質量のある可変量ワイヤとして導出する。Body1の質量は変化しない。
3. Body2をBody1にある巻き取りモーターから共有される有限量ワイヤとして導出する。Body1の質量も変化する。

また以下では表記の簡略化のため、$S_\theta = \sin(\theta), C_\theta = \cos(\theta)$ とする。

## Body2を質量ゼロの可変長ワイヤとする運動方程式

Body3の重心($X_3,Y_3$)を慣性座標系($x_0, y_0$)で表すと以下となる。

$$
\begin{bmatrix}
X_3 \\ Y_3
\end{bmatrix}
= \begin{bmatrix}
x_1 + l_2 S_{\theta_2} + a_3 S_{\theta_3} \\
-( a_1 + l_2 C_{\theta_2} + a_3 C_{\theta_3})
\end{bmatrix}
$$ 

### 力の関係

節点 $J_1, J_2$に生じる張力は以下の図となる。<br>
Body2は張力Tが$J_1$の位置で制御されている。そして、反対端の$J_2$ではその反力としてBody3へ張力が伝わる。

<p align="center">
<img src="../images/Tension-Body1-Body3.png" style="width:45%;" />
</p>

Body2はワイヤであるため、次のような特性がある。
- ワイヤの軸方向へは引っ張り力を伝える
- 横方向のせん断力は伝えない
- 曲げモーメントは伝えない
- 圧縮力は伝わらず、張力が負になると縮む

よって、ワイヤであるBody2がBody3へ作用させる力は常にBody2の方向である。

そして、重力はBody3の重心に加わるため、Body3を回転させる力にはならない。そして、接点である$J_3$は固定端ではなく、ワイヤ端であるため、重力はこの$J_3$には伝わらないため、やはり回転させる力にはならない。

よって、以下の図のようにBody1とBody3の力とトルクの関係が決まる。

### Body1 $x_0$方向の運動方程式

Body1の$\bar{y}_1, \theta_1$は拘束されている。よって、$x_0 = x_1 = \bar{x}_1$方向の運動方程式は以下となる。

$$
m_1 \ddot{x}_1 = f + T S_{\theta_2}
$$

### Body3 の並進方向運動方程式

運動方程式を導出するため、($X_3,Y_3$)の速度、加速度を求める。

$$
\begin{bmatrix}
\dot{X}_3 \\ \dot{Y}_3
\end{bmatrix}
= \begin{bmatrix}
\dot{x}_1 + \dot{l}_2 S_{\theta_2} + l_2 C_{\theta_2} \dot{\theta}_2 + a_3 C_{\theta_3} \dot{\theta}_3\\
 - \dot{l}_2 C_{\theta_2} + l_2 S_{\theta_2} \dot{\theta}_2 + a_3 S_{\theta_3} \dot{\theta}_3
\end{bmatrix}
$$ 

$$
\begin{bmatrix}
\ddot{X}_3 \\ \ddot{Y}_3
\end{bmatrix}
= \begin{bmatrix}
\ddot{x}_1 + \ddot{l}_2 S_{\theta_2} + \dot{l}_2 C_{\theta_2} \dot{\theta}_2 
+ \dot{l}_2 C_{\theta_2} \dot{\theta}_2 - l_2 S_{\theta_2} {\dot{\theta}_2}^2 + l_2 C_{\theta_2} \ddot{\theta}_2
- a_3 S_{\theta_3} {\dot{\theta}_3}^2 + a_3 C_{\theta_3} \ddot{\theta}_3\\
- \ddot{l}_2 C_{\theta_2} + \dot{l}_2 S_{\theta_2} \dot{\theta}_2 
+ \dot{l}_2 S_{\theta_2} \dot{\theta}_2 + l_2 C_{\theta_2} {\dot{\theta}_2}^2 + l_2 S_{\theta_2} \ddot{\theta}_2
+ a_3 C_{\theta_3} {\dot{\theta}_3}^2  + a_3 S_{\theta_3} \ddot{\theta}_3
\end{bmatrix}
$$ 

よって以下の運動方程式となる。

$$
\begin{aligned}
m_3 \ddot{X}_3 &= -T S_{\theta_2} \\
m_3 \ddot{Y}_3 &= T C_{\theta_2} - m_3 g
\end{aligned}
$$

### Body3 の回転方向運動方程式

Body3の重心を回転させる力は、張力$T$を$\bar{x}_3$方向へ射影した$T_r = T S_{\theta_2 - \theta_3}$である。この力に、$J_2$から重心までの距離$a_3$を掛け、トルクとすると、以下の回転の運動方式となる。

$$
I_{3} \ddot{\theta}_3 = a_3 T S_{\theta_2 - \theta_3} 
$$
