- Body1,Body2,Body3で連結されたホイストクレーン
- Body1は台車、Body2はワイヤ、Body3は吊り荷である。
- 制御入力はBody1の$\bar{x}_1$方向に与えられる$f$のみとする。
- Body1は水平方向のみ移動し、$\bar{y}_1, \theta_1$方向には移動しない。
- Body2はBody1と回転が自由な節点1($J_1$)で連結されている。
- Body2はその長さ$l_2$が可変であり、線密度$\rho$であり、その重心位置は$l_2/2$の位置になる。
- Body2はBody3から十分な力で引っ張られているため、撓みや曲げが生じないと仮定する。
- Body2はBody3と回転な自由な節点$(J_2)$で連結されている。
- Body3は並進と回転が自由である。
- 上線を付けた座標はlocal座標と示し、慣性座標系におけるlocal座標の原点位置を(X,Y)の大文字で示す。

## 運動方程式を求めるステップ

1. Body2を質量ゼロの可変量ワイヤとして導出する。
2. Body2をBody1から無限に供給される質量のある可変量ワイヤとして導出する。Body1の質量は変化しない。
3. Body2をBody1にある巻き取りモーターから共有される有限量ワイヤとして導出する。Body1の質量も変化する。

また以下では表記の簡略化のため、$S_\theta = \sin(\theta), C_\theta = \cos(\theta)$ とする。

## Body2を質量ゼロの可変長ワイヤとする運動方程式

Body3の重心($X_3,Y_3$)を完成座標系($x_0, y_0$)で表すと以下となる。

$$
\begin{bmatrix}
X_3 \\ Y_3
\end{bmatrix}
= \begin{bmatrix}
l_2 S_{\theta_2} + a_3 S_{\theta_3} \\
-( a_1 + l_2 C_{\theta_2} + a_3 C_{\theta_3})
\end{bmatrix}
$$

### Body1 $x_0$方向の運動方程式

Body1の$\bar{y}_1, \theta_1$は拘束されている。よって、$x_0 = x_1 = \bar{x}_1$方向の運動方程式は以下となる。

$$
m_1 \ddot{x}_1 = f + T S_{\theta_2}
$$

### Body3 の並進方向運動方程式

運動方程式を導出するため、($X_3,Y_3$)の速度、加速度を求める。

$$
\begin{bmatrix}
\dot{X}_3 \\ \dot{Y}_3
\end{bmatrix}
= \begin{bmatrix}
\dot{l}_2 S_{\theta_2} + l_2 C_{\theta_2} \dot{\theta}_2 + a_3 C_{\theta_3} \dot{\theta}_3\\
 - \dot{l}_2 C_{\theta_2} + l_2 S_{\theta_2} \dot{\theta}_2 + a_3 S_{\theta_3} \dot{\theta}_3
\end{bmatrix}
$$ 

$$
\begin{bmatrix}
\ddot{X}_3 \\ \ddot{Y}_3
\end{bmatrix}
= \begin{bmatrix}
\ddot{l}_2 S_{\theta_2} + \dot{l}_2 C_{\theta_2} \dot{\theta}_2 
+ \dot{l}_2 C_{\theta_2} \dot{\theta}_2 - l_2 S_{\theta_2} {\dot{\theta}_2}^2 + l_2 C_{\theta_2} \ddot{\theta}_2
- a_3 S_{\theta_3} {\dot{\theta}_3}^2 + a_3 C_{\theta_3} \ddot{\theta}_3\\
- \ddot{l}_2 C_{\theta_2} + \dot{l}_2 S_{\theta_2} \dot{\theta}_2 
+ \dot{l}_2 S_{\theta_2} \dot{\theta}_2 + l_2 C_{\theta_2} {\dot{\theta}_2}^2 + l_2 S_{\theta_2} \ddot{\theta}_2
+ a_3 C_{\theta_3} {\dot{\theta}_3}^2  + a_3 S_{\theta_3} \ddot{\theta}_3
\end{bmatrix}
$$ 

よって以下の運動方程式となる。

$$
\begin{aligned}
m_3 \ddot{X}_3 &= -T S_{\theta_2} \\
m_3 \ddot{Y}_3 &= T C_{\theta_2} - m_3 g
\end{aligned}
$$
