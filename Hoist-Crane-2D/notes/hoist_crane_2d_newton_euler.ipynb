# Hoist Crane 2D - Newton Euler 法

<p align="center">
<img src="../images/2D-Hoist-Bodys.png" style="width:65%;" />
</p>

- Body1,Body2,Body3で連結されたホイストクレーン
- Body1は台車、Body2はワイヤ、Body3は吊り荷である。
- Body1には$\bar{x}_1$方向に制御入力$f$が与えられる。
- Body1は水平方向のみ移動し、$\bar{y}_1, \theta_1$方向には移動しない。
- Body2はBody1と回転が自由な節点1($J_1$)で連結されている。
- Body2はその長さ$l_2$が可変であり、線密度$\rho$であり、その重心位置は$l_2/2$の位置になる。
- Body2の長さ$l_2$を制御し、ここから張力$T$が求まる。
- Body2はBody3から十分な力で引っ張られているため、撓みや曲げが生じないと仮定する。
- Body2はBody3と$l_2$で規制された並進移動と回転が自由な節点$(J_2)$で連結されている。
- Body3は並進と回転が自由である。
- 上線を付けた座標はlocal座標と示し、慣性座標系におけるlocal座標の原点位置を(X,Y)の大文字で示す。

## 運動方程式を求めるステップ

1. Step 0 : Body2を質量ゼロの可変長ワイヤとして導出する。
2. Step 1 : Body2をBody1から無限に巻き出される質量のある可変長ワイヤとして導出する。Body1の質量は変化しない。
3. Step 2 : Body2をBody1にある巻き取りモーターから巻き出される有限量ワイヤとして導出する。Body1の質量も変化する。

また以下では表記の簡略化のため、$S_\theta = \sin(\theta), C_\theta = \cos(\theta)$ とする。

## Step 0 : Body2を質量ゼロの可変長ワイヤとする運動方程式

Body3の重心($X_3,Y_3$)を慣性座標系($x_0, y_0$)で表すと以下となる。

$$
\begin{bmatrix}
X_3 \\ Y_3
\end{bmatrix}
= \begin{bmatrix}
x_1 + l_2 S_{\theta_2} + a_3 S_{\theta_3} \\
-( a_1 + l_2 C_{\theta_2} + a_3 C_{\theta_3})
\end{bmatrix}
$$ 

### 力の関係

節点 $J_1, J_2$に生じる張力は以下の図となる。<br>
節点 $J_1$においてワイヤ長$l_2(t)$が制御される。質量ゼロの理想ワイヤを仮定するため、ワイヤ両端には同じ大きさの張力$T$が作用する。張力$T$は指定された$l_2(t)$を実現するために必要な拘束力として運動方程式から求められる。

<p align="center">
<img src="../images/Tension-Body1-Body3.png" style="width:45%;" />
</p>

Body2はワイヤであるため、次のような特性がある。
- ワイヤの軸方向へは引っ張り力を伝える
- 横方向のせん断力は伝えない
- 曲げモーメントは伝えない
- 張力が負になる場合、ワイヤは圧縮力を伝えられず、弛む

よって、ワイヤであるBody2がBody3へ作用させる力は常にBody2の方向である。

重力$m_3 g$はBody3の重心に作用するため、Body3の重心周りの重力によるモーメントは0である。一方、張力$T$は重心から$a_3$離れた節点$J_2$に作用するため、回転トルクを与える。

よって、以下の図のようにBody1とBody3の力とトルクの関係が決まる。

</p>


### Body1 並進方向の運動方程式

Body1の$\bar{y}_1, \theta_1$は拘束されている。よって、$x_0 = x_1 = \bar{x}_1$方向の運動方程式は以下となる。

$$
m_1 \ddot{x}_1 = f + T S_{\theta_2}
$$

### Body3 の並進方向運動方程式

運動方程式を導出するため、慣性座標系のBody3の位置($X_3,Y_3$)の速度、加速度を求める。

$$
\begin{bmatrix}
\dot{X}_3 \\ \dot{Y}_3
\end{bmatrix}
= \begin{bmatrix}
\dot{x}_1 + \dot{l}_2 S_{\theta_2} + l_2 C_{\theta_2} \dot{\theta}_2 + a_3 C_{\theta_3} \dot{\theta}_3\\
 - \dot{l}_2 C_{\theta_2} + l_2 S_{\theta_2} \dot{\theta}_2 + a_3 S_{\theta_3} \dot{\theta}_3
\end{bmatrix}
$$ 

$$
\begin{bmatrix}
\ddot{X}_3 \\ \ddot{Y}_3
\end{bmatrix}
= \begin{bmatrix}
\ddot{x}_1 + \ddot{l}_2 S_{\theta_2} + \dot{l}_2 C_{\theta_2} \dot{\theta}_2 
+ \dot{l}_2 C_{\theta_2} \dot{\theta}_2 - l_2 S_{\theta_2} {\dot{\theta}_2}^2 + l_2 C_{\theta_2} \ddot{\theta}_2
- a_3 S_{\theta_3} {\dot{\theta}_3}^2 + a_3 C_{\theta_3} \ddot{\theta}_3\\
- \ddot{l}_2 C_{\theta_2} + \dot{l}_2 S_{\theta_2} \dot{\theta}_2 
+ \dot{l}_2 S_{\theta_2} \dot{\theta}_2 + l_2 C_{\theta_2} {\dot{\theta}_2}^2 + l_2 S_{\theta_2} \ddot{\theta}_2
+ a_3 C_{\theta_3} {\dot{\theta}_3}^2  + a_3 S_{\theta_3} \ddot{\theta}_3
\end{bmatrix}
$$ 

よって以下の運動方程式となる。

$$
\begin{aligned}
m_3 \ddot{X}_3 &= -T S_{\theta_2} \\
m_3 \ddot{Y}_3 &= T C_{\theta_2} - m_3 g
\end{aligned}
$$

### Body3 の回転方向運動方程式

Body3の重心まわりにトルクを生じさせる張力成分は、張力$T$を$\bar{x}_3$方向へ射影した$T_r = T S_{\theta_2 - \theta_3}$である。この力に、$J_2$から重心までの距離$a_3$を掛け、トルクとすると、以下の回転の運動方程式となる。

$$
I_{3} \ddot{\theta}_3 = a_3 T S_{\theta_2 - \theta_3} 
$$


### 式の整理

制御入力$f(t)$と、既知のワイヤ長軌道$l_2$を与える。この時、$l_2(t), \dot{l}_2(t), \ddot{l}_2(t)$ は既知の量である。
$$
m_1 \ddot{x}_1 - T S_{\theta_2}= f 
$$

$$
 m_3 \ddot{x}_1 
+ m_3 l_2 C_{\theta_2} \ddot{\theta}_2
+ m_3 a_3 C_{\theta_3} \ddot{\theta}_3  + T S_{\theta_2} = - m_3 \ddot{l}_2 S_{\theta_2} - 2 m_3 \dot{l}_2 C_{\theta_2} \dot{\theta}_2 + m_3 l_2 S_{\theta_2} {\dot{\theta}_2}^2 + m_3 a_3 S_{\theta_3} {\dot{\theta}_3}^2 \\
$$
$$
m_3 l_2 S_{\theta_2} \ddot{\theta}_2
 + m_3 a_3 S_{\theta_3} \ddot{\theta}_3 - T C_{\theta_2}= - m_3 g + m_3 \ddot{l}_2 C_{\theta_2} - 2 m_3 \dot{l}_2 S_{\theta_2} \dot{\theta}_2 - m_3 l_2 C_{\theta_2} {\dot{\theta}_2}^2 - m_3 a_3 C_{\theta_3} {\dot{\theta}_3}^2 
$$

$$
I_{3} \ddot{\theta}_3 - a_3 T S_{\theta_2 - \theta_3} = 0  
$$



上式の未知変数を$[\ddot{x}_1, \ddot{\theta}_2, \ddot{\theta}_3, T]$として行列形式で表すと、次のようになる。これは加速度$\ddot{x}_1, \ddot{\theta}_2, \ddot{\theta}_3$と拘束力$T$で表した式となっている。


$$
\begin{bmatrix}
m_1 & 0 & 0 & -S_{\theta_2} \\
m_3 & m_3 l_2 C_{\theta_2} & m_3 a_3 C_{\theta_3} & S_{\theta_2}  \\
0 & m_3 l_2 S_{\theta_2} & m_3 a_3 S_{\theta_3} & -  C_{\theta_2} \\
0 & 0 & I_{3} & - a_3 S_{\theta_2 - \theta_3}
\end{bmatrix}
\begin{bmatrix}
\ddot{x}_1 \\ \ddot{\theta}_2 \\ \ddot{\theta}_3 \\ T
\end{bmatrix} = 
\begin{bmatrix}
f \\
- m_3 \ddot{l}_2 S_{\theta_2} - 2 m_3 \dot{l}_2 C_{\theta_2} \dot{\theta}_2 + m_3 l_2 S_{\theta_2} {\dot{\theta}_2}^2 + m_3 a_3 S_{\theta_3} {\dot{\theta}_3}^2 \\
- m_3 g + m_3 \ddot{l}_2 C_{\theta_2} - 2 m_3 \dot{l}_2 S_{\theta_2} \dot{\theta}_2 - m_3 l_2 C_{\theta_2} {\dot{\theta}_2}^2 - m_3 a_3 C_{\theta_3} {\dot{\theta}_3}^2 \\
0
\end{bmatrix}
$$


この式でシミュレーションを行うには、上式を$A_k z_k = b_k$と表現し、次の流れで計算を行う。

- 1. 初期状態を決める

        $\boldsymbol{x}_0 = [x_{1,0} , \dot{x}_{1,0}, \theta_{2,0}, \dot{\theta}_{2,0}, \theta_{3,0}, \dot{\theta}_{3,0}]^T$
- 2. 初期入力を決める

        $\boldsymbol{u}_0 = [f_{0}, l_{2,0}, \dot{l}_{2,0}, \ddot{l}_{2,0}]^T$

- 3. 初期状態と初期入力により、$A_0, b_0$が決まるため、以下のように$z_0$を求める。

        $z_0 = A_0^{-1} b_0$

- 4. 状態 $\boldsymbol{x}_1$ を次のように更新する。これは、計算による発散を防止するsemi-implicit Euler 法である。

        $
        \dot{x}_{1,1} = \dot{x}_{1,0} + \Delta t \ \ddot{x}_{1,0}\\
        {x}_{1,1} = {x}_{1,0} + \Delta t \ \dot{x}_{1,1}\\
        \dot{\theta}_{2,1} = \dot{\theta}_{2,0} + \Delta t \ddot{\theta}_{2,0} \\
        {\theta}_{2,1} = {\theta}_{2,0} + \Delta t \dot{\theta}_{2,1} \\
        \dot{\theta}_{3,1} = \dot{\theta}_{3,0} + \Delta t \ddot{\theta}_{3,0} \\
        {\theta}_{3,1} = {\theta}_{3,0} + \Delta t \dot{\theta}_{3,1} \\
        $

- 5. 以下を繰り返す
    - 入力 $\boldsymbol{u}_k$ の決定
    - $z_k = A_k^{-1} b_k$ の計算
    - $\boldsymbol{x}_k$ の更新



## Step 1: Body2をBody1から無限に巻き出される質量のある可変長ワイヤとして導出する。

Step 0 では Body2を質量ゼロとして扱ったが、Step1では質量を考える。そして、Step1ではBody1の質量は変化しないという仮定を置く。
また、質量をもつ柔軟なワイヤは、一般的には運動中に曲がる。しかしStep1では、Body2へ働く張力が十分に大きく、たわみや曲率を無視して巻き出されたワイヤを直線として近似する。

### Body2の定義

#### 質量の流入と運動量について

Body2の質量はその長さ$l_2$と線密度$\rho$により次のようになる。$l_2$は時変のため、質量も時変である。

$$
m_2(t) = \rho l_2(t)
$$

また、この時間微分を行うと、質量の変化として以下の関係を得る。

$$
\dot{m}_2(t) = \rho \dot{l}_2(t)
$$

巻き出し中である場合、$\dot{l}_2 > 0$ であるため単位時間で$\rho \dot{l}_2$の質量が節点$J_1$からBody2へ流入する。

このように質点が境界を通過する系を検査体積、または開放系と呼ぶ。Body2には$J_1$を通過して新しいワイヤ材料が入ってくる、つまり、Body2を構成する物質粒子が時間とともに入れ替わる。

ここで重要になるので、Body2を構成する物体による質量の運動量の時間微分による慣性力と作用力のつり合いが取れないことである。

$$
\sum F = \frac{d}{dt}(m_2 \boldsymbol{v}_{G_2})
$$

それは、新しく流入する質量が運動量をBody2へ持ち込むからである。

##### ワイヤ上の座標の定義

<p align="center">
<img src="../images/body2-cordinate-system.png" style="width:30%;" />
</p>


$J_1$からワイヤに沿った距離を$s$とする。

$$
0 \le s \le l_2(t)
$$

- $s=0$ : 節点$J_1$
- $s=l_2(t)$ : 節点 $J_2$

ワイヤの下向き軸方向の単位ベクトルを$e_r$、その垂直方向を$e_t$とする。

$$
\boldsymbol{e}_r = \begin{bmatrix} S_{\theta_2} \\ - C_{\theta_2} \end{bmatrix}, \quad
\boldsymbol{e}_t = \begin{bmatrix} C_{\theta_2} \\ S_{\theta_2} \end{bmatrix}
$$

この時間微分は次の関係となる。

$$
\dot{\boldsymbol{e}}_r = \begin{bmatrix} C_{\theta_2} \dot{\theta}_2 \\ S_{\theta_2} \dot{\theta}_2 \end{bmatrix} = \dot{\theta}_2 \boldsymbol{e}_t , \quad
\dot{\boldsymbol{e}}_t = \begin{bmatrix} -S_{\theta_2} \dot{\theta}_2  \\ C_{\theta_2} \dot{\theta}_2 \end{bmatrix} = - \dot{\theta}_2 \boldsymbol{e}_r 
$$

慣性座標で節点$J_1$の位置を$\boldsymbol{r}_{J_1}$とする。

$$
\boldsymbol{r}_{J_1} = \begin{bmatrix} x_1 \\ -a_1 \end{bmatrix}
$$

よって、慣性座標上でのワイヤ上の位置は$\boldsymbol{r}$は、$J_1$からワイヤに沿った距離$s$を用いて以下となる。

$$
\boldsymbol{r}(s,t) = \boldsymbol{r}_{J_1}(t) + s \ \boldsymbol{e}_r(t)
$$



巻き出されるワイヤ材料の位置をワイヤ上の特定の位置を示す$s$で追跡すると、その位置は時間とともに変化する($s=s(t)$)。非伸縮ワイヤが一様な速度で巻き出されると仮定し、以下のように定義する。

$$
\frac{d s}{dt} = \dot{l}_2
$$

ワイヤ上の位置$\boldsymbol{r}$の速度$\boldsymbol{v}(s,t)$は、この時間微分であるため、

$$
\boldsymbol{v}(s(t),t) = \frac{d}{dt}\left( \boldsymbol{r}(s(t),t) \right) = \frac{d}{dt}\left( \boldsymbol{r}_{J_1}(t) + s \ \boldsymbol{e}_r(t) \right)
$$

より、以下となる。

$$
\boldsymbol{v}(s,t) = \boldsymbol{v}_{J_1}(t) + \dot{l}_2 \ \boldsymbol{e}_r(t) + s \ \dot{\theta}_2 \boldsymbol{e}_t(t)
$$

各項の意味は

- $\boldsymbol{v}_{J_1}(t)$ : Body1の並進速度
- $\dot{l}_2 \ \boldsymbol{e}_r(t)$ : ワイヤの巻き出し速度
- $s \ \dot{\theta}_2 \boldsymbol{e}_t(t)$ : ワイヤの旋回による接線速度


##### 運動量の差

Body2の幾何学的な重心位置$\boldsymbol{r}_{G_2}$は以下となる。

$$
\boldsymbol{r}_{G_2} = \boldsymbol{r}_{J_1} + \frac{l_2}{2} \boldsymbol{e}_r
$$

この時間微分は以下となる。

$$
\dot{\boldsymbol{r}}_{G_2} = \boldsymbol{v}_{J_1} + \frac{\dot{l}_2}{2} \boldsymbol{e}_r + \frac{l_2}{2} \dot{\theta}_2 \boldsymbol{e}_t
$$

よって運動量は次のように計算される。

$$
m_2 \dot{\boldsymbol{r}}_{G_2} =\rho l_2 \dot{\boldsymbol{r}}_{G_2} = \rho l_2 \boldsymbol{v}_{J_1} + \frac{1}{2} \rho l_2 \dot{l}_2 \boldsymbol{e}_r + \frac{1}{2} \rho l_2^2 \dot{\theta}_2 \boldsymbol{e}_t
$$

しかし、ワイヤ全体の本当の運動量$\boldsymbol{P}_2$はワイヤ上の速度$\boldsymbol{v}(s,t)$を用いて以下のように計算される。

$$
\boldsymbol{P}_2 = \int_0^{l_2} \rho \boldsymbol{v}(s,t) ds
$$

これを計算すると、

$$
\begin{aligned}
\boldsymbol{P}_2 &= \int_0^{l_2} \rho \left( \boldsymbol{v}_{J_1} + \dot{l}_2 \ \boldsymbol{e}_r + s \ \dot{\theta}_2 \boldsymbol{e}_t \right) ds \\
&= \rho l_2 \boldsymbol{v}_{J_1} + \rho l_2 \dot{l}_2 \ \boldsymbol{e}_r + \frac{1}{2}\rho l_2^2  \ \dot{\theta}_2 \boldsymbol{e}_t  \\
\end{aligned}
$$

よって、$m_2 \dot{\boldsymbol{r}}_{G_2} \neq \boldsymbol{P}_2 $ となる。これは、Body2へ新しい質量が流入しているためである。その差分は以下のように第２項の軸方向の速度で現れる。

- $m_2 \dot{\boldsymbol{r}}_{G_2}$ : $\dfrac{1}{2} \rho l_2 \dot{l}_2 \boldsymbol{e}_r$
- $\boldsymbol{P}_2$ : $\rho l_2 \dot{l}_2 \ \boldsymbol{e}_r$


#### 質量収支

Body2への流入質量流量は長さ$l_2$の時間変化に線密度$\rho$を掛けた値である。

$$
\dot{m}_{in} = \rho \dot{l}_2
$$

流入する質量が$J_1$を通過した直後の速度は、ワイヤ上の位置の速度$\boldsymbol{v}(s,t)$で$s=0$であるため、以下となる。

$$
v_{in} = \boldsymbol{v}_{J_1} + \dot{l}_2 \ \boldsymbol{e}_r
$$

#### 開放系の運動量収支

質量流入だけがある場合、運動量収支は以下のようになる。

$$
\sum \boldsymbol{F}_{\text{ext}} = \frac{d \boldsymbol{P}_2}{dt} - \dot{m}_{in} \boldsymbol{v}_{in}
$$

これは、以下のようにBody2の運動量の変化は、外力の合計と流入質量が外部からもたらす運動量を合計したものとなる。

$$
\frac{d \boldsymbol{P}_2}{dt} = \sum \boldsymbol{F}_{\text{ext}} + \dot{m}_{in} \boldsymbol{v}_{in} 
$$

運動量収支の右辺を計算する。

$$
\begin{aligned}
\frac{d \boldsymbol{P}_2}{dt } &= \rho \dot{l}_2 \boldsymbol{v}_{J_1} + \rho l_2 \dot{\boldsymbol{v}}_{J_1} 
+ \rho \dot{l}_2^2 \ \boldsymbol{e}_r + \rho l_2 \ddot{l}_2 \ \boldsymbol{e}_r + \rho l_2 \dot{l}_2 \ \dot{\theta}_2 \boldsymbol{e}_t 
+ \rho l_2 \dot{l}_2\ \dot{\theta}_2 \boldsymbol{e}_t + \frac{1}{2}\rho l_2^2 \ \ddot{\theta}_2 \boldsymbol{e}_t - \frac{1}{2}\rho l_2^2 \ \dot{\theta}_2^2 \boldsymbol{e}_r \\
&= \rho \dot{l}_2 \boldsymbol{v}_{J_1} + \rho l_2 \dot{\boldsymbol{v}}_{J_1} 
+ \rho \left( \dot{l}_2^2 + l_2 \ddot{l}_2  - \frac{1}{2} l_2^2 \ \dot{\theta}_2^2 \right) \boldsymbol{e}_r + \rho \left( 2 l_2 \dot{l}_2 \ \dot{\theta}_2 + \frac{1}{2} l_2^2 \ \ddot{\theta}_2 \right) \boldsymbol{e}_t 
\end{aligned} 
$$

$$
\dot{m}_{in} \boldsymbol{v}_{in} = \rho \dot{l}_2 \boldsymbol{v}_{J_1} + \rho \dot{l}_2^2 \ \boldsymbol{e}_r
$$

よって、以下のようになる。

$$
\frac{d \boldsymbol{P}_2}{dt} - \dot{m}_{in} \boldsymbol{v}_{in} = \rho l_2 \dot{\boldsymbol{v}}_{J_1} + \rho \left(l_2 \ddot{l}_2  - \frac{1}{2} l_2^2 \ \dot{\theta}_2^2 \right) \boldsymbol{e}_r + \rho \left( 2 l_2 \dot{l}_2 \ \dot{\theta}_2 + \frac{1}{2} l_2^2 \ \ddot{\theta}_2 \right) \boldsymbol{e}_t
$$

これがBody2に対する並進運動量の変化である。$\dot{l}_2=\ddot{l}_2=0$であれば、固定長の質量分布の式となる。

#### Body2に作用する外力

<p align="center">
<img src="../images/body2-ext-force.png" style="width:30%;" />
</p>

- $T_1$ : $J_1$側の張力
- $T_2$ : $J_2$側の張力

質量のあるワイヤでは、ワイヤ自身の向心力や接線力、また重力も支える必要があるため、$T_1 \neq T_2$となる。

よってBody2に作用する外力の合計は

$$
\sum \boldsymbol{F}_{ext} = -T_1 \boldsymbol{e}_r + T_2 \boldsymbol{e}_r + m_2 \begin{bmatrix} 0 \\ -g \end{bmatrix}
$$

#### 並進の運動方程式

よってBody2の並進の運動方程式は以下となる。


$$
\sum \boldsymbol{F}_{\text{ext}} = \frac{d \boldsymbol{P}_2}{dt} - \dot{m}_{in} \boldsymbol{v}_{in}
$$


$$
-T_1 \boldsymbol{e}_r + T_2 \boldsymbol{e}_r + m_2 \begin{bmatrix} 0 \\ -g \end{bmatrix} = \rho l_2 \dot{\boldsymbol{v}}_{J_1} + \rho \left(l_2 \ddot{l}_2  - \frac{1}{2} l_2^2 \ \dot{\theta}_2^2 \right) \boldsymbol{e}_r + \rho \left( 2 l_2 \dot{l}_2 \ \dot{\theta}_2 + \frac{1}{2} l_2^2 \ \ddot{\theta}_2 \right) \boldsymbol{e}_t
$$


$$
\begin{bmatrix}
 (T_2 - T_1)S_{\theta_2} \\[8pt]
 - (T_2 - T_1)C_{\theta_2} -\rho l_2 g \\
\end{bmatrix} = 
\begin{bmatrix}
\rho l_2 \ddot{x}_1 + \rho \left(l_2 \ddot{l}_2  - \frac{1}{2} l_2^2 \ \dot{\theta}_2^2 \right) S_{\theta_2} + \rho \left( 2 l_2 \dot{l}_2 \ \dot{\theta}_2 + \frac{1}{2} l_2^2 \ \ddot{\theta}_2 \right)C_{\theta_2} \\[8pt]
- \rho \left(l_2 \ddot{l}_2  - \frac{1}{2} l_2^2  \dot{\theta}_2^2 \right) C_{\theta_2}  + \rho \left( 2 l_2 \dot{l}_2 \ \dot{\theta}_2 + \frac{1}{2} l_2^2 \ \ddot{\theta}_2 \right) S_{\theta_2} \\[8pt]
\end{bmatrix}
$$


#### 回転の運動方程式を立てる

Body2の上端$J_1$まわりで角運動量を導出する。

ワイヤ上の$J_1$からのある点までの位置ベクトルは次で表される。

$$
\boldsymbol{r}_{J_1-P} = s \boldsymbol{e}_r
$$

この点の材料速度は次となる。

$$
\boldsymbol{v}(s,t) = \boldsymbol{v}_{J_1} + \dot{l}_2 \boldsymbol{e}_r + s \dot{\theta}_2 \boldsymbol{e}_t
$$

$J_1$まわりの角運動量$H_{J_1}$は、$J_1$ から微小質量 $\rho ds$ までの距離 $s \boldsymbol{e}_r$とその質点の速度 $\boldsymbol{v}(s,t)$ の外積を全体積分の積算で表される。

$$
H_{J_1} = \int_0^{l_2} (s \boldsymbol{e}_r) \times \boldsymbol{v}(s,t) \rho ds
$$

<img src="../images/angular_moment_of_body2.png" style="width:25%;" />
</p>

速度を代入して計算すると、

$$
\begin{aligned}
H_{J_1} &= \int_0^{l_2}  \rho(s \boldsymbol{e}_r) \times \left( \boldsymbol{v}_{J_1} + \dot{l}_2 \boldsymbol{e}_r + s \dot{\theta}_2 \boldsymbol{e}_t
\right)  ds \\
&= \int_0^{l_2}  \rho s ( \boldsymbol{e}_r \times \boldsymbol{v}_{J_1}) ds + \int_0^{l_2}  \rho s^2 \dot{\theta}_2( \boldsymbol{e}_r \times  \boldsymbol{e}_t) ds
\end{aligned}
$$

$\boldsymbol{e}_r \times  \boldsymbol{e}_t = [0, 0, 1]^T$となるため、これを $\theta_2$ 軸の回転軸と捉える。すると角運動量$H_{J_1}$は以下となる。

$$
H_{J_1} = \frac{1}{2} \rho l_2^2 ( \boldsymbol{e}_r \times \boldsymbol{v}_{J_1}) + \frac{1}{3} \rho l_2^3 \dot{\theta}_2
$$

角運動量を時間微分して整理する。

$$
\begin{aligned}
\dot{H}_{J_1} &= 
 \rho l_2 \dot{l}_2 ( \boldsymbol{e}_r \times \boldsymbol{v}_{J_1})
+\frac{1}{2} \rho l_2^2 (\dot{\theta_2} \boldsymbol{e}_t \times \boldsymbol{v}_{J_1} + \boldsymbol{e}_r \times \dot{\boldsymbol{v}}_{J_1})
+ \rho l_2^2 \dot{l}_2 \dot{\theta}_2 + \frac{1}{3} \rho l_2^3 \ddot{\theta}_2 \\
 &= \rho l_2 \dot{l}_2 ( \boldsymbol{e}_r \times \boldsymbol{v}_{J_1}) +\frac{1}{2} \rho l_2^2 \dot{\theta_2} (\boldsymbol{e}_t \times \boldsymbol{v}_{J_1})
 +\frac{1}{2} \rho l_2^2 (\boldsymbol{e}_r \times \dot{\boldsymbol{v}}_{J_1}) + \rho l_2^2 \dot{l}_2 \dot{\theta}_2 + \frac{1}{3} \rho l_2^3 \ddot{\theta}_2  \\
 &= - (\boldsymbol{v}_{J_1} \times \rho l_2 \dot{l}_2 \ \boldsymbol{e}_r ) - ( \boldsymbol{v}_{J_1} \times \frac{1}{2} \rho l_2^2 \dot{\theta_2} \ \boldsymbol{e}_t) +\frac{1}{2} \rho l_2^2 (\boldsymbol{e}_r \times \dot{\boldsymbol{v}}_{J_1}) + \rho l_2^2 \dot{l}_2 \dot{\theta}_2 + \frac{1}{3} \rho l_2^3 \ddot{\theta}_2  \\
 &= - \boldsymbol{v}_{J_1} \times  \left(\rho l_2 \dot{l}_2 \ \boldsymbol{e}_r + \frac{1}{2} \rho l_2^2 \dot{\theta_2} \ \boldsymbol{e}_t \right) +\frac{1}{2} \rho l_2^2 (\boldsymbol{e}_r \times \dot{\boldsymbol{v}}_{J_1}) + \rho l_2^2 \dot{l}_2 \dot{\theta}_2 + \frac{1}{3} \rho l_2^3 \ddot{\theta}_2  \\
\end{aligned}
$$

前節で求めたBody2の並進の運動量$\boldsymbol{P}_2$は以下であった。

$$
\boldsymbol{P}_2 = \rho l_2 \boldsymbol{v}_{J_1} + \rho l_2 \dot{l}_2 \ \boldsymbol{e}_r + \frac{1}{2}\rho l_2^2  \ \dot{\theta}_2 \boldsymbol{e}_t 
$$

$\boldsymbol{P}_2$ と $\boldsymbol{v}_{J_1}$ の外積を取ると以下となる。

$$
\begin{aligned}
\boldsymbol{v}_{J_1} \times \boldsymbol{P}_2 &= \boldsymbol{v}_{J_1} \times \left(\rho l_2 \boldsymbol{v}_{J_1} + \rho l_2 \dot{l}_2 \ \boldsymbol{e}_r + \frac{1}{2}\rho l_2^2  \ \dot{\theta}_2 \boldsymbol{e}_t \right) \\
&= \boldsymbol{v}_{J_1} \times \left(\ \rho l_2 \dot{l}_2 \ \boldsymbol{e}_r + \frac{1}{2}\rho l_2^2  \ \dot{\theta}_2 \boldsymbol{e}_t \right)
\end{aligned}
$$

よって、次のようにまとめる。

$$
\dot{H}_{J_1} = - \boldsymbol{v}_{J_1} \times \boldsymbol{P}_2 +\frac{1}{2} \rho l_2^2 (\boldsymbol{e}_r \times \dot{\boldsymbol{v}}_{J_1}) + \rho l_2^2 \dot{l}_2 \dot{\theta}_2 + \frac{1}{3} \rho l_2^3 \ddot{\theta}_2 
$$

次のように表現すると、

- $m_2 = \rho l_2$ : Body2の質量
- $I_{J_1} = \frac{1}{3}\rho l_2^3$ : Body2の$J_1$周りの慣性モーメント
- $(\boldsymbol{e}_r \times \dot{\boldsymbol{v}}_{J_1})_{z} = \boldsymbol{e}_t \cdot \dot{\boldsymbol{v}}_{J_1} = C_{\theta_2} \ddot{x}_1$
 : Body1の加速度$\ddot{x}_1$のBody2の接線方向への射影

後ろの三つの項はそれぞれ、次のように意味なる。

$J_1$の並進加速度の接線方向成分$C_{\theta_2} \ddot{x}_1$ に、ワイヤ質量$m_2$と重心までの距離$l_2/2$を掛けた慣性項

$$
\frac{1}{2} \rho l_2^2 (\boldsymbol{e}_r \times \dot{\boldsymbol{v}}_{J_1}) = \frac{1}{2} l_2 m_2 C_{\theta_2} \ddot{x}_1
$$

巻き出し$\dot{l}_2$によって、旋回に伴う角運動量の変化によるモーメント項

$$
\dot{I}_{J_1} = \rho l_2^2 \dot{l}_2  \rightarrow \dot{I}_{J_1} \dot{\theta}_2 =  \rho l_2^2 \dot{l}_2 \dot{\theta}_2 
$$

$J_1$周りの慣性モーメントと角加速度$\ddot{\theta}_2$による慣性項

$$
\frac{1}{3} \rho l_2^3 \ddot{\theta}_2 = \frac{1}{3} m_2 l_2^2 \ddot{\theta}_2 
$$

</p>

流入点$J_1$では腕の長さがゼロなので、流入する角運動量の流速はゼロである。また、下端$J_2$では材料の出入りがない。したがって、上記の式を慣性項を左側へまとめて表現すると、

$$
\frac{1}{2} l_2 m_2 C_{\theta_2} \ddot{x}_1 + \rho l_2^2 \dot{l}_2 \dot{\theta}_2 + \frac{1}{3} \rho l_2^3 \ddot{\theta}_2 = \dot{H}_{J_1} + \boldsymbol{v}_{J_1} \times \boldsymbol{P}_2
$$
