## 1. 慣性テンソル

### 1.1. 質点の運動量

質点の運動量は並進運動であれば以下のように、質量と速度の積として以下の式で表される。

$$
P=mv
$$

質点が回転している場合は角運動量$L$となり、質点の運動量に回転中心から質点までの距離$l$を用いて以下の式となる。

$$
L = l \times mv
$$

そして、接線速度 $v$ は以下であるため、

$$
v = \omega \times l
$$

角運動量は以下となる。

$$
L = m \cdot \left( l \times (\omega \times l) \right)
$$

### 1.2. 剛体の運動量

剛体が世界座標系で示された $\omega$ で回転しているとする。1.1.節では質点についてであったが、剛体の場合は以下の図のように、微小質量 $dm$ と物体座標系原点から微小質量までの距離 $l$、そして$\omega$ により、以下のように、剛体容積分の積分が角運動量となる。ここでは物体座標系の原点は物体の重心にあるとし、物体座標系の原点と世界座標系の原点が一致しているとする。

$$
L = \int_B \left( l \times (\omega \times l) \right)  dm
$$

</p>


$$
l = \begin{bmatrix} l_x \\ l_y \\ l_z \end{bmatrix}, \quad \omega = \begin{bmatrix} \omega_x \\ \omega_y \\ \omega_z \end{bmatrix}
$$

とすると、

$$
\omega \times l = \begin{bmatrix} \omega_x \\ \omega_y \\ \omega_z \end{bmatrix} \times \begin{bmatrix} l_x \\ l_y \\ l_z \end{bmatrix} = 
\begin{bmatrix}
\omega_y l_z - \omega_z l_y \\
-\omega_x l_z + \omega_z l_x \\
\omega_x l_y - \omega_y l_x
\end{bmatrix}
$$

$$
l \times (\omega \times l) = \begin{bmatrix} l_x \\ l_y \\ l_z \end{bmatrix} \times \begin{bmatrix}
\omega_y l_z - \omega_z l_y \\ -\omega_x l_z + \omega_z l_x \\ \omega_x l_y - \omega_y l_x
\end{bmatrix} =
\begin{bmatrix}
l_y (\omega_x l_y - \omega_y l_x) - l_z ( -\omega_x l_z + \omega_z l_x) \\
- l_x (\omega_x l_y - \omega_y l_x) + l_z (\omega_y l_z - \omega_z l_y ) \\
l_x (-\omega_x l_z + \omega_z l_x) - l_y (\omega_y l_z - \omega_z l_y)
\end{bmatrix}
$$

上式を $\omega$ で整理すると、次のようになる。

$$
\begin{bmatrix}
(l_y^2 + l_z^2) \omega_x - l_x l_y \omega_y - l_x l_z \omega_z \\
-l_x l_y \omega_x + (l_x^2 + l_z^2) \omega_y - l_y l_z \omega_z \\
- l_x l_z \omega_x - l_y l_z \omega_y +(l_x^2 + l_y^2) \omega_z
\end{bmatrix} = 
\begin{bmatrix}
(l_y^2 + l_z^2)  & - l_x l_y & - l_x l_z \\
-l_x l_y &  (l_x^2 + l_z^2) & - l_y l_z \\
- l_x l_z & - l_y l_z & (l_x^2 + l_y^2) 
\end{bmatrix}
\begin{bmatrix} \omega_x \\ \omega_y \\ \omega_z \end{bmatrix}
$$

よって、角運動量 $L$ は以下のようになる。

$$
L = \begin{bmatrix}
\displaystyle\int (l_y^2 + l_z^2) dm  & - \displaystyle\int l_x l_y dm& - \displaystyle\int l_x l_z dm\\
- \displaystyle\int l_x l_y dm&  \displaystyle\int (l_x^2 + l_z^2) dm & - \displaystyle\int l_y l_z dm \\
- \displaystyle\int l_x l_z dm& - \displaystyle\int l_y l_z dm & \displaystyle\int (l_x^2 + l_y^2) dm 
\end{bmatrix}
\begin{bmatrix} \omega_x \\ \omega_y \\ \omega_z \end{bmatrix}
$$

慣性モーメントを対角項から以下のように定義し、

$$
\begin{aligned}
I_{xx} &= \int (l_y^2 + l_z^2) dm \\
I_{yy} &= \int (l_x^2 + l_z^2) dm \\
I_{zz} &= \int (l_x^2 + l_y^2) dm \\
\end{aligned}
$$

慣性乗積を非対角項から以下のように定義する。

$$
\begin{aligned}
I_{xy} = I_{yx} &= \int (l_x l_y) dm \\
I_{yz} = I_{zy} &= \int (l_y l_z) dm \\
I_{zx} = I_{xz} &= \int (l_x l_z) dm \\
\end{aligned}
$$

すると角運動量は以下で表すことができ、この右辺の行列を慣性テンソルと呼ぶ。

$$
L = \begin{bmatrix}
I_{xx}  & - I_{xy} & - I_{xz}\\
- I_{xy} &  I_{yy}  & - I_{yz} \\
- I_{xz} & - I_{zy} & I_{zz} 
\end{bmatrix}
\begin{bmatrix} \omega_x \\ \omega_y \\ \omega_z \end{bmatrix}
$$

この式は物体の重心周りの角運動量であるため、角運動量と慣性テンソルを重心周りであることを示すため$G$を付け、以下のように記述する。

$$
L_G = I_G \ \omega
$$
