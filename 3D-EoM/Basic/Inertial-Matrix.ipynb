## 1. 慣性テンソル

### 1.1. 質点の運動量

質点の運動量は並進運動であれば以下のように、質量と速度の積として以下の式で表される。

$$
P=mv
$$

質点が回転している場合は角運動量$L$となり、質点の運動量に回転中心から質点までの距離$l$を用いて以下の式となる。

$$
L = l \times mv
$$

そして、接線速度 $v$ は以下であるため、

$$
v = \omega \times l
$$

角運動量は以下となる。

$$
L = m \cdot \left( l \times (\omega \times l) \right)
$$

### 1.2. 剛体の運動量

剛体が世界座標系で示された $\omega$ で回転しているとする。1.1.節では質点についてであったが、剛体の場合は以下の図のように、微小質量 $dm$ と物体座標系原点から微小質量までの距離 $l$、そして重心周りの角速度 $\omega$ により、以下のように、剛体容積分の積分が角運動量となる。ここでは、$l$と$\omega$が世界座標で表されているとする。

$$
L = \int_B \left( l \times (\omega \times l) \right)  dm
$$

</p>


$$
l = \begin{bmatrix} l_x \\ l_y \\ l_z \end{bmatrix}, \quad \omega = \begin{bmatrix} \omega_x \\ \omega_y \\ \omega_z \end{bmatrix}
$$

とすると、

$$
\omega \times l = \begin{bmatrix} \omega_x \\ \omega_y \\ \omega_z \end{bmatrix} \times \begin{bmatrix} l_x \\ l_y \\ l_z \end{bmatrix} = 
\begin{bmatrix}
\omega_y l_z - \omega_z l_y \\
-\omega_x l_z + \omega_z l_x \\
\omega_x l_y - \omega_y l_x
\end{bmatrix}
$$

$$
l \times (\omega \times l) = \begin{bmatrix} l_x \\ l_y \\ l_z \end{bmatrix} \times \begin{bmatrix}
\omega_y l_z - \omega_z l_y \\ -\omega_x l_z + \omega_z l_x \\ \omega_x l_y - \omega_y l_x
\end{bmatrix} =
\begin{bmatrix}
l_y (\omega_x l_y - \omega_y l_x) - l_z ( -\omega_x l_z + \omega_z l_x) \\
- l_x (\omega_x l_y - \omega_y l_x) + l_z (\omega_y l_z - \omega_z l_y ) \\
l_x (-\omega_x l_z + \omega_z l_x) - l_y (\omega_y l_z - \omega_z l_y)
\end{bmatrix}
$$

上式を $\omega$ で整理すると、次のようになる。

$$
\begin{bmatrix}
(l_y^2 + l_z^2) \omega_x - l_x l_y \omega_y - l_x l_z \omega_z \\
-l_x l_y \omega_x + (l_x^2 + l_z^2) \omega_y - l_y l_z \omega_z \\
- l_x l_z \omega_x - l_y l_z \omega_y +(l_x^2 + l_y^2) \omega_z
\end{bmatrix} = 
\begin{bmatrix}
(l_y^2 + l_z^2)  & - l_x l_y & - l_x l_z \\
-l_x l_y &  (l_x^2 + l_z^2) & - l_y l_z \\
- l_x l_z & - l_y l_z & (l_x^2 + l_y^2) 
\end{bmatrix}
\begin{bmatrix} \omega_x \\ \omega_y \\ \omega_z \end{bmatrix}
$$

よって、角運動量 $L$ は以下のようになる。

$$
L = \begin{bmatrix}
\displaystyle\int (l_y^2 + l_z^2) dm  & - \displaystyle\int l_x l_y dm& - \displaystyle\int l_x l_z dm\\
- \displaystyle\int l_x l_y dm&  \displaystyle\int (l_x^2 + l_z^2) dm & - \displaystyle\int l_y l_z dm \\
- \displaystyle\int l_x l_z dm& - \displaystyle\int l_y l_z dm & \displaystyle\int (l_x^2 + l_y^2) dm 
\end{bmatrix}
\begin{bmatrix} \omega_x \\ \omega_y \\ \omega_z \end{bmatrix}
$$

慣性モーメントを対角項から以下のように定義し、

$$
\begin{aligned}
I_{xx} &= \int (l_y^2 + l_z^2) dm \\
I_{yy} &= \int (l_x^2 + l_z^2) dm \\
I_{zz} &= \int (l_x^2 + l_y^2) dm \\
\end{aligned}
$$

慣性乗積を非対角項から以下のように定義する。

$$
\begin{aligned}
I_{xy} = I_{yx} &= \int (l_x l_y) dm \\
I_{yz} = I_{zy} &= \int (l_y l_z) dm \\
I_{zx} = I_{xz} &= \int (l_x l_z) dm \\
\end{aligned}
$$

すると角運動量は以下で表すことができ、この右辺の行列を慣性テンソルと呼ぶ。

$$
L = \begin{bmatrix}
I_{xx}  & - I_{xy} & - I_{xz}\\
- I_{xy} &  I_{yy}  & - I_{yz} \\
- I_{xz} & - I_{zy} & I_{zz} 
\end{bmatrix}
\begin{bmatrix} \omega_x \\ \omega_y \\ \omega_z \end{bmatrix}
$$

この式は物体の重心周りの角運動量であるため、角運動量と慣性テンソルを重心周りであることを示すため$G$を付け、以下のように記述する。

$$
L_G = I_G \ \omega
$$



### 1.3. 慣性テンソルの座標変換

剛体の重心$G$周りの角運動量を以下の式で示した。

$$
L_G = I_G \omega
$$

ここで、

- $L_G$ : 重心周りの角運動量
- $I_G$ : 重心周りの慣性テンソル
- $\omega$ : 剛体の角速度ベクトル

この式はすべての量が同じ座標系で表現されていれば成立する。

- 世界座標系 $\Sigma_0$ で示すと以下となる

$$
{^0}L_G = {^0}I_G {^0}\omega
$$

- 物体座標系 $\Sigma_1$ で示すと以下となる

$$
{^1}L_G = {^1}I_G {^1}\omega
$$

ベクトルの座標変換は以下の変換式である。

$$
{^0}v = {^0}R{_1} {^1}v
$$

角速度ベクトル$\omega$と角運動量$L$はベクトルであるため以下の関係式で示せる。

$$
\begin{aligned}
{^0}L_G = {^0}R{_1} {^1}L_G , \quad {^0} \omega = {^0}R{_1} {^1}\omega \space \rightarrow {^1} \omega = \left({^0}R{_1}\right)^T {^0}\omega 
\end{aligned}
$$

これより、世界座標系で示した角運動量 ${^0}L_G$は、${^1}L_G$と回転行列を用いて次のようになる。 

$$
\begin{aligned}
{^0}L_G &= {^0}R{_1} {^1}L_G \\
&= {^0}R{_1} {^1}I_G {^1}\omega \\
&= {^0}R{_1} {^1}I_G \left({^0}R{_1}\right)^T {^0}\omega =  {^0}I_G {^0}\omega
\end{aligned}
$$

これより、物体座標系の重心周りの慣性テンソル ${^1}I_G$ は、回転行列 ${^0}R{_1}$ を用いて、世界座標では以下のように示すことが出来る。

$$
{^0}I_G = {^0}R{_1} {^1}I_G \left({^0}R{_1}\right)^T
$$

この式の意味は、変換した過程を逆に進むと

- ${^1}\omega = \left({^0}R{_1}\right)^T {^0}\omega$ : 世界座標系の角速度を物体座標系へ変換する。
- ${^1}L_G = {^1}I_G {^1}\omega$ : 物体座標系で角運動量を計算する。
- ${^0}R{_1} {^1}L_G$ : 物体座標系の角運動量を世界座標系へ変換する。

## 2. オイラーの回転運動方程式

運動量の時間微分が力になるように、角運動量$L_G$の時間微分もモーメント$N_G$となる。

$$
{^0} N_G = {^0}\dot{L}_G
$$

ここで角運動量を物体座標系$\Sigma_1$で表すと以下である。

$$
{^0}L_G = {^0}R_1 {^1}L_G
$$

これを時間微分し、物体座標系で表した回転運動方程式を導出する。

$$
{^0}\dot{L}_G = {^0}\dot{R}_1 {^1}L_G + {^0}R_1 {^1}\dot{L}_G
$$

回転行列の時間微分は以下である。これを上式に代入すると、

$$
{^0}\dot{R}_1 = {^0}R{_1} \left[{^1}\omega \right]^\land
$$

$$
{^0}\dot{L}_G = {^0}R{_1} \left[{^1}\omega \right]^\land {^1}L_G + {^0}R_1 {^1}\dot{L}_G = {^0}R{_1}\left( \left[{^1}\omega \right]^\land {^1}L_G + {^1}\dot{L}_G \right)
$$

物体座標系で示した角運動量は以下である。

$$
{^1}L_G = {^1}I_G {^1} \ \omega
$$

これを時間微分すると、積の微分より、

$$
{^1}\dot{L}_G = {^1}\dot{I}_G {^1}\omega + {^1}I_G {^1}\dot{\omega}
$$

剛体の質量分布が変化しなければ、以下となる。

$$
{^1}\dot{I}_G = 0
$$

よって、以下となる。

$$
{^0}\dot{L}_G ={^0}R{_1}\left( \left[{^1}\omega \right]^\land {^1}I_G {^1} \ \omega + {^1}I_G {^1}\dot{\omega} \right)
$$

角運動量の時間微分${^0}\dot{L}_G$がモーメント${^0}N_G$である。

$$
{^0}N_G ={^0}R{_1}\left( \left[{^1}\omega \right]^\land {^1}I_G {^1} \ \omega + {^1}I_G {^1}\dot{\omega} \right)
$$

これを物体座標系で表すため、両辺に${^1}R_0$を掛けると、回転の運動方程式が導出される。

$$
{^1}N_G ={^1}I_G \ {^1}\dot{\omega} + \left[{^1}\omega \right]^\land {^1}I_G \ {^1}\omega
$$

世界座標系における回転の運動方程式は、同様にすべての項を世界座標で表して、以下となる。

$$
{^0}N_G ={^0}I_G \ {^0}\dot{\omega} + \left[{^0}\omega \right]^\land {^0}I_G \ {^0}\omega
$$
