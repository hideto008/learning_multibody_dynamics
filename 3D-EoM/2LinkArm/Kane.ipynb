$s_1, s_2$は回転軸とともに、回転方向も示しており、このベクトルの右ねじの方向を正回転とする。


## 1. 設定

初期状態では図のようにアームは世界座標系 $\Sigma_0$ の$z_0$ 方向を向いている。

この状態で、それぞれを次のように定義する。

#### 座標系

Link1 座標 $\Sigma_1$ : 座標原点は$\Sigma_0$の原点$o$と一致し、基底ベクトルは$x_0, y_0, z_0$と同じ方向を向いている。

Link2 座標 $\Sigma_2$ : 座標原点は、Link1先端と一致し、基底ベクトルは$x_0, y_0, z_0$と同じ方向を向いている。

#### リンク長

$$
{^1}\boldsymbol{l}_1 = \begin{bmatrix} 0 \\ 0 \\ l_1 \end{bmatrix}, \quad {^2}\boldsymbol{l}_2 = \begin{bmatrix} 0 \\ 0 \\ l_2 \end{bmatrix}
$$

#### リンク重心

各Linkの重心はLinkの中心にあるとする。

$$
{^1}\boldsymbol{r}_{1G} = \begin{bmatrix} 0 \\ 0 \\ r_{1G} \end{bmatrix}, \quad {^2}\boldsymbol{r}_{2G} = \begin{bmatrix} 0 \\ 0 \\ r_{2G} \end{bmatrix}
$$

#### 回転軸

単位ベクトル $\boldsymbol{s}$ で回転軸と回転方向を示している。


$$
{^1}\boldsymbol{s}_1 = \begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix} , \quad {^2}\boldsymbol{s}_2 = \begin{bmatrix} 0 \\ 1 \\ 0 \end{bmatrix} 
$$

#### 回転行列

今回のシステムで用いる回転行列は $\boldsymbol{s}_1$ に係る $z$ 軸回転と、 $\boldsymbol{s}_2$ に係る $y$ 軸回転である。それぞれの回転角度は$\theta_1, \theta_2$として、それぞれ次の回転行列である。表記の簡略化のため、以降は$C_\theta = \cos(\theta), S_\theta = \sin(\theta)$ とする。


$$
E_z(\theta_1) = 
\begin{bmatrix}
C_{\theta_1} & - S_{\theta_1} & 0 \\
S_{\theta_1} & C_{\theta_1} & 0 \\
0 & 0 & 1
\end{bmatrix} , \quad
E_y(\theta_2) = 
\begin{bmatrix}
C_{\theta_2} & 0 & S_{\theta_2} \\
0 & 1 & 0 \\
- S_{\theta_2} & 0 & C_{\theta_2}\\
\end{bmatrix}
$$

$E_z(\theta_1)$は $\Sigma_0$ と $\Sigma_1$ の姿勢変換に係る回転行列であるため、

$$
{^0}R_1 = E_z(\theta_1)
$$

$E_y(\theta_2)$は $\Sigma_1$ と $\Sigma_2$ の姿勢変換に係る回転行列であるため、

$$
{^1}R_2 = E_y(\theta_2)
$$

となる。


## 2. 各Linkの関係

$s_1, s_2$により回転を行うと、システムは全体の姿勢が変化する。この時の重心位置と回転姿勢を求める。

### 2.1. 重心位置

<p align="center">
<img src="./images/rotation-order.png" style="width:40%;" />
</p>

まず先端であるLink2のみがあり、この状態で$s_2$の回転を行い、その後、Link1が下から突き出てLink2を押し上げたのちに$s_1$でLink1とLink2を回転させると考える。すると、まず、アーム先端の位置$P$は次のようになる。

$$
\begin{aligned}
{^0}P &= {^0}R_1 \left({^1}\boldsymbol{l}_1 +  {^1}R_2 {^2}\boldsymbol{l}_2 \right) \\
&= {^0}R_1  {^1}\boldsymbol{l}_1 + {^0}R_1 {^1}R_2  {^2}\boldsymbol{l}_2 \\
&= {^0}\boldsymbol{l}_1 +  {^0}\boldsymbol{l}_2
\end{aligned}
$$

${^0}R_2$ は具体的には以下の回転行列である。

$$
{^0}R_2 = \begin{bmatrix}
C_{\theta_1} & - S_{\theta_1} & 0 \\
S_{\theta_1} & C_{\theta_1} & 0 \\
0 & 0 & 1
\end{bmatrix}
\begin{bmatrix}
C_{\theta_2} & 0 & S_{\theta_2} \\
0 & 1 & 0 \\
- S_{\theta_2} & 0 & C_{\theta_2}\\
\end{bmatrix} = 
\begin{bmatrix}
C_{\theta_1}C_{\theta_2} & - S_{\theta_1} & C_{\theta_1}S_{\theta_2} \\
S_{\theta_1}C_{\theta_2} & C_{\theta_1} & S_{\theta_1}S_{\theta_2} \\
-S_{\theta_2} & 0 & C_{\theta_2}
\end{bmatrix}
$$

各Linkの重心位置は上式の各項を用いて次のように書ける。

$$
{^0}\boldsymbol{r}_{1G} = {^0}R_1 {^1}\boldsymbol{r}_{1G} = \begin{bmatrix} 0 \\ 0 \\ r_{1G} \end{bmatrix}
$$

$$
{^0}\boldsymbol{r}_{2G} = {^0}\boldsymbol{l}_1 + {^0}R_2 {^2}\boldsymbol{r}_{2G} =
\begin{bmatrix} 0 \\ 0 \\ l_{1} \end{bmatrix} + \begin{bmatrix} C_{\theta_1}S_{\theta_2}  \\ S_{\theta_1}S_{\theta_2}  \\ C_{\theta_2} \end{bmatrix} r_{2G}
$$


### 2.2 回転姿勢

回転姿勢は2.1節で示した${^0}R_1, {^0}R_2$である。

- Link1の回転姿勢 : ${^0}R_1$
- Link2の回転姿勢 : ${^0}R_2$

回転ベクトルの単位ベクトル $\boldsymbol{s}$ は以下となる。

$$
\begin{aligned}
{^0}\boldsymbol{s}_1 &= {^0}R_1 {^1}\boldsymbol{s}_1 = \begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix} \\
{^0}\boldsymbol{s}_2 &= {^0}R_2 {^2}\boldsymbol{s}_2 = \begin{bmatrix} -S_{\theta_1} \\ C_{\theta_1} \\ 0 \end{bmatrix} \\
\end{aligned}
$$


### 3. 一般化座標 $q$

$\theta_1, \theta_2$ を決めるとシステムを構成するすべての物体の位置と姿勢が決まる。よって一般化座標を次のように決定する。

$$
\boldsymbol{q} = \begin{bmatrix}
q_1 \\ q_2
\end{bmatrix} = \begin{bmatrix}
\theta_1 \\ \theta_2
\end{bmatrix}
$$

### 4. 一般化速度 $\boldsymbol{u}$

システムの運動を表す互いに独立な変数として一般化速度を定義する。系の運動を簡単に表すことができるように選択する。

今回は、一般化速度を一般化座標の時間微分として次のように選択する。

$$
\boldsymbol{u} = \begin{bmatrix}
u_1 \\ u_2 
\end{bmatrix} =
\begin{bmatrix}
\dot{\theta}_1 \\ \dot{\theta}_2 
\end{bmatrix} 
$$

一般化座標と一般化速度を結ぶ運動学的微分方程式は以下となる。

$$
\dot{q}_1 = u_1, \quad \dot{q}_2 = u_2
$$

### 5. 各物体の重心の並進速度と角速度ベクトル

それぞれのLinkの重心の並進速度と角速度ベクトルを$\Sigma_0$ 座標で表す。

#### 5.1. Link1

Link1の重心の位置は以下である。

$$
{^0}\boldsymbol{r}_{1G} = {^0}R_1 {^1}\boldsymbol{r}_{1G}
$$

これを時間微分する。${^1}\boldsymbol{r}_{1G}$は時不変であるため、以下となる。

$$
{^0}\dot{\boldsymbol{r}}_{1G} = {^0}\dot{R}_1 {^1}\boldsymbol{r}_{1G}
$$

${^0}\dot{R}_1$ は ${^0}\omega_1$ を用いて、以下となる。

$$
{^0}\dot{R}_1 = [{^0}\omega_1]^\land \ {^0}R_1 
$$

よって、重心位置の並進速度は以下となる。

$$
{^0}\dot{\boldsymbol{r}}_{1G} = [{^0}\omega_1]^\land \ {^0}R_1 {^1}\boldsymbol{r}_{1G}
$$

Link1の角速度ベクトル ${^0}\omega_1$ は ${^0}\boldsymbol{s}_1$を回転軸とした角速度 $\dot{q}_1$ となる。

$$
\boxed{
{^0}\omega_1 = \begin{bmatrix} 0 \\ 0 \\ \dot{q}_1 \end{bmatrix}
}
$$

よって、Link1の重心の並進速度は以下となる。

$$
\boxed{
{^0}{\boldsymbol{v}}_{1G} = [{^0}\omega_1]^\land \ {^0}\boldsymbol{r}_{1G} = \begin{bmatrix} 0 \\ 0 \\ \dot{q}_1 \end{bmatrix} \times \begin{bmatrix} 0 \\ 0 \\ r_{1G} \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}
}
$$



#### 5.2. Link2

Link2の重心位置は以下である。

$$
{^0}\boldsymbol{r}_{2G} = {^0}\boldsymbol{l}_1 + {^0}R_2 {^2}\boldsymbol{r}_{2G} = {^0}R_1{^1}\boldsymbol{l}_1 + {^0}R_2 {^2}\boldsymbol{r}_{2G}
$$

時間微分を行うと、${^1}\boldsymbol{l}_1$と${^2}\boldsymbol{r}_{2G}$は時不変であるため以下となる。

$$
{^0}\dot{\boldsymbol{r}}_{2G} = {^0}\dot{R}_1  {^1}\boldsymbol{l}_1 + {^0}\dot{R}_2 {^2}\boldsymbol{r}_{2G}
$$

回転行列の時間微分は次となるため、

$$
\begin{aligned}
{^0}\dot{R}_2 &= [{^0}\omega_2]^\land \ {^0}R_2 \\
\end{aligned}
$$

${^0}{R}_2 = {^0}{R}_1 {^1}{R}_2$ であるため、上式は以下となる。

$$
\begin{aligned}
{^0}\dot{R}_2 &= {^0}\dot{R}_1 {^1}{R}_2 + {^0}{R}_1 {^1}\dot{R}_2 \\
\end{aligned}
$$

回転行列の時間微分はそれぞれ以下である。${^2}\omega_{2/1}$は$\Sigma_1$との$\Sigma_2$の相対角速度である。

$$
{^0}\dot{R}_1 = [{^0}\omega_1]^\land \ {^0}R_1 , \quad {^1}\dot{R}_2 = {^1}R_{2} \ [{^2}\omega_{2/1}]^\land 
$$ 

これより、時間微分は以下となる。

$$
{^0}\dot{R}_2 =  [{^0}\omega_1]^\land {^0}R_1 {^1}{R}_2 + {^0}{R}_1 {^1}R_{2} \ [{^2}\omega_{2/1}]^\land
$$

これを代入する。

$$
\begin{aligned}
{^0}\dot{\boldsymbol{r}}_{2G} &= [{^0}\omega_1]^\land \ {^0}R_1 {^1}\boldsymbol{l}_1 + \left( [{^0}\omega_1]^\land {^0}R_1 {^1}{R}_2 + {^0}{R}_1 {^1}R_{2} \ [{^2}\omega_{2/1}]^\land\right) {^2}\boldsymbol{r}_{2G} \\
&= [{^0}\omega_1]^\land \ {^0}\boldsymbol{l}_1 + \left( [{^0}\omega_1]^\land {^0}R_2 + {^0}{R}_2 \ [{^2}\omega_{2/1}]^\land\right) {^2}\boldsymbol{r}_{2G} 
\end{aligned}
$$

歪対称行列は $R \omega^\land = (R \omega)^\land R$ となるため、

$$
\begin{aligned}
{^0}\dot{\boldsymbol{r}}_{2G} &= [{^0}\omega_1]^\land \ {^0}\boldsymbol{l}_1 + \left( [{^0}\omega_1]^\land {^0}R_2 + [{^0}{R}_2 \ {^2}\omega_{2/1}]^\land {^0}{R}_2 \right) {^2}\boldsymbol{r}_{2G} \\
&= [{^0}\omega_1]^\land \ {^0}\boldsymbol{l}_1 + \left( [{^0}\omega_1]^\land + [{^0}{R}_2 \ {^2}\omega_{2/1}]^\land  \right) {^0}{R}_2 {^2}\boldsymbol{r}_{2G} \\
&= [{^0}\omega_1]^\land \ {^0}\boldsymbol{l}_1 + \left[{^0}\omega_1 + {^0}{R}_2 \ {^2}\omega_{2/1} \right]^\land {^0}{R}_2 {^2}\boldsymbol{r}_{2G} \\
\end{aligned}
$$

${^0}\boldsymbol{r}_{2G}$より以下の関係が導ける。

$$
{^0}\boldsymbol{r}_{2G} = {^0}\boldsymbol{l}_1 + {^0}R_2 {^2}\boldsymbol{r}_{2G} \rightarrow {^0}R_2 {^2}\boldsymbol{r}_{2G} =  {^0}\boldsymbol{r}_{2G} - {^0}\boldsymbol{l}_1
$$

これを代入すると、Link2重心の並進速度が求まる。

$$
{^0}{\boldsymbol{v}}_{2G} = [{^0}\omega_1]^\land \ {^0}\boldsymbol{l}_1 +  \left[{^0}\omega_1 + {^0}{R}_2 \ {^2}\omega_{2/1} \right]^\land \left( {^0}\boldsymbol{r}_{2G} - {^0}\boldsymbol{l}_1 \right)
$$

Link2の角速度ベクトルは回転行列${^0}R_2$が与えられているため、その時間微分から次のように求めることが出来る。

$$
\begin{aligned}
{^0}\dot{R}_2 = \left[{^0}\omega_1 + {^0}{R}_2 \ {^2}\omega_{2/1} \right]^\land {^0}{R}_2 \quad \rightarrow  \left[{^0}\omega_1 + {^0}{R}_2 \ {^2}\omega_{2/1} \right]^\land = {^0}\dot{R}_2 {^0}{R}_2^T = \left[{^0}\omega_2\right]^\land
\end{aligned}
$$

よって、Link2の角速度ベクトルは以下で求まる。

$$
{^0}\omega_2 = \text{vee}(\left[{^0}\omega_1 + {^0}{R}_2 \ {^2}\omega_{2/1} \right]^\land) = {^0}\omega_1 + {^0}{R}_2 \ {^2}\omega_{2/1}
$$

Link2の相対角速度ベクトル ${^2}\omega_{2/1}$ は ${^2}\boldsymbol{s}_2$を回転軸とした角速度 $\dot{q}_2$ である。

$$
{^2}\omega_{2/1} = \begin{bmatrix} 0 \\ 1 \\ 0 \end{bmatrix} \dot{q}_2
$$

これより、${^0}{R}_2 \ {^2}\omega_{2/1}$ は

$$
{^0}{R}_2 \ {^2}\omega_{2/1} = \begin{bmatrix} -S_{q_1} \\ C_{q_1} \\ 0 \end{bmatrix}\dot{q}_2
$$

よってLink2の角速度ベクトルは以下となる。

$$
\boxed{
{^0}\omega_2 = \begin{bmatrix} 0 \\ 0 \\ \dot{q}_1 \end{bmatrix} + \begin{bmatrix} -S_{q_1} \\ C_{q_1} \\ 0 \end{bmatrix}\dot{q}_2 = \begin{bmatrix} -S_{q_1}\dot{q}_2 \\ C_{q_1}\dot{q}_2 \\ \dot{q}_1 \end{bmatrix}
}
$$


${^0}{\boldsymbol{v}}_{2G}$を具体的に計算すると以下となる。

$$
\begin{aligned}
{^0}{\boldsymbol{v}}_{2G} &= [{^0}\omega_1]^\land \ {^0}\boldsymbol{l}_1 +  \left[{^0}\omega_1 + {^0}{R}_2 \ {^2}\omega_{2/1} \right]^\land \left( {^0}\boldsymbol{r}_{2G} - {^0}\boldsymbol{l}_1 \right) \\
&= {^0}\omega_1 \times {^0}\boldsymbol{l}_1 + {^0}\omega_2 \times \left( {^0}\boldsymbol{r}_{2G} - {^0}\boldsymbol{l}_1 \right)
\end{aligned}
$$

各項を計算すると、

$$
{^0}\omega_1 \times {^0}\boldsymbol{l}_1 = \begin{bmatrix} 0 \\ 0 \\ \dot{q}_1 \end{bmatrix} \times \begin{bmatrix} 0 \\ 0 \\ l_1 \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}
$$ 

$$
\begin{aligned}
{^0}\omega_2 \times \left( {^0}\boldsymbol{r}_{2G} - {^0}\boldsymbol{l}_1 \right) &= \begin{bmatrix} -S_{q_1}\dot{q}_2 \\ C_{q_1}\dot{q}_2 \\ \dot{q}_1 \end{bmatrix} \times \left( \begin{bmatrix} 0 \\ 0 \\ l_{1} \end{bmatrix} + \begin{bmatrix} C_{q_1}S_{q_2}  \\ S_{q_1}S_{q_2}  \\ C_{q_2} \end{bmatrix} r_{2G} - \begin{bmatrix} 0 \\ 0 \\ l_1 \end{bmatrix} \right) \\
&= \begin{bmatrix} -S_{q_1}\dot{q}_2 \\ C_{q_1}\dot{q}_2 \\ \dot{q}_1 \end{bmatrix} \times \begin{bmatrix} C_{q_1}S_{q_2}  \\ S_{q_1}S_{q_2}  \\ C_{q_2} \end{bmatrix} r_{2G} \\
&= r_{2G} \begin{bmatrix} C_{q_1}C_{q_2} \\ S_{q_1}C_{q_2} \\ -S_{q_2} \end{bmatrix} \dot{q}_2 + r_{2G}\begin{bmatrix} -S_{q_1} S_{q_2} \\ C_{q_1}S_{q_2} \\ 0 \end{bmatrix} \dot{q}_1
\end{aligned}
$$

よって、${^0}{\boldsymbol{v}}_{2G}$は以下となる。

$$
\boxed{
{^0}{\boldsymbol{v}}_{2G} = r_{2G} \begin{bmatrix} C_{q_1}C_{q_2} \\ S_{q_1}C_{q_2} \\ -S_{q_2} \end{bmatrix} \dot{q}_2 + r_{2G}\begin{bmatrix} -S_{q_1} S_{q_2} \\ C_{q_1}S_{q_2} \\ 0 \end{bmatrix} \dot{q}_1
}
$$

### 6. 偏速度(partial velocity)と偏角速度(partial angular velocity)

#### 6.1. Link1

$$
\begin{aligned}
{^0}\boldsymbol{v}_1^{(1)} &= \frac{\partial {^0}{\boldsymbol{v}}_{1G}}{\partial u_1} = \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix} \\
{^0}\boldsymbol{v}_1^{(2)} &= \frac{\partial {^0}{\boldsymbol{v}}_{1G}}{\partial u_2} = \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix} \\
{^0}\omega_1^{(1)} &= \frac{\partial {^0}\omega_1}{\partial u_1} = \begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix} \\
{^0}\omega_1^{(2)} &= \frac{\partial {^0}\omega_1}{\partial u_2} = \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix} \\
\end{aligned}
$$

#### 6.2 Link2

$$
\begin{aligned}
{^0}\boldsymbol{v}_2^{(1)} &= \frac{\partial {^0}{\boldsymbol{v}}_{2G}}{\partial u_1} = \begin{bmatrix} -S_{q_1} S_{q_2} \\ C_{q_1}S_{q_2} \\ 0 \end{bmatrix}  r_{2G} \\
{^0}\boldsymbol{v}_2^{(2)} &= \frac{\partial {^0}{\boldsymbol{v}}_{2G}}{\partial u_2} = \begin{bmatrix} C_{q_1}C_{q_2}  \\ S_{q_1} C_{q_2}  \\ -S_{q_2} \end{bmatrix} r_{2G}\\
{^0}\omega_2^{(1)} &= \frac{\partial {^0}\omega_2}{\partial u_1} = \begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix} \\
{^0}\omega_2^{(2)} &= \frac{\partial {^0}\omega_2}{\partial u_2} = \begin{bmatrix} -S_{q_1} \\ C_{q_1} \\ 0 \end{bmatrix} \\
\end{aligned}
$$


### 7. 一般化作用力 ${F}_r$

Linkに作用する力とトルクを、一般化速度$u_r$に対応する偏速度、偏角速度方向へ射影し、その一般化運動方向への寄与を求める。

以下の式で$F_r$を求める。

$$
F_r = \sum_i \left( {^0}{\boldsymbol{v}_i^{(r)}}^T {^0}\boldsymbol{F}_i  + {^0}{\boldsymbol{\omega}_i^{(r)}}^T {^0}\boldsymbol{T}_i \right)
$$

${^0}\boldsymbol{F}_i$は各Linkに作用する力、${^0}\boldsymbol{T}_i$ は各Linkの重心まわりに作用するトルクを世界座標$(x_0, y_0, z_0)$で表したものである。

今回は、制御入力、および重力による力を一般化作用力の計算に用いる。

#### 作用する力

Link1, Link2の重心には重力が生じる。Link1の質量を$m_1$、Link2の質量を$m_2$とすると、それぞれ$z$方向に重力が生じる。

Link1と世界座標をつなぐJoint $J_0$ 、Link1とLink2をつなぐJoint $J_1$ではそれぞれ作用・反作用の力を受けている。

$J_0$は動かない点であるため、偏速度はゼロとなる。よって、作用・反作用の力は一般化作用力には寄与しない。

$J_1$は同じ偏速度を持つ点であるため、この点における一般化作用力への寄与は以下のように相殺される。

$$
F_{r,{J_1}} = {\boldsymbol{v}_{J_1}^{(r)}}^T \boldsymbol{F}_{J_1}^{(1)} + {\boldsymbol{v}_{J_1}^{(r)}}^T \boldsymbol{F}_{J_1}^{(2)} = 0 , \quad \because \boldsymbol{F}_{J_1}^{(1)}=- \boldsymbol{F}_{J_1}^{(2)} 
$$

よって、作用力は以下となる。

$$
{^0}\boldsymbol{F}_1 = \begin{bmatrix} 0 \\ 0 \\ -m_1 g \end{bmatrix} , \quad {^0}\boldsymbol{F}_2 = \begin{bmatrix} 0 \\ 0 \\ -m_2 g \end{bmatrix}
$$ 

#### 作用するトルク

Link1とLink2はそれぞれ各Jointでモーターによるトルクが印加されるとする。Link1に対して$\tau_1$、Link2に対して$\tau_2$とする。

$\tau_1$は${^0}\boldsymbol{s}_1$軸を回転中心として、固定されたベースとLink1の間に作用し、Link1を回転させる。Link1の回転するとLink2も回転するが、これは拘束により動かされているのであり、$\tau_1$がLink2へ直接作用しているのでない。

$\tau_2$は${^0}\boldsymbol{s}_2$軸を回転中心としてLink1とLink2の間に作用する。Link2には作用モーメント、Link1には反作用モーメントが生じる。

これより各Linkへの作用するトルクは以下となる。

$$
{^0}\boldsymbol{T}_1 = {^0}\boldsymbol{s}_1 \tau_1 - {^0}\boldsymbol{s}_2 \tau_2 = \begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix} \tau_1 - \begin{bmatrix} -S_{q_1} \\ C_{q_1} \\ 0 \end{bmatrix} \tau_2
$$

$$
{^0}\boldsymbol{T}_2 = {^0}\boldsymbol{s}_2 \tau_2 = \begin{bmatrix} -S_{q_1} \\ C_{q_1} \\ 0 \end{bmatrix} \tau_2
$$



#### 一般作用力 $F_r$

##### $r=1$

$$
F_1 =  {{^0}\boldsymbol{v}_1^{(1)}}^T {^0}\boldsymbol{F}_1 + {{^0}\boldsymbol{v}_2^{(1)}}^T {^0}\boldsymbol{F}_2 + {{^0}\boldsymbol{\omega}_1^{(1)}}^T {^0}\boldsymbol{T}_1 + {{^0}\boldsymbol{\omega}_2^{(1)}}^T {^0}\boldsymbol{T}_2 
$$

各項を計算する。

$$
{{^0}\boldsymbol{v}_1^{(1)}}^T {^0}\boldsymbol{F}_1 = \begin{bmatrix} 0 & 0 & 0 \end{bmatrix} \begin{bmatrix} 0 \\ 0 \\ -m_1 g \end{bmatrix} = 0
$$

$$
{{^0}\boldsymbol{v}_2^{(1)}}^T {^0}\boldsymbol{F}_2 =  r_{2G}\begin{bmatrix} -S_{q1} S_{q_2} & C_{q_1}S_{q_2} & 0 \end{bmatrix} \begin{bmatrix} 0 \\ 0 \\ -m_2 g \end{bmatrix} = 0
$$

$$

### 8. 一般化慣性力 $F_r^*$

並進の慣性力をニュートンの運動方程式により以下でまとめる。

$$
\boldsymbol{F}_i^{*} = - m_i {^0}\dot{\boldsymbol{v}}_{iG}
$$

また回転の慣性トルクをオイラーの回転運動方程式により以下でまとめる。

$$
\boldsymbol{T}_i^{*} = -\left( {^0}\boldsymbol{I}_{iG} {^0}\dot{\boldsymbol{\omega}}_i + {^0}\boldsymbol{\omega}_i \times {^0}\boldsymbol{I}_{iG} {^0}\boldsymbol{\omega}_i \right)
$$

これにより一般化慣性力 $F_r^{*}$ を次のように求める。これは、一般化速度$u_r$に対応する偏速度、偏角速度方向へ慣性力を射影し、その方向への寄与を求めている。

$$
F_r^{*} = \sum_i \left( {{^0}\boldsymbol{v}_i^{(r)}}^T {^0}\boldsymbol{F}_i^{*}  + {{^0}\boldsymbol{\omega}_i^{(r)}}^T {^0}\boldsymbol{T}_i^{*} \right)
$$

${^0}\dot{\boldsymbol{v}}_{iG}$は各物体の重心の並進加速度を世界座標($x_0, y_0, z_0$)で表したものである。<br>
${^0}\boldsymbol{\omega}_i, {^0}\dot{\boldsymbol{\omega}}_i$は物体(Link)$i$の角速度と角加速度を世界座標($x_0, y_0, z_0$)で表したものである。<br>
${^0}\boldsymbol{I}_{iG}$は世界座標系で示したLink $i$ の重心周りの慣性テンソルである。重心周りのオイラーの回転運動方程式をまとめるため、Link $i$ 座標系で示した重心周りの慣性テンソル ${^i}\boldsymbol{I}_{iG}$ の座標変換により次の式で計算を行う。

$$
{^0}\boldsymbol{I}_{iG} = {^0}R_i {^i}\boldsymbol{I}_{iG} \left({^0}R_i \right)^T
$$

S_{q_1} \left( C_{q_2}^2 I_{2zz} + S_{q_2}^2 I_{2xx} - I_{2yy} \right) \\
S_{q_2} C_{q_2} (I_{2xx} - I_{2zz})
\end{bmatrix} \dot{q}_1 \dot{q}_2 +
\begin{bmatrix}
- S_{q_1} S_{q_2} C_{q_2} (I_{2zz} - I_{2xx})\\
C_{q_1} S_{q_2} C_{q_2} (I_{2zz} - I_{2xx})  \\
0
\end{bmatrix} \dot{q}_1^2 


\end{aligned}
$$

これより、$\boldsymbol{T}_2^{*}$は以下となる。

$$

\begin{aligned}
{^0}\boldsymbol{T}_2^{*}
= -\Bigg(
&\begin{bmatrix}

0
\end{bmatrix} \ddot{q}_2

\begin{bmatrix}

\end{bmatrix} \ddot{q}_1
\\
&+
\begin{bmatrix}
C_{q_1}\left[(I_{2zz}-I_{2xx})(C_{q_2}^2-S_{q_2}^2)-I_{2yy}\right]\\
S_{q_1}\left[(I_{2zz}-I_{2xx})(C_{q_2}^2-S_{q_2}^2)-I_{2yy}\right]\\
2S_{q_2}C_{q_2}(I_{2xx}-I_{2zz})
\end{bmatrix}\dot{q}_1\dot{q}_2
+
\begin{bmatrix}
-S_{q_1}S_{q_2}C_{q_2}(I_{2zz}-I_{2xx}) \\
C_{q_1}S_{q_2}C_{q_2}(I_{2zz}-I_{2xx}) \\
0
\end{bmatrix}\dot{q}_1^2
\Bigg)
\end{aligned}
$$

以上で${^0}\boldsymbol{F}_i^{*}$と$ {^0}\boldsymbol{T}_i^{*}$が求めった。これより、$F_r^{*}$の計算を行う。

$$
F_r^{*} = \sum_i \left( {{^0}\boldsymbol{v}_i^{(r)}}^T {^0}\boldsymbol{F}_i^{*}  + {{^0}\boldsymbol{\omega}_i^{(r)}}^T {^0}\boldsymbol{T}_i^{*} \right)
$$

F_1^{*} = {{^0}\boldsymbol{v}_1^{(1)}}^T {^0}\boldsymbol{F}_1^{*}  + {{^0}\boldsymbol{\omega}_1^{(1)}}^T {^0}\boldsymbol{T}_1^{*} + {{^0}\boldsymbol{v}_2^{(1)}}^T {^0}\boldsymbol{F}_2^{*}  + {{^0}\boldsymbol{\omega}_2^{(1)}}^T {^0}\boldsymbol{T}_2^{*}

$$
\begin{aligned}

\end{aligned}
$$

$$
\begin{aligned}
 {{^0}\boldsymbol{\omega}_2^{(1)}}^T {^0}\boldsymbol{T}_2^{*} &= -\begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix}^T \Bigg(
\begin{bmatrix}

\\

## 9. Kaneの方程式 $F_r + F_r^* = 0$

一般化作用力$F_r$と一般化慣性力$F_r^*$ は以下として求めた。

$$
\begin{bmatrix}
F_1 \\ F_2
\end{bmatrix} = 
\begin{bmatrix}
\tau_1 \\
\tau_2 + r_{2G} S_{q_2} m_2 g
\end{bmatrix}
$$

$$
\begin{bmatrix}
F_1^{*} \\ F_2^{*}
\end{bmatrix} =
\begin{bmatrix}
 - (S_{q_2}^2 I_{2xx}+C_{q_2}^2 I_{2zz} + I_{1zz} + m_2r_{2G}^2 S_{q_2}^2) \ddot{q}_1 -2S_{q_2}C_{q_2} \left(m_2r_{2G}^2 + I_{2xx}-I_{2zz} \right)\dot{q}_1\dot{q}_2 \\
 -\left(m_2r_{2G}^2 + I_{2yy}\right) \ddot{q}_2  -S_{q_2}C_{q_2}(I_{2zz}-I_{2xx} - m_2r_{2G}^2)\dot{q}_1^2
\end{bmatrix}
$$

これをKaneの方程式へ代入する。

#### $r=1$

$$
 ( I_{1zz} +  S_{q_2}^2 (I_{2xx} + m_2r_{2G}^2) + C_{q_2}^2 I_{2zz} ) \ddot{q}_1  = \tau_1 -2S_{q_2}C_{q_2} \left(m_2r_{2G}^2 + I_{2xx} - I_{2zz} \right)\dot{q}_1\dot{q}_2
$$

この式は一般化速度 $u_1 = \dot{q}_1$ 方向に対する運動方程式である。

左辺の$\ddot{q}_1$の係数は$q_2 = \theta_2$の姿勢により慣性モーメントの影響を示している。$q_2=0$の場合、慣性モーメントは$(I_{1zz} + I_{2zz})$ となり、Link1とLink2が一直線のLinkとして見えている。$q_2=90$ deg の場合、慣性モーメントは$(I_{1zz} + I_{2xx} + m_2r_{2G}^2)$ となり、Link2の慣性モーメントがJoint1周りの慣性モーメントになると解釈でき、平行軸の定理とも整合する。

左辺の係数を以下でまとめる。

$$
M_{11}(q_2) = I_{1zz} +  S_{q_2}^2 (I_{2xx} + m_2r_{2G}^2) + C_{q_2}^2 I_{2zz} 
$$

これを$q_2$で偏微分すると、以下となり、右辺の$\dot{q}_1 \dot{q}_2$の係数と一致する。

$$
\frac{\partial M_11(q_2)}{\partial q_2 } = 2 S_{q_2} C_{q_2} (I_{2xx} + m_2r_{2G}^2) - 2 C_{q_2} S_{q_2} I_{2zz} = 2 S_{q_2} C_{q_2} (I_{2xx} + m_2r_{2G}^2 - I_{2zz}) 
$$

よって、以下となる。

$$
M_{11}(q_2) \ddot{q}_1 = \tau_1 - \frac{\partial M_{11}(q_2)}{\partial q_2 } \dot{q}_1 \dot{q}_2
$$

つまり、$\dot{q}_1 \dot{q}_2$の項は第二関節$q_2$が動くことにより、第1関節$q_1$まわりの慣性モーメントが時間変化することで発生する慣性項の影響を示している。


実際に、時間微分は次のようになるため、

$$
\frac{M_{11}(q_2)}{dt} = \frac{\partial M_{11(q_2)}}{ \partial q_2} \frac{d \ q_2} {dt}
$$

運動法方程式を以下のように構成すると、左辺は角運動量$M_{11}\dot{q}_1$の時間変化を表している。

$$
M_{11}\ddot{q}_1 + \dot{M}_{11} \dot{q}_1 = \tau_1
$$


#### $r=2$

$$
\left(m_2r_{2G}^2 + I_{2yy}\right) \ddot{q}_2 = \tau_2 + r_{2G} S_{q_2} m_2 g -S_{q_2}C_{q_2}(I_{2zz}-I_{2xx} - m_2r_{2G}^2)\dot{q}_1^2
$$

この式は一般化速度 $u_2 = \dot{q}_2$ 方向に対する運動方程式である。

左辺の$\ddot{q}_2$の係数は慣性モーメントであり、Jointと重心までの距離が平行軸として現れている。

右辺の $\dot{q}_1^2$の項は、Joint1が回転することで、Link2に生じる遠心力に対応する一般化力表す。重心の並進運動に加えて、Link2の慣性テンソルによる影響も含まれる。



運動方程式を以下の形へまとめる。

$$
M(q)\ddot{q} + c(q, \dot{q}) + g(q) = \tau
$$

$$
 ( I_{1zz} +  S_{q_2}^2 (I_{2xx} + m_2r_{2G}^2) + C_{q_2}^2 I_{2zz} ) \ddot{q}_1 + 2S_{q_2}C_{q_2} \left(m_2r_{2G}^2 + I_{2xx} - I_{2zz} \right)\dot{q}_1\dot{q}_2= \tau_1  \\
 \left(m_2r_{2G}^2 + I_{2yy}\right) \ddot{q}_2 - r_{2G} S_{q_2} m_2 g  + S_{q_2}C_{q_2}(I_{2zz}-I_{2xx} - m_2r_{2G}^2)\dot{q}_1^2 = \tau_2 
$$

$$
\begin{bmatrix}
I_{1zz} +  S_{q_2}^2 (I_{2xx} + m_2r_{2G}^2) + C_{q_2}^2 I_{2zz} & 0 \\
0 & m_2r_{2G}^2 + I_{2yy}
\end{bmatrix}
\begin{bmatrix}
\ddot{q}_1 \\ \ddot{q}_2 
\end{bmatrix} +
\begin{bmatrix}
2S_{q_2}C_{q_2} \left(m_2r_{2G}^2 + I_{2xx} - I_{2zz} \right)\dot{q}_1\dot{q}_2 \\
 S_{q_2}C_{q_2}(I_{2zz}-I_{2xx} - m_2r_{2G}^2)\dot{q}_1^2
\end{bmatrix} +
\begin{bmatrix}
0 \\
- r_{2G} S_{q_2} m_2 g 
\end{bmatrix}=
\begin{bmatrix}
\tau_1 \\ \tau_2
\end{bmatrix}
$$


## 10. Pythonによる検算

Sympyを用いて運動方程式を導出し、上記の式が正しいかを確認する。

```python
import sympy as sp
from sympy.physics.mechanics import(
    ReferenceFrame,
    Point,
    RigidBody,
    KanesMethod,
    dynamicsymbols,
    inertia
)

# 一般化座標

q1, q2 = dynamicsymbols("q1 q2")
u1, u2 = dynamicsymbols("u1 u2")

m1, m2 = sp.symbols("m1 m2", positive=True)
l1, r1G, r2G = sp.symbols("l1 r1G r2G", positive=True)
g = sp.symbols("g", positive=True)

I1xx, I1yy, I1zz = sp.symbols("I1xx I1yy I1zz")
I2xx, I2yy, I2zz = sp.symbols("I2xx I2yy I2zz")

tau1, tau2 = sp.symbols("tau1 tau2")

# 座標系
N = ReferenceFrame("N") # Sigma0
A = ReferenceFrame("A") # Sigma1
B = ReferenceFrame("B") # Sigma2

# Joint1 : World Z軸まわり
A.orient_axis(N, N.z, q1)

# Joint2 : Link1 Y軸まわり
B.orient_axis(A, A.y, q2)

#
O = Point("O") # Joint1
J2 = Point("J2") # Joint2
G1 = Point("G1") # Link1 CoM
G2 = Point("G2") # Link2 CoM

O.set_vel(N, 0) # 

# 位置
G1.set_pos(O, r1G * A.z) 
J2.set_pos(O, l1 * A.z)
G2.set_pos(J2, r2G * B.z)

# 速度
G1.v2pt_theory(O, N, A)
J2.v2pt_theory(O, N, A)
G2.v2pt_theory(J2, N , B)

# 運動学的微分方程式の定義
t = dynamicsymbols._t

kd = [
    q1.diff(t) - u1,
    q2.diff(t) - u2
]

# 剛体と慣性テンソルの定義
I1 = inertia(A, I1xx, I1yy, I1zz)
I2 = inertia(B, I2xx, I2yy, I2zz)

body1 = RigidBody("Link1", G1, A, m1, (I1, G1))
body2 = RigidBody("Link2", G2, B, m2, (I2, G2))

bodies = [body1, body2]

# 重力と関節トルクの定義
loads = [
    (G1, -m1 * g * N.z),
    (G2, -m2 * g * N.z),

    # Joint1 torque
    (A, tau1 * N.z),

    # Joint2 torque: Link1とLink2への作用・反作用
    (A, -tau2 * A.y),
    (B,  tau2 * A.y)
]

# KaneMethodで運動方程式を生成

KM = KanesMethod(
    N,
    q_ind=[q1, q2],
    u_ind=[u1, u2],
    kd_eqs=kd
)

KM.kanes_equations(bodies, loads)

M_sym = sp.simplify(KM.mass_matrix)
forcing_sym = sp.trigsimp(KM.forcing)

from IPython.display import display, Math

display(Math(r"M(q) = "))
display(M_sym)

print("Forcing:")
display(Math(r"\mathrm{forcing}(q,u,\tau) ="))
display(forcing_sym)

```

手計算との検算

```python

Acoef = m2*r2G**2 + I2xx - I2zz

M_kane = sp.Matrix([
    [
        I1zz + Acoef*sp.sin(q2)**2 + I2zz,
        0
    ],
    [
        0,
        I2yy + m2*r2G**2
    ]
])

h_kane = sp.Matrix([
    2*Acoef*sp.sin(q2)*sp.cos(q2)*u1*u2,
    -Acoef*sp.sin(q2)*sp.cos(q2)*u1**2
    -m2*g*r2G*sp.sin(q2)
])

tau = sp.Matrix([tau1, tau2])

# SymPy側の h = tau - forcing
h_sym = tau - forcing_sym

M_error = sp.trigsimp(M_sym - M_kane)
h_error = sp.trigsimp(h_sym - h_kane)

assert M_error == sp.zeros(2, 2)
assert h_error == sp.zeros(2, 1)

print("Kane EOM verified!")
```
