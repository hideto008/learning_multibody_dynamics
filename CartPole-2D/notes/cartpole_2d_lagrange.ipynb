- Link1 と Link2は、Link1の重心位置にある回転関節で接続されている。回転は自由で、位置は一致するように拘束するように構成されている。
- Link1の質量は$m_1$
- Link2の質量は$m_2$で、重心まわりの慣性モーメントは$I_2$
- Link1は垂直方向($y_0$)を固定されており、水平方向($x_0$)に動作できる。
- 制御入力 $f$ はLink1に加えられ、Link2には直接的な制御入力や関節トルクを加えない。
- $l_2$は関節点からLink2の重心までの距離
- $\theta_2$は鉛直上向きから時計回りを正とするLink2の角度
- $\mu_1$はLink1の並進粘性係数
- $\mu_2$は関節の回転粘性係数
- $g$は重力加速度

Lagrange法では、運動エネルギー、位置エネルギーおよびRayleighの散逸関数を用いて運動方程式を構成する。

今回の対象とするシステムは次のようなものがある。

|項|並進運動|回転運動|単位|
|:---|:---:|:---:|:---:|
|運動エネルギー $W$| $\frac{1}{2} m \dot{x}^2(t)$ | $\frac{1}{2} I \dot{\theta}^2(t)$|$\mathrm{J}$|
|位置エネルギー $U$| $m g h(t)$ |-|$\mathrm{J}$|
|Rayleighの散逸関数 $D$ | $\frac{1}{2} \mu \dot{x}^2(t)$| $\frac{1}{2} \mu_{\theta} \dot{\theta}^2(t)$|$\mathrm{W}$|

以降は表記を簡潔にするため、$\sin(\theta) = S_\theta, \cos(\theta) = C_\theta$ として表す。

## 運動方程式を構成する項

### 運動エネルギー

運動エネルギーは重心$(x_1,y_1), (X_2, Y_2)$の運動と回転 $(I_2)$ の運動により以下で構成される。
$y_1$方向は拘束されて動作しないため、この項は現れない。

$$
W(t) = \frac{1}{2}m_1 (\dot{x}_1(t))^2 + \frac{1}{2}m_2 (\dot{X}_2(t))^2 + \frac{1}{2}m_2 (\dot{Y}_2(t))^2 + \frac{1}{2}I_2 (\dot{\theta}_2(t))^2 
$$

### 位置エネルギー

位置エネルギーは重心が高さ方向($y_0$)に変化する座標で現れる。今回はLink2の重心高さ $Y_2$ が変化するため以下となる。

$$
U(t) = m_2 g Y_2(t)
$$

## Lagrange方程式で使用する関数

### Rayleighの散逸関数

粘性項を用いて以下で構成される。

$$
D(t) = \frac{1}{2}\mu_1 (\dot{x}_1(t))^2 + \frac{1}{2} \mu_2 (\dot{\theta}_2(t))^2  
$$

ここで、$(X_2, Y_2)$ を $(x_0, y_0)$ 座標で表すと

$$
\begin{aligned}
X_2(t) &= x_1(t) + l_2 S_{\theta_2} \\
Y_2(t) &= l_2 C_{\theta_2}
\end{aligned}
$$

よって、速度は以下となる。

$$
\begin{aligned}
\dot{X}_2(t) &= \dot{x}_1(t) + l_2 C_{\theta_2} \dot{\theta}_2(t)\\
\dot{Y}_2(t) &= - l_2 S_{\theta_2} \dot{\theta}_2(t)
\end{aligned}
$$


これより、運動エネルギー $W(t)$と位置エネルギー$U(t)$は以下のようになる。

$$
\begin{aligned}
W(t) &= \frac{1}{2}m_1 (\dot{x}_1(t))^2 + \frac{1}{2}m_2 \left(\dot{x}_1(t)^2 + l_2^2 {C_{\theta_2}}^2 (\dot{\theta}_2(t))^2 + 2 \dot{x}_1(t) l_2 C_{\theta_2} \dot{\theta}_2(t)  \right)  \\ &+ \frac{1}{2}m_2 \left(l_2^2 {S_{\theta_2}}^2 \dot{\theta_2}(t)^2 \right) + \frac{1}{2}I_2 (\dot{\theta}_2(t))^2 \\
&= \frac{1}{2}\left(m_1+m_2 \right) (\dot{x}_1(t))^2 + \frac{1}{2}m_2 l_2^2 {C_{\theta_2}}^2 (\dot{\theta}_2(t))^2 + m_2 \dot{x}_1(t) l_2 C_{\theta_2} \dot{\theta}_2(t)  
\\ & + \frac{1}{2}m_2 \left(l_2^2 {S_{\theta_2}}^2 \dot{\theta_2}(t)^2 \right) + \frac{1}{2}I_2 (\dot{\theta}_2(t))^2 \\
&= \frac{1}{2}\left(m_1+m_2 \right) (\dot{x}_1(t))^2 + \frac{1}{2}\left( m_2 l_2^2 + I_2 \right) (\dot{\theta}_2(t))^2 + m_2  l_2 C_{\theta_2} \dot{x}_1(t) \dot{\theta}_2(t) \\
U(t) &= m_2 g {l_2} {C_{\theta_2}}
\end{aligned}
$$

## Lagragian $L$ と一般化座標

$$
\begin{aligned}
L(t) &= W(t) - U(t) \\
&= \frac{1}{2}\left(m_1+m_2 \right) (\dot{x}_1(t))^2 + \frac{1}{2}\left( m_2 l_2^2 + I_2 \right) (\dot{\theta}_2(t))^2 + m_2  l_2 C_{\theta_2} \dot{x}_1(t) \dot{\theta}_2(t) - m_2 g {l_2} {C_{\theta_2}}
\end{aligned}
$$

一般化座標は系の姿勢を一意に決めるための独立な変数として選ぶ。
今回はLink1の$x_1$方向と、Link2の$\theta_2$角度が決まれば Cart Poleの姿勢が決まる。
よって、一般化座標 $\boldsymbol{q}$ を以下のように設定する。

$$
\boldsymbol{q} = [q_1, q_2]^T = [x_1(t), \theta_2(t)]^T
$$

一般化力は一般化座標 $q_j$ の微小変化に対して仕事をする、実在の力・トルクの合成力として選ぶ。
よって、$\boldsymbol{q}$ に対応する一般化力 $\boldsymbol{Q}$ を以下のように設定する。

$$
\boldsymbol{Q} = [Q_1, Q_2]^T = [f(t), 0]^T
$$

## 運動方程式の導出

Lagrangian $L$ , Rayleighの散逸関数 $D$, 一般化座標 $\boldsymbol{q}$, 一般化力 $\boldsymbol{Q}$ を用いて 運動方程式を次のように導出する。

$$
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i} \right) - \frac{\partial L}{\partial q_i} + \frac{\partial D}{\partial \dot{q}_i} = Q_i , \quad (i=1,2)
$$

### $i = 1$

$$
\begin{aligned}
\frac{\partial L}{\partial \dot{q}_1} &= \frac{\partial L}{\partial \dot{x}_1} = (m_1 + m_2) \dot{x}_1(t) + m_2 l_2 C_{\theta_2} \dot{\theta}_2(t) \\
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_1} \right) &= \frac{d}{dt}\left(\frac{\partial L}{\partial \dot{x}_1} \right) = (m_1 + m_2) \ddot{x}_1(t) - m_2 l_2 S_{\theta_2} {\dot{\theta}_2(t)}^2 + m_2 l_2 C_{\theta_2} \ddot{\theta}_2(t) \\
\frac{\partial L}{\partial q_1} &= \frac{\partial L}{\partial x_1} = 0 \\
\frac{\partial D}{\partial \dot{q}_1} &= \frac{\partial D}{\partial \dot{x}_1} = \mu_1 \dot{x}_1(t) \\
Q_1 &= f(t)
\end{aligned}
$$

よって運動方程式は以下となる。

$$
(m_1 + m_2) \ddot{x}_1(t) + m_2 l_2 C_{\theta_2} \ddot{\theta}_2(t) = f(t)- \mu_1 \dot{x}_1(t) + m_2 l_2 S_{\theta_2} {\dot{\theta}_2(t)}^2 
$$

### $i=2$

$$
\begin{aligned}
\frac{\partial L}{\partial \dot{q}_2} &= \frac{\partial L}{\partial \dot{\theta}_2} =  (m_2 {l_2}^2 + I_2) \dot{\theta}_2(t) + m_2 l_2 C_{\theta_2} \dot{x}_1(t) \\
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_2} \right) &= \frac{d}{dt}\left(\frac{\partial L}{\partial \dot{\theta}_2} \right) = (m_2 {l_2}^2 + I_2) \ddot{\theta}_2(t) - m_2 l_2 S_{\theta_2} \dot{\theta}_2 (t) \dot{x}_1(t) + m_2 l_2 C_{\theta_2} \ddot{x}_1(t)\\
\frac{\partial L}{\partial q_2} &= \frac{\partial L}{\partial \theta_2} = - m_2 l_2 S_{\theta_2} \dot{x}_1(t) \dot{\theta}_2(t) + m_2 g l_2 S_{\theta_2}  \\
\frac{\partial D}{\partial \dot{q}_2} &= \frac{\partial D}{\partial \dot{\theta}_2} = \mu_2 \dot{\theta}_2 (t)  \\
Q_2 &= 0
\end{aligned}
$$

よって運動方程式は以下となる。

$$
m_2 l_2 C_{\theta_2} \ddot{x}_1(t) + (I_2 + m_2 {l_2}^2) \ddot{\theta}_2(t) =  m_2 g l_2 S_{\theta_2} - \mu_2 \dot{\theta}_2 (t) 
$$

行列形式でまとめると以下となる。

$$
\begin{bmatrix}
m_1+m_2 & m_2 l_2 C_{\theta_2} \\
m_2 l_2 C_{\theta_2} & I_2 + m_2 {l_2}^2
\end{bmatrix}
\begin{bmatrix}
\ddot{x}_1(t) \\ \ddot{\theta}_2(t)
\end{bmatrix} =
\begin{bmatrix}
f(t)- \mu_1 \dot{x}_1(t) + m_2 l_2 S_{\theta_2} {\dot{\theta}_2(t)}^2 \\
 m_2 g l_2 S_{\theta_2} - \mu_2 \dot{\theta}_2 (t) 
\end{bmatrix}
$$
