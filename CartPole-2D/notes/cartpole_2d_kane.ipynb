# Cart Pole 2D - Kane法

- Link1 と Link2は、Link1の重心位置にある回転関節で接続されている。回転は自由で、位置は一致するように拘束するように構成されている。
- Link1の質量は$m_1$
- Link2の質量は$m_2$で、重心まわりの慣性モーメントは$I_2$
- Link1は垂直方向($y_0$)を固定されており、水平方向($x_0$)に動作できる。
- 制御入力 $f$ はLink1に加えられ、Link2には直接的な制御入力や関節トルクを加えない。
- $l_2$は関節点からLink2の重心までの距離
- $\theta_2$は鉛直上向きから時計回りを正とするLink2の角度
- $\mu_1$はLink1の並進粘性係数
- $\mu_2$は関節の回転粘性係数
- $g$は重力加速度

Kane法は各物体に働く力と慣性力を一般化速度に対応する運動方向に射影して運動方程式を導出する。

おおむね次の順番で進める。

1. 一般化座標の決定
1. 一般化速度の決定
1. 各物体の重心の並進速度と角速度を求める
1. 偏速度(partial velocity)、偏角速度(partial angular velocity)を求める
1. 一般化作用力 $F_r$ を求める
1. 一般化慣性力 $F^\star_r$を求める
1. Kaneの方程式 $F_r + F^\star_r=0$ より運動方程式を求める

以降は表記の簡略化のため、$S_\theta \coloneqq \sin(\theta), C_\theta \coloneqq \cos(\theta) $ と記述する。

### 1. 一般化座標 $\boldsymbol{q}$ の決定

この値を決めればシステムを構成するすべての物体(今回の場合はLink)の位置と姿勢が決まる、互いに独立な変数を選択する。

今回 $(X_2, Y_2)$ は $(x_1, \theta_2)$ により表されている。また、Link1の $y_0$ 方向の移動と回転 $\theta_1$ は固定されている。

$$
\begin{bmatrix}
X_2 \\ Y_2
\end{bmatrix} = 
\begin{bmatrix}
x_1 + l_2 S_{\theta_2} \\ l_2 C_{\theta_2}
\end{bmatrix}
$$

よって、一般化座標 $q$ は以下を選択する。

$$
\boldsymbol{q}= \begin{bmatrix}
q_1 \\ q_2
\end{bmatrix} =
\begin{bmatrix}
x_1 \\ \theta_2
\end{bmatrix}
$$


### 2. 一般化速度 $\boldsymbol{u}$ の決定

Kane法では、系の運動を表す互いに独立な変数として一般化速度を定義する。一般化速度は必ずしも一般化座標の時間微分と一致させる必要はなく、系の運動を簡単に表すことができるように自由に選ぶことが出来る。

まず、一般化速度を一般化座標の時間微分として次のように選択して進める。後で異なる一般化速度を設定し比較を行う。

$$
\boldsymbol{u} =
\begin{bmatrix}u_1 \\ u_2 \end{bmatrix}
=\begin{bmatrix} \dot{x}_1 \\ \dot{\theta}_2 \end{bmatrix}
$$

一般化座標と一般化速度を結ぶ運動学的微分方程式は以下となる。

$$
\dot{q}_1 = u_1 , \quad \dot{q}_2 = u_2
$$

### 3. 各物体の重心の並進速度と角速度を求める

それぞれのLinkの重心の並進速度と角速度を世界座標の$x_0, y_0, z_0$で表す。並進速度は$z_0$方向には生じないため、2次元ベクトルとして表し、角速度については回転軸周りの速度として表すため、3次元ベクトルで表記する。

#### Link 1

$$
\begin{aligned}
\boldsymbol{v}_1 &= \begin{bmatrix} \dot{x}_1 \\ 0 \end{bmatrix} = \begin{bmatrix} u_1 \\ 0 \end{bmatrix} \\
\boldsymbol{\omega}_1 &= \begin{bmatrix}  0 \\ 0 \\ 0 \end{bmatrix}
\end{aligned}
$$

#### Link 2

右手直交座標系の$e_z$軸($x_0 \times y_0$の外積方向)周りを正方向とすると、$\theta_2$ の正方向は $e_z$と反対方向になる。よって$\boldsymbol{\omega}_2$のように表す。

$$
\begin{aligned}
\boldsymbol{v}_2 &= \begin{bmatrix} \dot{X}_2 \\ \dot{Y}_2 \end{bmatrix} = \begin{bmatrix} u_1 + l_2 C_{q_2} u_2\\ -l_2 S_{q_2} u_2 \end{bmatrix} \\
\boldsymbol{\omega}_2 &= - \dot{\theta}_2 \begin{bmatrix}  0 \\ 0 \\ 1 \end{bmatrix} = -u_2 \boldsymbol{e}_z
\end{aligned}
$$


### 4. 偏速度(partial velocity)と偏角速度(partial angular velocity)を求める

偏速度と偏角速度は、それぞれの一般化速度 $\boldsymbol{u}$ が物体(Link)の並進速度 $\boldsymbol{v}_i$ および角速度 $ \boldsymbol{\omega}_i$ にどのように寄与するのかを表す。
そのため、以下のように各物体の動く方向に対して各一般化速度で偏微分を行う。

例えば、$\boldsymbol{v}_1^{(1)}$は他の一般化速度を一定として、$u_1$を単位量変化させたときの$\boldsymbol{v}_1$の変化方向である。これは、一般化速度に対する物体速度の感度ともいえる。

##### Link1

$$
\begin{aligned}
\boldsymbol{v}_1^{(1)} &= \frac{\partial \boldsymbol{v}_1}{\partial u_1} = \begin{bmatrix} 1 \\ 0 \end{bmatrix} \\
\boldsymbol{v}_1^{(2)} &= \frac{\partial \boldsymbol{v}_1}{\partial u_2} = \begin{bmatrix} 0 \\ 0 \end{bmatrix} \\
\boldsymbol{\omega}_1^{(1)} &= \frac{\partial \boldsymbol{\omega}_1}{\partial u_1} = \begin{bmatrix} 0 \\ 0 \\ 0\end{bmatrix} \\
\boldsymbol{\omega}_1^{(2)} &= \frac{\partial \boldsymbol{\omega}_1}{\partial u_2} = \begin{bmatrix} 0 \\ 0 \\ 0\end{bmatrix} \\
\end{aligned}
$$

##### Link2

$$
\begin{aligned}
\boldsymbol{v}_2^{(1)} &= \frac{\partial \boldsymbol{v}_2}{\partial u_1} 
= \begin{bmatrix} 1 \\ 0 \end{bmatrix} \\
\boldsymbol{v}_2^{(2)} &= \frac{\partial \boldsymbol{v}_2}{\partial u_2}
= \begin{bmatrix} l_2 C_{q_2} \\ -l_2 S_{q_2} \end{bmatrix} \\
\boldsymbol{\omega}_2^{(1)} &= \frac{\partial \boldsymbol{\omega}_2}{\partial u_1} 
= \begin{bmatrix} 0 \\ 0 \\ 0\end{bmatrix} \\
\boldsymbol{\omega}_2^{(2)} &= \frac{\partial \boldsymbol{\omega}_2}{\partial u_2} 
= \begin{bmatrix} 0 \\ 0 \\ -1\end{bmatrix} \\
\end{aligned}
$$

### 5. 一般化作用力 $F_r$を求める

各物体(Link)に作用する力とトルクを、一般化速度$u_r$に対応する偏速度、偏角速度方向へ射影し、その一般化運動方向への寄与を求める。

以下の式で$F_r$を求める。

$$
F_r = \sum_i \left( {\boldsymbol{v}_i^{(r)}}^T \boldsymbol{F}_i  + {\boldsymbol{\omega}_i^{(r)}}^T \boldsymbol{T}_i \right)
$$

$\boldsymbol{F}_i$は物体 $(i)$ に作用する力、$\boldsymbol{T}_i$ は各物体の重心まわりに作用するトルクを世界座標$(x_0, y_0, z_0)$で表したものである。

今回は、制御入力、重力および粘性項による力を一般化作用力の計算に用いる。

#### 拘束力について

理想拘束力は、拘束条件を満たす許容運動に対して仕事をしない。例えばレールからの拘束反力は許容される偏速度と直交するため寄与がゼロになる。これはLink1の偏速度 ${\boldsymbol{v}_1}^{(1)}$とLink1がレールから受ける拘束力 $\boldsymbol{N}$ は以下のように直交の関係があるため一般化作用力には表れない。

$$
{{\boldsymbol{v}_1}^{(1)}}^T \boldsymbol{N} = 
\begin{bmatrix} 1 & 0 \end{bmatrix} \begin{bmatrix} 0 \\ N \end{bmatrix} = 0
$$

また、Link1とLink2が接続されている関節点における拘束力も一般化作用力では次のように相殺される。<br>
関節点を$P$とすると、この点においてLink1がLink2から受ける反力を $-\boldsymbol{R}$、Link2がLink1から受ける反力を $\boldsymbol{R}$ とする。<br>
Link1でもLink2でも関節点は同じ速度であり、この点の偏速度も以下のように同じである。$\boldsymbol{v}_{P,i}^{(r)}$の右下の意味は、Link $i$における点$P$の速度である。

$$
\boldsymbol{v}_{P,1}^{(r)} = \boldsymbol{v}_{P,2}^{(r)} = \boldsymbol{v}_P^{(r)}
$$

関節点はLink1とともに水平方向へ動くので、その偏速度はゼロではない。よって、関節点(joint)の反力による一般化作用力への寄与は以下のようにゼロとなり現れない。

$$
\begin{aligned}
F_{r, \text{joint}} &= {\boldsymbol{v}_{P,1}^{(r)}}^{T} (-\boldsymbol{R}) + {\boldsymbol{v}_{P,2}^{(r)}}^{T} (\boldsymbol{R}) \\
&= {\boldsymbol{v}_{P}^{(r)}}^{T} (-\boldsymbol{R} +\boldsymbol{R}) \\
&= 0
\end{aligned}
$$

これは、同じ偏速度を持つ接続点に作用・反作用が働くため相殺される、ということである。

#### 粘性項による力の方向について

Link2 に加わる $\boldsymbol{T}_2$ には$\mu_2 u_2$がある。粘性トルクは$u_2$と反対方向に加わるが、$u_2$が$e_z$の方向とは反対であるため、以下のようになることに注意する。また、この粘性トルクの反作用もLink1に作用するが、Link1の偏角速度はゼロであるため、一般化作用力には寄与しない。

$$
\boldsymbol{T}_2 = \begin{bmatrix} 0 & 0 & \mu_2 u_2 \end{bmatrix}^T
$$

#### $r=1$

$u_1$方向の偏速度、偏角速度方向へ力を射影することで、$u_1$に関する作用力が現れる。

$$
\begin{aligned}
F_1 &= {\boldsymbol{v}_1^{(1)}}^T \boldsymbol{F}_1  + {\boldsymbol{\omega}_1^{(1)}}^T \boldsymbol{T}_1 +
{\boldsymbol{v}_2^{(1)}}^T \boldsymbol{F}_2  + {\boldsymbol{\omega}_2^{(1)}}^T \boldsymbol{T}_2 \\
&= \begin{bmatrix} 1 & 0 \end{bmatrix}\begin{bmatrix} f - \mu_1 u_1 \\ - m_1 g \end{bmatrix}
 + \begin{bmatrix} 0 & 0 & 0 \end{bmatrix}\begin{bmatrix} 0 \\ 0 \\ -\mu_2 u_2  \end{bmatrix}
 + \begin{bmatrix} 1 & 0 \end{bmatrix}\begin{bmatrix} 0 \\ - m_2 g \end{bmatrix}
 + \begin{bmatrix} 0 & 0 & 0 \end{bmatrix}\begin{bmatrix} 0 \\ 0 \\ \mu_2 u_2  \end{bmatrix} \\
 &= f - \mu_1 u_1
\end{aligned}
$$

#### $r=2$

$u_2$方向の偏速度、偏角速度方向へ力を射影することで、$u_2$に関する作用力が現れる。

$$
\begin{aligned}
F_2 &= {\boldsymbol{v}_1^{(2)}}^T \boldsymbol{F}_1  + {\boldsymbol{\omega}_1^{(2)}}^T \boldsymbol{T}_1 +
{\boldsymbol{v}_2^{(2)}}^T \boldsymbol{F}_2  + {\boldsymbol{\omega}_2^{(2)}}^T \boldsymbol{T}_2 \\
&= \begin{bmatrix} 0 & 0 \end{bmatrix}\begin{bmatrix} f - \mu_1 u_1 \\ - m_1 g \end{bmatrix}
 + \begin{bmatrix} 0 & 0 & 0 \end{bmatrix}\begin{bmatrix} 0 \\ 0 \\ -\mu_2 u_2  \end{bmatrix}
 + \begin{bmatrix} l_2 C_{q_2} & -l_2 S_{q_2} \end{bmatrix}\begin{bmatrix} 0 \\ - m_2 g \end{bmatrix}
 + \begin{bmatrix} 0 & 0 & -1 \end{bmatrix} \begin{bmatrix} 0 \\ 0 \\ \mu_2 u_2 \end{bmatrix} \\
 &= l_2 S_{q_2} m_2 g - \mu_2 u_2
\end{aligned}
$$


よって一般化作用力は以下でまとめる。

$$
\begin{bmatrix}
F_1 \\ F_2
\end{bmatrix}=
\begin{bmatrix}
f - \mu_1 u_1 \\ l_2 S_{q_2} m_2 g - \mu_2 u_2
\end{bmatrix}
$$


### 6. 一般化慣性力 $F_r^{*}$ を求める

並進の慣性力をニュートンの運動方程式により以下でまとめる。

$$
\boldsymbol{F}_i^{*} = - m_i \dot{\boldsymbol{v}}_i
$$

また回転の慣性トルクをオイラーの方程式により以下でまとめる。

$$
\boldsymbol{T}_i^{*} = -\left( \boldsymbol{I}_i \dot{\boldsymbol{\omega}}_i + \boldsymbol{\omega}_i \times \boldsymbol{I}_i \boldsymbol{\omega}_i \right)
$$

これにより一般化慣性力 $F_r^{*}$ を次のように求める。これは、一般化速度$u_r$に対応する偏速度、偏角速度方向へ慣性力を射影し、その方向への寄与を求めている。

$$
F_r^{*} = \sum_i \left( {\boldsymbol{v}_i^{(r)}}^T \boldsymbol{F}_i^{*}  + {\boldsymbol{\omega}_i^{(r)}}^T \boldsymbol{T}_i^{*} \right)
$$

$\dot{\boldsymbol{v}}_i$は各物体の重心の並進加速度を世界座標($x_0, y_0, z_0$)で表したものである。<br>
$\boldsymbol{\omega}_i, \dot{\boldsymbol{\omega}}_i$は物体(Link)$i$の角速度と角加速度を世界座標($x_0, y_0, z_0$)で表したものである。<br>

"3. 各物体の重心の並進速度と角速度を求める" で求められているものをそれぞれ時間微分を行って求める。

- 並進加速度

Link1が動く方向は$x_0$の方向だけであるため、次のようになる。

$$
\dot{\boldsymbol{v}}_1 = \begin{bmatrix} \dot{u}_1 \\ 0 \end{bmatrix}
$$

Link2の ${\boldsymbol{v}}_2$ から

$$
{\boldsymbol{v}}_2 =
\begin{bmatrix}
\dot{X}_2 \\ \dot{Y}_2
\end{bmatrix} = 
\begin{bmatrix}
u_1 + l_2 C_{q_2}u_2 \\
- l_2 S_{q_2} u_2
\end{bmatrix}
$$

$\dot{\boldsymbol{v}}_2$は以下となる。

$$
\dot{\boldsymbol{v}}_2 =
\begin{bmatrix}
\ddot{X}_2 \\ \ddot{Y}_2
\end{bmatrix} = 
\begin{bmatrix}
\dot{u}_1 - l_2 S_{q_2}{u_2}^2 + l_2 C_{q_2}\dot{u}_2 \\
 - l_2 C_{q_2} {u_2}^2 - l_2 S_{q_2} \dot{u}_2
\end{bmatrix}
$$

- 角加速度

Link1は回転しないため、以下となる。

$$
\boldsymbol{\omega}_1 = \dot{\boldsymbol{\omega}}_1 = \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}
$$

Link2の $\boldsymbol{\omega}_2$ より、

$$
\boldsymbol{\omega}_2 = - u_2 \begin{bmatrix}  0 \\ 0 \\ 1 \end{bmatrix}
$$

$\dot{\boldsymbol{\omega}}_2$ は以下となる。

$$
\dot{\boldsymbol{\omega}}_2 = - \dot{u}_2 \begin{bmatrix}  0 \\ 0 \\ 1 \end{bmatrix}
$$

$m_i$は重心の質量、 $\boldsymbol{I}_i$は重心周りの慣性テンソルである。今回のモデルでは、回転が発生するのはLink2の$z_0$軸周りのみであり、$z_0$軸を主軸慣性とする。計算では$I_{zz,2}=I_2$のみが現れるが、慣性テンソルは行列形式であることを意識するため、次のように表す。

$$
\boldsymbol{I}_1 = 
\begin{bmatrix}
I_{1,xx} &  &  \\
 & I_{1,yy} &  \\
 &  & I_{1,zz}
\end{bmatrix} , \quad
\boldsymbol{I}_2 = 
\begin{bmatrix}
I_{2,xx} &  &  \\
 & I_{2,yy} &  \\
 &  & I_{2}
\end{bmatrix} 
$$

- 並進の慣性力

Link1

$$
\boldsymbol{F}_1^{*} = \begin{bmatrix} -m_1\dot{u}_1 \\ 0 \end{bmatrix}
$$

Link2

$$
\boldsymbol{F}_2^{*} = 
\begin{bmatrix}
 -m_2\ddot{X}_2 \\ -m_2 \ddot{Y}_2
\end{bmatrix} = 
\begin{bmatrix}
-m_2\dot{u}_1 + m_2 l_2 S_{q_2}{u_2}^2 - m_2 l_2 C_{q_2}\dot{u}_2 \\
m_2 l_2 C_{q_2} {u_2}^2 + m_2 l_2 S_{q_2} \dot{u}_2
\end{bmatrix}
$$

- 回転の慣性トルク

Link1

$$
\begin{aligned}
\boldsymbol{T}_1^{*} &= -\left( \boldsymbol{I}_1 \dot{\boldsymbol{\omega}}_1 + \boldsymbol{\omega}_1 \times \boldsymbol{I}_1 \boldsymbol{\omega}_1 \right) \\
&= -\left(
\begin{bmatrix}
I_{1,xx} &  &  \\
 & I_{1,yy} &  \\
 &  & I_{1,zz}
\end{bmatrix}
\begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}
+
\begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix} \times
\begin{bmatrix}
I_{1,xx} &  &  \\
 & I_{1,yy} &  \\
 &  & I_{1,zz}
\end{bmatrix}
\begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}
\right) \\
&=
\begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix}
\end{aligned}
$$

Link2

$$
\begin{aligned}
\boldsymbol{T}_2^{*} &= -\left( \boldsymbol{I}_2 \dot{\boldsymbol{\omega}}_2 + \boldsymbol{\omega}_2 \times \boldsymbol{I}_2 \boldsymbol{\omega}_2 \right) \\
&= -\left(
\begin{bmatrix}
I_{2,xx} &  &  \\
 & I_{2,yy} &  \\
 &  & I_{2}
\end{bmatrix} 
\begin{bmatrix}  0 \\ 0 \\ - \dot{u}_2  \end{bmatrix}+
\begin{bmatrix}  0 \\ 0 \\ - u_2 \end{bmatrix} \times
\begin{bmatrix}
I_{2,xx} &  &  \\
 & I_{2,yy} &  \\
 &  & I_{2}
\end{bmatrix} 
\begin{bmatrix}  0 \\ 0 \\ - u_2 \end{bmatrix}
\right) \\
&=
-\left(
\begin{bmatrix} 0 \\ 0 \\ - \dot{u}_2 I_2 \end{bmatrix} +
I_2 
\begin{bmatrix}  0 \\ 0 \\ - u_2 \end{bmatrix} \times
\begin{bmatrix}  0 \\ 0 \\ - u_2 \end{bmatrix}
\right) \\
&= \begin{bmatrix} 0 \\ 0 \\ \dot{u}_2 I_2 \end{bmatrix}
\end{aligned}
$$



- $r=1$

$u_1$方向への偏速度、偏角速度方向へ慣性力と慣性トルクを射影することで、$u_1$に関する慣性力が現れる。

$$
\begin{aligned}
F_1^{*} &= {\boldsymbol{v}_1^{(1)}}^T \boldsymbol{F}_1^{*}  + {\boldsymbol{\omega}_1^{(1)}}^T \boldsymbol{T}_1^{*} + {\boldsymbol{v}_2^{(1)}}^T \boldsymbol{F}_2^{*}  + {\boldsymbol{\omega}_2^{(1)}}^T \boldsymbol{T}_2^{*}  \\
&= \begin{bmatrix} 1 & 0 \end{bmatrix} \begin{bmatrix} -m_1\dot{u}_1 \\ 0 \end{bmatrix} +
\begin{bmatrix} 0 & 0 & 0\end{bmatrix} \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix} \\
& \quad +
\begin{bmatrix} 1 & 0 \end{bmatrix} 
\begin{bmatrix}
-m_2\dot{u}_1 + m_2 l_2 S_{q_2}{u_2}^2 - m_2 l_2 C_{q_2}\dot{u}_2 \\
m_2 l_2 C_{q_2} {u_2}^2 + m_2 l_2 S_{q_2} \dot{u}_2
\end{bmatrix} +
\begin{bmatrix} 0 & 0 & 0\end{bmatrix} 
\begin{bmatrix} 0 \\ 0 \\ \dot{u}_2 I_2 \end{bmatrix} \\
&= -m_1\dot{u}_1 -m_2\dot{u}_1 + m_2 l_2 S_{q_2}{u_2}^2 - m_2 l_2 C_{q_2}\dot{u}_2 \\
&= -(m_1 + m_2) \dot{u}_1 + m_2 l_2 S_{q_2}{u_2}^2 - m_2 l_2 C_{q_2}\dot{u}_2 \\
\end{aligned}
$$

- $r=2$

$u_2$方向への偏速度、偏角速度方向へ慣性力と慣性トルクを射影することで、$u_2$に関する慣性力が現れる。

$$
\begin{aligned}
F_2^{*} &= {\boldsymbol{v}_1^{(2)}}^T \boldsymbol{F}_1^{*}  + {\boldsymbol{\omega}_1^{(2)}}^T \boldsymbol{T}_1^{*} + {\boldsymbol{v}_2^{(2)}}^T \boldsymbol{F}_2^{*}  + {\boldsymbol{\omega}_2^{(2)}}^T \boldsymbol{T}_2^{*}  \\
&= \begin{bmatrix} 0 & 0 \end{bmatrix} \begin{bmatrix} -m_1\dot{u}_1 \\ 0 \end{bmatrix} +
\begin{bmatrix} 0 & 0 & 0\end{bmatrix} \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix} \\
& \quad +
\begin{bmatrix} l_2 C_{q_2} & -l_2 S_{q_2} \end{bmatrix} 
\begin{bmatrix}
-m_2\dot{u}_1 + m_2 l_2 S_{q_2}{u_2}^2 - m_2 l_2 C_{q_2}\dot{u}_2 \\
m_2 l_2 C_{q_2} {u_2}^2 + m_2 l_2 S_{q_2} \dot{u}_2
\end{bmatrix} +
\begin{bmatrix} 0 & 0 & -1\end{bmatrix} 
\begin{bmatrix} 0 \\ 0 \\ \dot{u}_2 I_2 \end{bmatrix} \\
&= l_2 C_{q_2}\left( -m_2 \dot{u}_1 + m_2 l_2 S_{q_2}{u_2}^2 - m_2 l_2 C_{q_2}\dot{u}_2 \right)
-l_2 S_{q_2} \left( m_2 l_2 C_{q_2} {u_2}^2 + m_2 l_2 S_{q_2} \dot{u}_2 \right) - I_2 \dot{u}_2 \\
&= -m_2 l_2 C_{q_2} \dot{u}_1 + l_2 C_{q_2} m_2 l_2 S_{q_2}{u_2}^2 - l_2 C_{q_2} m_2 l_2 C_{q_2}\dot{u}_2 - l_2 S_{q_2} m_2 l_2 C_{q_2} {u_2}^2 - l_2 S_{q_2} m_2 l_2 S_{q_2} \dot{u}_2 - I_2 \dot{u}_2 \\
&= - I_2 \dot{u}_2 -m_2 l_2 C_{q_2} \dot{u}_1 + \cancel{m_2 l_2^2 C_{q_2} S_{q_2}{u_2}^2} - \cancel{m_2 l_2^2 S_{q_2} C_{q_2} {u_2}^2} - m_2 l_2^2 C_{q_2} C_{q_2}\dot{u}_2 - m_2 l_2^2 S_{q_2} S_{q_2} \dot{u}_2 \\
&= - I_2 \dot{u}_2 -m_2 l_2 C_{q_2} \dot{u}_1 - m_2 l_2^2 \left( C_{q_2}^2 + S_{q_2}^2 \right) \dot{u}_2 \\
&= - I_2 \dot{u}_2 -m_2 l_2 C_{q_2} \dot{u}_1 - m_2 l_2^2 \dot{u}_2
\end{aligned}
$$

よって、一般化慣性力は次のようになる。

$$
\begin{bmatrix}
F_1^{*} \\ F_2^{*}
\end{bmatrix} =
\begin{bmatrix}
-(m_1 + m_2) \dot{u}_1 + m_2 l_2 S_{q_2}{u_2}^2 - m_2 l_2 C_{q_2}\dot{u}_2 \\
 - I_2 \dot{u}_2  -m_2 l_2 C_{q_2} \dot{u}_1 - m_2 l_2^2 \dot{u}_2
\end{bmatrix}
$$


### 7. Kane の方程式 $F_r + F_r^{*} =0$ より運動方程式を求める

Kaneの方程式 $F_r + F_r^{*}=0$ は、一般化作用力 $F_r$ と一般化慣性力 $F_r^{*}$ のつり合いを表す。以下では $F_r^{*}$ に含まれる加速度項に比例する項を符号反転して左辺へ移し、慣性項として整理する。なお、速度二乗項も一般化慣性力に由来するが、質量行列形式では右辺へ配置する。

求めた一般化作用力$F_r$と一般化慣性力$F_r^{*}$を以下に示す。

$$
\begin{bmatrix}
F_1 \\ F_2
\end{bmatrix}=
\begin{bmatrix}
f - \mu_1 u_1 \\ l_2 S_{q_2} m_2 g - \mu_2 u_2
\end{bmatrix}
$$

$$
\begin{bmatrix}
F_1^{*} \\ F_2^{*}
\end{bmatrix} =
\begin{bmatrix}
-(m_1 + m_2) \dot{u}_1 + m_2 l_2 S_{q_2}{u_2}^2 - m_2 l_2 C_{q_2}\dot{u}_2 \\
 - I_2 \dot{u}_2  -m_2 l_2 C_{q_2} \dot{u}_1 - m_2 l_2^2 \dot{u}_2
\end{bmatrix}
$$

これらをKaneの方程式$F_r + F_r^{*}=0$へ代入する。

- $r=1$

$$
f - \mu_1 u_1 -(m_1 + m_2) \dot{u}_1 + m_2 l_2 S_{q_2}{u_2}^2 - m_2 l_2 C_{q_2}\dot{u}_2 = 0
$$

加速度 $\dot{\boldsymbol{u}}$ を含む項を左辺へ集めて整理する。 

$$
(m_1 + m_2) \dot{u}_1 + m_2 l_2 C_{q_2}\dot{u}_2 =  f - \mu_1 u_1 + m_2 l_2 S_{q_2}{u_2}^2
$$

これは一般化速度$u_1=\dot{x}_1$方向に対する運動である。

左辺は以下の項の合計である。
- Link1とLink2の合計質量$m_1 + m_2$に対し$\dot{u}_1 = \ddot{x}_1$の加速度を掛けた慣性項
- [Newton-Euler法のNote](./cartpole_2d_newton_euler.ipynb)の"拘束力について"の節で求めた$u_1$方向に対するLink2の接線加速度$a_{t,x}=l_2 C_{q_2}\dot{u}_2$にLink2の質量 $m_2$を掛けた慣性項

右辺は以下の項の合計である。
- 制御入力$f$
- 運動方向と反対に作用する粘性力$-\mu_1 u_1$
- Link2の向心加速度$a_{c,x}=-l_2 S_{q_2}{u_2}^2$にLink2の質量 $m_2$を掛けた慣性項。これを右辺へ移項しているため、符号が反転している。

- $r=2$

$$
l_2 S_{q_2} m_2 g - \mu_2 u_2  - I_2 \dot{u}_2  -m_2 l_2 C_{q_2} \dot{u}_1 - m_2 l_2^2 \dot{u}_2 = 0
$$

加速度 $\dot{\boldsymbol{u}}$ を含む項を左辺へ集めて整理する。 

$$
(I_2 + m_2 l_2^2) \dot{u}_2 + m_2 l_2 C_{q_2} \dot{u}_1 = l_2 S_{q_2} m_2 g - \mu_2 u_2 
$$

これは、一般化速度$u_2=\dot{\theta}_2$方向に対する関節点周りの回転運動である。

左辺は以下の項の合計である。

- ($I_2 + m_2 l_2^2$)は平行軸の定理を適用したLink2の慣性モーメントであり、回転関節周りの慣性モーメントであることを示している。これに$\dot{u}_2 = \ddot{\theta}_2$を掛けたトルクに関する慣性項
- 関節点の水平方向加速度$\dot{u}_1 = \ddot{x}_1$によりLink2の重心にも$\dot{u}_1$の加速度が生じる。その加速度をLink2を回転させる方向へ射影した加速度$C_{\theta_2} \dot{u}_1$にLink2の質量と関節から質点までの距離$l_2$を掛けたトルクに関する慣性項

右辺は以下の項の合計である。

- Link2に作用する重力の接点方向成分$S_{q_2} m_2 g$に、関節点から重心までの距離$l_2$を掛けた重力トルク
- 運動方向と反対に作用する粘性トルク $-\mu_2 u_2$

## 別の一般化速度 $\boldsymbol{u}$ を選んだ場合の運動方程式


上記で設定した一般速度 $\boldsymbol{u}$ とは異なる、$\tilde{\boldsymbol{u}}$ を以下に設定する。変更は $u_2$ のみであるため、これを $\tilde{u}_2$ で表す。設定した $\tilde{u}_2 = l_2 \dot{\theta}_2$ はLink2の接線方向の速度であり、上記で$u_2=\dot{theta}_2$として求めた回転に関するものとは異なる形で運動方程式が現れることを確認する。

$$
\tilde{\boldsymbol{u}} = 
\begin{bmatrix}
{u}_1 \\ \tilde{u}_2
\end{bmatrix} =
\begin{bmatrix}
\dot{x}_1 \\ l_2 \dot{\theta}_2
\end{bmatrix}
$$

運動学的微分方程式は次となる。

$$
\dot{q}_2 = \tilde{u}_2 / l_2
$$

物体の重心の並進速度と角速度は$\boldsymbol{\tilde{v}}_2, \boldsymbol{\tilde{\omega}}_2$ となる。

$$
\boldsymbol{\tilde{v}}_2 =
\begin{bmatrix}
\dot{X}_2 \\ \dot{Y}_2
\end{bmatrix} = 
\begin{bmatrix}
\dot{x}_1 + l_2 C_{\theta_2} \dot{\theta}_2 \\
- l_2 S_{\theta_2} \dot{\theta}_2
\end{bmatrix} =
\begin{bmatrix}
u_1 + C_{q_2} \tilde{u}_2 \\
- S_{q_2} \tilde{u}_2
\end{bmatrix}
$$

$$
\boldsymbol{\tilde{\omega}}_2 = -\dot{\theta}_2 \begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix} =
-\frac{\tilde{u}_2}{l_2} \boldsymbol{e}_z
$$

これにより、偏速度と偏角速度は以下のように変化する。

$$
\begin{aligned}
\boldsymbol{\tilde{v}}_2^{(2)} &= \frac{\partial \boldsymbol{\tilde{v}}_2}{\partial \tilde{u}_2}
= \begin{bmatrix}  C_{q_2} \\ -S_{q_2} \end{bmatrix} \\
\boldsymbol{\tilde{\omega}}_2^{(2)} &= \frac{\partial \boldsymbol{\tilde{\omega}}_2}{\partial \tilde{u}_2} 
= \begin{bmatrix} 0 \\ 0 \\ -1/l_2\end{bmatrix} \\
\end{aligned}
$$


一般化作用力に$r=2$の項が以下のように変化する。

Link2に作用する粘性項は次のように変化する。

$$
\boldsymbol{\tilde{T}}_2 = \begin{bmatrix} 0 & 0 & \mu_2 \tilde{u}_2/l_2\end{bmatrix}^T
$$

よって$\tilde{u}_2$方向への偏速度と偏角速度へ作用力を射影することで、$\tilde{u}_2$に関する一般化作用力が現れる。

$$
\begin{aligned}
\tilde{F}_2 &= {\boldsymbol{{v}}_1^{(2)}}^T \boldsymbol{F}_1  + {\boldsymbol{{\omega}}_1^{(2)}}^T \boldsymbol{T}_1 +
{\boldsymbol{\tilde{v}}_2^{(2)}}^T \boldsymbol{F}_2  + {\boldsymbol{\tilde{\omega}}_2^{(2)}}^T \boldsymbol{\tilde{T}}_2 \\
&= \begin{bmatrix} 0 & 0 \end{bmatrix}\begin{bmatrix} f - \mu_1 u_1 \\ - m_1 g \end{bmatrix}
 + \begin{bmatrix} 0 & 0 & 0 \end{bmatrix}\begin{bmatrix} 0 \\ 0 \\ -\mu_2 \tilde{u}_2/l_2  \end{bmatrix}
 + \begin{bmatrix} C_{q_2} & -S_{q_2} \end{bmatrix}\begin{bmatrix} 0 \\ - m_2 g \end{bmatrix}
 + \begin{bmatrix} 0 & 0 & -1/l_2 \end{bmatrix} \begin{bmatrix} 0 \\ 0 \\ \mu_2 \tilde{u}_2/l_2 \end{bmatrix} \\
 &= S_{q_2} m_2 g - \mu_2 \tilde{u}_2 / l_2^2
\end{aligned}
$$

よって一般化作用力は以下でまとめる。

$$
\begin{bmatrix}
F_1 \\ \tilde{F}_2
\end{bmatrix}=
\begin{bmatrix}
f - \mu_1 u_1 \\
 S_{q_2} m_2 g - \mu_2 \tilde{u}_2 / l_2^2
\end{bmatrix}
$$



一般化慣性力 $F_r^{*}$はLink2に関する並進加速度と角加速度が次のようになる。

$$
\dot{\boldsymbol{\tilde{v}}}_2 =
\begin{bmatrix}
\ddot{X}_2 \\ \ddot{Y}_2
\end{bmatrix} = 
\begin{bmatrix}
\dot{u}_1 - S_{q_2} \tilde{u}_2^2/l_2 + C_{q_2} \dot{\tilde{u}}_2 \\
- C_{q_2} \tilde{u}_2^2/l_2 - S_{q_2} \dot{\tilde{u}}_2
\end{bmatrix}
$$

$$
\boldsymbol{\tilde{\omega}}_2 = -\frac{\tilde{u}_2}{l_2} \begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix}
$$

$$
\dot {\boldsymbol{\tilde{\omega}}}_2 = -\frac{\dot{\tilde{u}}_2}{l_2} \begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix}
$$

Link2の並進の慣性力は以下となる。

$$
\tilde{\boldsymbol{F}}_2^{*} = 
\begin{bmatrix}
 -m_2\ddot{X}_2 \\ -m_2 \ddot{Y}_2
\end{bmatrix} = 
\begin{bmatrix}
-m_2 \dot{u}_1 + m_2 S_{q_2} \tilde{u}_2^2/l_2 -m_2 C_{q_2} \dot{\tilde{u}}_2 \\
 m_2 C_{q_2} \tilde{u}_2^2/l_2 + m_2 S_{q_2} \dot{\tilde{u}}_2
\end{bmatrix}
$$

Link2の回転の慣性トルクは以下となる。

$$
\begin{aligned}
\tilde{\boldsymbol{T}}_2^{*} = 
\begin{bmatrix} 0 \\ 0 \\ \dot{\tilde{u}}_2 I_2/l_2 \end{bmatrix}
\end{aligned}
$$

よって一般化慣性力$F_r$は以下のようになる。

- $r=1$

$$
\begin{aligned}
\tilde{F}_1^{*} &= {\boldsymbol{v}_1^{(1)}}^T \boldsymbol{F}_1^{*}  + {\boldsymbol{\omega}_1^{(1)}}^T \boldsymbol{T}_1^{*} + {\boldsymbol{v}_2^{(1)}}^T \boldsymbol{\tilde{F}}_2^{*}  + {\boldsymbol{\omega}_2^{(1)}}^T \boldsymbol{\tilde{T}}_2^{*}  \\
&= \begin{bmatrix} 1 & 0 \end{bmatrix} \begin{bmatrix} -m_1\dot{u}_1 \\ 0 \end{bmatrix} +
\begin{bmatrix} 0 & 0 & 0\end{bmatrix} \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix} \\
& \quad +
\begin{bmatrix} 1 & 0 \end{bmatrix} 
\begin{bmatrix}
-m_2 \dot{u}_1 + m_2 S_{q_2} \tilde{u}_2^2/l_2 -m_2 C_{q_2} \dot{\tilde{u}}_2 \\
 m_2 C_{q_2} \tilde{u}_2^2/l_2 + m_2 S_{q_2} \dot{\tilde{u}}_2
\end{bmatrix} +
\begin{bmatrix} 0 & 0 & 0\end{bmatrix} 
\begin{bmatrix} 0 \\ 0 \\ \dot{\tilde{u}}_2 I_2/l_2\end{bmatrix} \\
&= -m_1\dot{u}_1 -m_2 \dot{u}_1 + m_2 S_{q_2} \tilde{u}_2^2/l_2 -m_2 C_{q_2} \dot{\tilde{u}}_2  \\
&= -(m_1 + m_2) \dot{u}_1 + m_2 S_{q_2} \tilde{u}_2^2/l_2 -m_2 C_{q_2} \dot{\tilde{u}}_2 \\
\end{aligned}
$$

- $r=2$

$$
\begin{aligned}
\tilde{F}_2^{*} &= {\boldsymbol{v}_1^{(2)}}^T \boldsymbol{F}_1^{*}  + {\boldsymbol{\omega}_1^{(2)}}^T \boldsymbol{T}_1^{*} + {\boldsymbol{\tilde{v}}_2^{(2)}}^T {\boldsymbol{\tilde{F}}}_2^{*}  + {\boldsymbol{\tilde{\omega}}_2^{(2)}}^T \boldsymbol{\tilde{T}}_2^{*}  \\
&= \begin{bmatrix} 0 & 0 \end{bmatrix} \begin{bmatrix} -m_1\dot{u}_1 \\ 0 \end{bmatrix} +
\begin{bmatrix} 0 & 0 & 0\end{bmatrix} \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix} \\
& \quad +
\begin{bmatrix} C_{q_2} & -S_{q_2} \end{bmatrix} 
\begin{bmatrix}
-m_2 \dot{u}_1 + m_2 S_{q_2} \tilde{u}_2^2/l_2 -m_2 C_{q_2} \dot{\tilde{u}}_2 \\
 m_2 C_{q_2} \tilde{u}_2^2/l_2 + m_2 S_{q_2} \dot{\tilde{u}}_2
\end{bmatrix} +
\begin{bmatrix} 0 & 0 & -1/l_2\end{bmatrix} 
\begin{bmatrix} 0 \\ 0 \\ \dot{\tilde{u}}_2 I_2/l_2 \end{bmatrix} \\
&= C_{q_2}\left( -m_2 \dot{u}_1 + m_2 S_{q_2} \tilde{u}_2^2/l_2 -m_2 C_{q_2} \dot{\tilde{u}}_2 \right)
-S_{q_2} \left( m_2 C_{q_2} \tilde{u}_2^2/l_2 + m_2 S_{q_2} \dot{\tilde{u}}_2 \right) - I_2 \dot{\tilde{u}}_2 /l_2^2 \\
&= -m_2 C_{q_2} \dot{u}_1 + m_2 C_{q_2} S_{q_2} \tilde{u}_2^2/l_2 - m_2 C_{q_2}^2 \dot{\tilde{u}}_2 - m_2 S_{q_2} C_{q_2} \tilde{u}_2^2/l_2 - m_2 S_{q_2}^2 \dot{\tilde{u}}_2 - I_2 \dot{\tilde{u}}_2/l_2^2 \\
&= - I_2 \dot{\tilde{u}}_2/l_2^2 - m_2 C_{q_2} \dot{u}_1 +\cancel{m_2 C_{q_2} S_{q_2} \tilde{u}_2^2/l_2} - \cancel{m_2 S_{q_2} C_{q_2} \tilde{u}_2^2/l_2} - m_2 C_{q_2}^2 \dot{\tilde{u}}_2 - m_2 S_{q_2}^2 \dot{\tilde{u}}_2 \\
&= - I_2 \dot{\tilde{u}}_2/l_2^2 -m_2 C_{q_2} \dot{u}_1 - m_2 \left( C_{q_2}^2 + S_{q_2}^2 \right) \dot{\tilde{u}}_2 \\
&= - I_2 \dot{\tilde{u}}_2/l_2^2 -m_2 C_{q_2} \dot{u}_1 - m_2 \dot{\tilde{u}}_2
\end{aligned}
$$


よって一般化慣性力は以下でまとめる。

$$
\begin{bmatrix}
\tilde{F}_1^{*} \\ \tilde{F}_2^{*}
\end{bmatrix} = 
\begin{bmatrix}
-(m_1 + m_2) \dot{u}_1 + m_2 S_{q_2} \tilde{u}_2^2/l_2 -m_2 C_{q_2} \dot{\tilde{u}}_2 \\
- I_2 \dot{\tilde{u}}_2/l_2^2 -m_2 C_{q_2} \dot{u}_1 - m_2 \dot{\tilde{u}}_2
\end{bmatrix}
$$



Kaneの方程式 $F_r + F_r^{*}=0$ へ代入し、運動方程式を求めると次のようになる。

- $r=1$ : $F_1 + \tilde{F}_1^{*}=0$

f - \mu_1 u_1 -(m_1 + m_2) \dot{u}_1 + m_2 S_{q_2} \tilde{u}_2^2/l_2 -m_2 C_{q_2} \dot{\tilde{u}}_2 =0
$$

加速度$\dot{\tilde{\boldsymbol{u}}}$を含む項を左辺へ集めて整理する。

$$
(m_1 + m_2) \dot{u}_1 + m_2 C_{q_2} \dot{\tilde{u}}_2 = f - \mu_1 u_1 + m_2 S_{q_2} \tilde{u}_2^2/l_2
$$

この式は一般化速度$u_1$方向に対する運動である。そして、$\tilde{u}_2$は接線速度であるため、以下のように運動方程式が構成される。

- Link2の質量 $m_2$ に接線方向の加速度 $\dot{\tilde{u}}_2$ を$u_1$方向に射影した加速度 $C_{q_2} \dot{\tilde{u}}_2$ を掛けた慣性項

右辺は以下の項の合計である。

- 制御入力 $f$
- 運動方向と反対に作用する粘性力$-\mu_1 u_1$
- Link2の向心加速度$a_c = l_2 \tilde{\boldsymbol{\omega}}^2$は$\tilde{u}_2$を使い、$a_c = \tilde{u}_2^2/l_2$となる。これを$u_1$方向へ射影した加速度$a_{c,x} = -S_{q_2}\tilde{u}_2^2 / l_2$ に、Link2の質量$m_2$を掛けた慣性項。これを右辺へ移項しているため、符号が反転している。

$$

加速度$\dot{\tilde{\boldsymbol{u}}}$を含む項を左辺へ集めて整理する。

$$
(I_2/l_2^2 + m_2) \dot{\tilde{u}}_2 + m_2 C_{q_2} \dot{u}_1 = S_{q_2} m_2 g - \mu_2 \tilde{u}_2 / l_2^2
$$

この式は一般化速度$\tilde{u}_2$方向に対する運動である。

- ($I_2/l_2^2 + m_2 l_2^2$)は,,,,

右辺は以下の項の合計である。

- Link2に作用する重力の接点方向成分$S_{q_2} m_2 g$
- 粘性項$\mu_2$は角速度に対応するトルクを$-\mu_2 \tilde{\boldsymbol{\omega}}_2$のように生じさせる。トルクを力に変換するため、支点である関節点から力点である$m_2$までの距離で割り、$\tilde{\boldsymbol{\omega}}_2 = \tilde{u}_2/l_2$ であるため、接線方向に対する粘性項$\mu_2$による粘性力は$-\mu_2 \tilde{u}_2 / l_2^2$となる。
