- Link1 と Link2は、Link1の重心位置にある回転関節で接続されている。回転は自由で、位置は一致するように拘束するように構成されている。
- Link1の質量は$m_1$
- Link2の質量は$m_2$で、重心まわりの慣性モーメントは$I_2$
- Link1は垂直方向($y_0$)を固定されており、水平方向($x_0$)に動作できる。
- 制御入力 $f$ はLink1に加えられ、Link2には直接的な制御入力や関節トルクを加えない。
- $l_2$は関節点からLink2の重心までの距離
- $\theta_2$は鉛直上向きから時計回りを正とするLink2の角度
- $\mu_1$はLink1の並進粘性係数
- $\mu_2$は関節の回転粘性係数
- $g$は重力加速度

KANE法は各物体に働く力と慣性力を一般化速度に対応する運動方向に射影して運動方程式を導出する。

おおむね次の順番で進める。

1. 一般化座標の決定
1. 一般化速度の決定
1. 各物体の重心の並進速度と角速度を求める
1. 偏速度(partial velocity)、偏角速度(partial anguler velocity)を求める
1. 一般化作用力 $F_r$ を求める
1. 一般化慣性力 $F^\star_r$を求める
1. KANEの方程式 $F_r + F^\star_r=0$ より運動方程式を求める

以降は表記の簡略化のため、$S_\theta \coloneqq \sin(\theta), C_\theta \coloneqq \cos(\theta) $ と記述する。

### 1. 一般化座標 $\boldsymbol{q}$ の決定

この値を決めればシステムを構成するすべての物体(今回の場合はLink)の位置と姿勢が決まる、互いに独立な変数を選択する。

今回 $(X_2, Y_2)$ は $(x_1, \theta_2)$ により表されている。また、Link1の $y_0$ 方向の移動と回転 $\theta_1$ は固定されている。

$$
\begin{bmatrix}
X_2 \\ Y_2
\end{bmatrix} = 
\begin{bmatrix}
x_1 + l_2 S_{\theta_2} \\ l_2 C_{\theta_2}
\end{bmatrix}
$$

よって、一般化座標 $q$ は以下を選択する。

$$
\boldsymbol{q}= \begin{bmatrix}
q_1 \\ q_2
\end{bmatrix} =
\begin{bmatrix}
x_1 \\ \theta_2
\end{bmatrix}
$$


### 2. 一般化速度 $\boldsymbol{u}$ の決定

Kane法では、系の運動を表す互いに独立な変数として一般化速度を定義する。一般化速度は必ずしも一般化座標の時間微分と一致させる必要はない。

今回のシステムでは、一般化速度を一般化座標の時間微分として次のように選択する。

$$
\boldsymbol{u} =
\begin{bmatrix}u_1 \\ u_2 \end{bmatrix}
=\begin{bmatrix} \dot{x}_1 \\ \dot{\theta}_2 \end{bmatrix}
$$

一般化座標と一般化速度を結ぶ運動学的微分方程式は以下となる。

$$
\dot{q}_1 = u_1 , \quad \dot{q}_2 = u_2
$$

### 3. 各物体の重心の並進速度と角速度を求める

それぞれのLinkの重心の並進速度と角速度は以下となる。角速度については回転軸周りの速度として表すため、3次元ベクトルで表記する。

#### Link 1

$$
\begin{aligned}
\boldsymbol{v}_1 &= \begin{bmatrix} \dot{x}_1 \\ 0 \end{bmatrix} = \begin{bmatrix} u_1 \\ 0 \end{bmatrix} \\
\boldsymbol{\omega}_1 &= \begin{bmatrix}  0 \\ 0 \\ 0 \end{bmatrix}
\end{aligned}
$$

#### Link 2

$e_z$軸周りを正方向とすると、$\theta_2$ の正方向は $e_z$と反対方向になる。

$$
\begin{aligned}
\boldsymbol{v}_2 &= \begin{bmatrix} \dot{X}_2 \\ \dot{Y}_2 \end{bmatrix} = \begin{bmatrix} u_1 + l_2 C_{q_2} u_2\\ -l_2 S_{q_2} \end{bmatrix} \\
\boldsymbol{\omega}_1 &= - \dot{\theta}_2 \begin{bmatrix}  0 \\ 0 \\ 1 \end{bmatrix} = -u_2 \boldsymbol{e}_z\
\end{aligned}
$$
