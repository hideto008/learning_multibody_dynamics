- Link1 と Link2が、Link1の重心で、回転は自由で、位置は一致するように拘束するように構成されている。
- Link1の質量は$m_1$
- Link2の質量は$m_2$で、重心まわりの慣性モーメントは$I_2$
- Link1は垂直方向($y_0$)を固定されており、水平方向($x_0$)に動作できる。
- Link2はLink1に印加される力 $f$ により回転する。


Link1とLink2は関節点で回転が自由に動くように構成されているが、位置は動かないように拘束されている。このため位置を拘束するように力が発生するが、この値は拘束を満たしながら運動を成立させるように決まるため未知数として扱い運動方程式を立てる。

拘束力を下の図のように$H,V$で描く。
関節点では、水平方向と鉛直方向に拘束力が図の方向に発生しているとする。これを$H,V$で表す。

作用・反作用の法則によりLink1とLink2は図のように逆方向を向いている。

### 運動方程式

以降は式の簡略化のため$\sin(\theta)=S_\theta, \cos(\theta) = C_\theta$ のように表す。

#### Link1

$x_0$方向に運動が生じるので、入力、減衰力、拘束力により次のようになる。

$$
m_1 \ddot{x}_1(t) = f(t) - \mu_1 \dot{x}_1(t) - H(t)
$$

#### Link2

重心$(X_2, Y_2)$の運動方程式を立てる。重力と拘束力により次のようになる。

$$
\begin{aligned}
m_2 \ddot{X}_2(t) &= H(t) \\
m_2 \ddot{Y}_2(t) &= V(t)  - m_2 g
\end{aligned}
$$

Link2の重心まわりの運動方程式を立てる。重力は重心に作用するため重心まわりの加点モーメントには寄与しない。よって拘束力と減衰力により次となる。時計回りと回転の正方向としている。

$$
I_2 \ddot{\theta}_2(t) = V(t) S_{\theta_2} l_2 - H(t) C_{\theta_2} l_2  - \mu_2 \dot{\theta_2}(t)
$$


上式から未知数の$H(t), V(t)$を消去する。

$$
\begin{aligned}
H(t) &= m_2 \ddot{X}_2(t) \\
V(t) &= m_2 \ddot{Y}_2(t) + m_2 g
\end{aligned}
$$

これを代入すると、以下の2式となる。

$$
\begin{aligned}
m_1 \ddot{x}_1(t) &= f(t) - \mu_1 \dot{x}_1(t) - m_2 \ddot{X}_2(t) \\
I_2 \ddot{\theta}_2(t) &= \left[m_2 \ddot{Y}_2(t) + m_2 g\right] S_{\theta_2} l_2 - m_2 \ddot{X}_2(t) C_{\theta_2} l_2 - \mu_2 \dot{\theta_2}(t) \\
&= m_2 g S_{\theta_2} l_2 - \mu_2 \dot{\theta_2}(t) + m_2 l_2 \ddot{Y}_2(t) S_{\theta_2} - m_2 l_2 \ddot{X}_2(t) C_{\theta_2} 
\end{aligned}
$$

$(X_2, Y_2)$ を $(x_0, y_0)$ で表すと以下となる。

$$
\begin{aligned}
X_2(t) &= x_1(t) + l_2 S_{\theta_2} \\
Y_2(t) &= l_2 C_{\theta_2}
\end{aligned}
$$

よって、速度は

$$
\begin{aligned}
\dot{X}_2(t) &= \dot{x}_1(t) + l_2 C_{\theta_2} \dot{\theta}_2(t)\\
\dot{Y}_2(t) &= - l_2 S_{\theta_2} \dot{\theta}_2(t)
\end{aligned}
$$

加速度は以下となる。

$$
\begin{aligned}
\ddot{X}_2(t) &= \ddot{x}_1(t) - l_2 S_{\theta_2} {\dot{\theta}_2}^2(t) + l_2 C_{\theta_2} \ddot{\theta}_2(t) \\
\ddot{Y}_2(t) &= - l_2 C_{\theta_2} {\dot{\theta}_2}^2(t) - l_2 S_{\theta_2} \ddot{\theta}_2(t)
\end{aligned}
$$

上記の2式に代入すると、


$$
\begin{aligned}
m_1 \ddot{x}_1(t) &= f(t) - \mu_1 \dot{x}_1(t) - m_2 \left[ \ddot{x}_1(t) - l_2 S_{\theta_2} {\dot{\theta}_2}^2(t) + l_2 C_{\theta_2} \ddot{\theta}_2(t)\right] \\
(m_1 + m_2 ) \ddot{x}_1(t) + m_2 l_2 C_{\theta_2} \ddot{\theta}_2(t) &=  f(t) - \mu_1 \dot{x}_1(t) + m_2 l_2 S_{\theta_2} {\dot{\theta}_2}^2(t)
\end{aligned}
$$

$$
\begin{aligned}
I_2 \ddot{\theta}_2(t) &= m_2 g S_{\theta_2} l_2 - \mu_2 \dot{\theta_2}(t)  + m_2 l_2 \left(  - l_2 C_{\theta_2} {\dot{\theta}_2}^2(t) - l_2 S_{\theta_2} \ddot{\theta}_2(t) \right)  S_{\theta_2}\\
& - m_2 l_2 \left( \ddot{x}_1(t) - l_2 S_{\theta_2} {\dot{\theta}_2}^2(t) + l_2 C_{\theta_2} \ddot{\theta}_2(t)\right) C_{\theta_2} \\
m_2 l_2 C_{\theta_2} \ddot{x}_1(t) + I_2 \ddot{\theta}_2(t) &= m_2 l_2 S_{\theta_2}  g - \mu_2 \dot{\theta_2}(t) - m_2 {l_2}^2 S_{\theta_2} C_{\theta_2} {\dot{\theta}_2}^2(t)  -m_2 {l_2}^2 {S_{\theta_2}}^2 \ddot{\theta}_2(t) \\ & + m_2 {l_2}^2 S_{\theta_2} C_{\theta_2} {\dot{\theta}_2}^2(t) - m_2 {l_2}^2 C_{{\theta_2}}^2 \ddot{\theta}_2(t) \\
m_2 l_2 C_{\theta_2} \ddot{x}_1(t) + I_2 \ddot{\theta}_2(t) &= m_2 l_2 S_{\theta_2}  g - \mu_2 \dot{\theta_2}(t)  -m_2 {l_2}^2 \left({S_{\theta_2}}^2 +C_{{\theta_2}}^2 \right) \ddot{\theta}_2(t) \\
m_2 l_2 C_{\theta_2} \ddot{x}_1(t) + \left(I_2 + m_2 {l_2}^2 \right) \ddot{\theta}_2(t) &= m_2 l_2 S_{\theta_2}  g - \mu_2 \dot{\theta_2}(t)  \\
\end{aligned}
$$


以上より運動方程式は次となる。

$$
\boxed{
\begin{aligned}
(m_1 + m_2 ) \ddot{x}_1(t) + m_2 l_2 C_{\theta_2} \ddot{\theta}_2(t) &=  f(t) - \mu_1 \dot{x}_1(t) + m_2 l_2 S_{\theta_2} {\dot{\theta}_2}^2(t) \\
m_2 l_2 C_{\theta_2} \ddot{x}_1(t) + \left(I_2 + m_2 {l_2}^2 \right) \ddot{\theta}_2(t) &= m_2 l_2 S_{\theta_2}  g - \mu_2 \dot{\theta_2}(t)  \\
\end{aligned}
}
$$

行列形式でまとめると、以下となる。
$$
\begin{bmatrix}
m_1 + m_2 & m_2 l_2 C_{\theta_2} \\
m_2 l_2 C_{\theta_2} & I_2 + m_2 {l_2}^2
\end{bmatrix}
\begin{bmatrix}
\ddot{x}_1(t) \\ \ddot{\theta}_2(t)
\end{bmatrix} = 
\begin{bmatrix}
f(t) - \mu_1 \dot{x}_1(t) + m_2 l_2 S_{\theta_2} {\dot{\theta}_2}^2(t) \\
m_2 l_2 S_{\theta_2}  g - \mu_2 \dot{\theta_2}(t)
\end{bmatrix}
$$




### 拘束力について




$H,V$の拘束力について検討を行う。

$H,V$は次の式となる。

$$
\begin{aligned}
H(t) &= m_2 \ddot{X}_2(t) \\
V(t) &= m_2 \ddot{Y}_2(t) + m_2 g
\end{aligned} \Rightarrow
\begin{aligned}
H(t) &= m_2 \ddot{x}_1(t) - m_2 l_2  S_{\theta_2} {\dot{\theta}_2}^2(t) + m_2 l_2 C_{\theta_2} \ddot{\theta}_2(t) \\
V(t) &= - m_2 l_2 C_{\theta_2} {\dot{\theta}_2}^2(t) - m_2 l_2 S_{\theta_2} \ddot{\theta}_2(t) + m_2 g
\end{aligned}
$$

$H,V$はLink2の重心に必要な加速度を生じさせる関節反力であるため、その重心加速度を実現するように拘束力が決まる。。<br>

$H(t)$はLink1の移動に伴うLink2の並進加速度、Link2の関節点における回転運動による接線加速度と向心加速度の$H$方向成分から決定される。<br>
$V(t)$は重力、Link2の関節点における回転運動による接線加速度と向心加速度の$V$方向成分から決定される。

Link2が関節点で$\theta_2$方向に回転運動を行うときの加速度を考える。これは以下の図のように接線加速度と向心加速度の2つになる。

<img src="../images/acceleration.png" style="width:65%;" />
</p>

まず、$\theta_2$の回転を行うと、$m_2$は円弧$s$を描く。

$$
s= l_2 \theta_2
$$

これを2階微分すると、Link2の垂直方向に接線加速度 $a_t$が発生する。これを$H,V$のそれぞれの方向に成分として表すと、次のようになる。

$$
\begin{aligned}
a_{t,x} &= a_t C_{\theta_2}= l_2 \ddot{\theta}_2 C_{\theta_2} \\
a_{t,y} &= a_t S_{\theta_2}= l_2 \ddot{\theta}_2 S_{\theta_2}
\end{aligned}
$$

また、重心の運動中の軌道は回転中心から常に$l_2$の距離を保ち続けるために、回転中心側へ引っ張るような力が必要である。これは、例えばひもを付けたボールを一定の速さで回す場合、ボールの速さが一定でも、速度ベクトルの向きは常に変化する。その向きを曲げ続けるために、円の中心へ向かう加速度が必要になる。これが向心加速度 $a_c$ であり、回転中心から回転する物体までの距離を用いて次で表される。

$$
a_c = l_2 {\dot{\theta}_2}^2
$$

これを$H,V$のそれぞれの方向に成分として表すと、次のようになる。

$$
\begin{aligned}
a_{c,x} &= a_c S_{\theta_2}= l_2 {\dot{\theta}_2}^2 S_{\theta_2} \\
a_{c,y} &= a_c C_{\theta_2}= l_2 {\dot{\theta}_2}^2 C_{\theta_2}
\end{aligned}
$$

よって$H$には

- 接線加速度の成分 : $a_{t,x} = a_t C_{\theta_2}= l_2  C_{\theta_2} \ddot{\theta}_2$
- 向心加速度の成分 : $a_{c,x} = a_c S_{\theta_2}= l_2 S_{\theta_2} {\dot{\theta}_2}^2 $ \\
$

よって$V$には

- 接線加速度の成分 : $a_{t,y} = a_t S_{\theta_2}= l_2  S_{\theta_2} \ddot{\theta}_2$
- 向心加速度の成分 : $a_{c,y} = a_c C_{\theta_2}= l_2  C_{\theta_2} {\dot{\theta}_2}^2$

があることが分かる。

$V,H$の方向に合わせると、以下のように表現される。

\begin{aligned}
H(t) &= m_2 \ddot{x}_1(t) - m_2 a_{c,x} + m_2 a_{t,x} \\
V(t) &= - m_2 a_{c,y} - m_2 a_{t,y} + m_2 g
\end{aligned}
