- Link1 と Link2が、Link1の重心で、自由回転で接続されている。
- Link1の重量は$m_1$
- Link2の重量は$m_2$で、周りの慣性は$I_2$
- Link1は垂直方向($y_0$)を固定さており、水平方向($x_0$)に動作できる。
- Link2はLink2に印加される力 $f$ により回転する。


<p align="center">
<img src="../images/force_relationship.png", style="width:40%;"/>
</p>

Lnk1とLink2は回転が自由に動くように構成されているが、位置は動かないように拘束されている。
よって、接続されている点では、水平方向と鉛直方向に拘束力が発生している。これを$H,V$で表す。

作用・反作用の法則によりLink1とLink2は図のように逆方向を向ている。

$H,V$の拘束力について検討を行う。

まず、Link2が回転中心で$\theta_2$方向に回転運動を行うときの力を考える。これは結果的に以下の図のように接線加速度と向心加速度による力が生じることなる。

<p align="center">
<img src="../images/acceleration.png" style="width:85%;" />
</p>

まず、$\theta_2$の回転を行うと、$m_2$は円弧$s$を描く。

$$
s= l_2 \theta
$$

これを2階微分すると、Link2の鉛直方向に接線加速度 $a_t$が発生する。これを$H,V$のそれぞれの方向に成分として表すと、次のようになる。以降は式の簡略化のため$\sin(\theta)=S_\theta, \cos(\theta) = C_\theta$ のように表す。

$$
\begin{aligned}
a_{t,x} &= a_t C_{\theta_2}= l_2 \ddot{\theta}_2 C_{\theta_2} \\
a_{t,y} &= a_t S_{\theta_2}= l_2 \ddot{\theta}_2 S_{\theta_2}
\end{aligned}
$$

また、重心の運動中の軌道は回転中心から常に$l_2$の距離を保ち続けるために、回転中心側へ引っ張るような力が必要である。これは、例えばひもを付けたボールを一定の速さで回す場合、ボールの速さが一定でも、速度ベクトルの向きは常に変化する。その向きを曲げ続けるために、円の中心へ向かう加速度が必要になる。これが向心加速度 $a_c$ であり、回転中心から回転する物体までの距離を用いて次で表される。

$$
a_c = l_2 {\dot{\theta}_2}^2
$$

これを$H,V$のそれぞれの方向に成分として表すと、次のようになる。

$$
\begin{aligned}
a_{c,x} &= a_c S_{\theta_2}= l_2 {\dot{\theta}_2}^2 S_{\theta_2} \\
a_{c,y} &= a_c C_{\theta_2}= l_2 \dot{\theta}_2 C_{\theta_2}
\end{aligned}
$$

重力加速度 $g$ は-V方向に加わる。Link1からの運動による影響は$V$方向には生じない。よって、V方向の加速度は以下となる。

$$
-g - a_{t,y} - a_{c_y} = -g - l_2 \ddot{\theta}_2 S_{\theta_2} - l_2 \dot{\theta}_2 C_{\theta_2}
$$

Vは$y_0$方向と一致しており、この方向へ動くのは$m_2$のみである。よって運動方程式は以下となる。

$$
m_2 \ddot{Y}_2 = - m_2g  - l_2 m_2 \ddot{\theta}_2 S_{\theta_2} - l_2 m_2\dot{\theta}_2 C_{\theta_2}
$$

また、Link2の$x_0$方向に影響するのは$H$であり、この方向の加速度は以下となる。

$$
a_{t,x} - a_{c,x} =  l_2 \ddot{\theta}_2 C_{\theta_2} - l_2 {\dot{\theta}_2}^2 S_{\theta_2}
$$

よって、Link2の$x_0$方向の運動方程式は以下となる。

$$
m_2 \ddot{X}_2 = m_2 l_2 \ddot{\theta}_2 C_{\theta_2} - l_2 m_2 {\dot{\theta}_2}^2 S_{\theta_2}
$$


Link2の重心の位置 $(X_2, Y_2)$ は世界座標では次のように表される。

$$
\begin{aligned}
X_2 &= x_1 + l_2 S_{\theta_2} \\
Y_2 &= l_2 C_{\theta_2}
\end{aligned}
$$

よって、速度は

$$
\begin{aligned}
\dot{X}_2 &= \dot{x}_1 + l_2 C_{\theta_2} \dot{\theta}_2\\
\dot{Y}_2 &= - l_2 S_{\theta_2} \dot{\theta}_2
\end{aligned}
$$

加速度は以下となる。

$$
\begin{aligned}
\ddot{X}_2 &= \ddot{x}_1 - l_2 S_{\theta_2} {\dot{\theta}_2}^2 + l_2 C_{\theta_2} \ddot{\theta}_2\\
\ddot{Y}_2 &= - l_2 C_{\theta_2} {\dot{\theta}_2}^2
\end{aligned}
$$
