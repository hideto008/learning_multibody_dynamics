- Link1 と Link2が、Link1の重心で、回転は自由で、位置は一致するように拘束するように構成されている。
- Link1の質量は$m_1$
- Link2の質量は$m_2$で、重心まわりの慣性モーメントは$I_2$
- Link1は垂直方向($y_0$)を固定されており、水平方向($x_0$)に動作できる。
- 制御入力 $f$ はLink1に加えられ、Link2には直接的な制御入力や関節トルクを加えない。
- $l_2$は関節点からLink2の重心までの距離
- $\theta_2$は鉛直上向きから時計回りを正とするLink2の角度
- $\mu_1$はLink1の並進粘性係数
- $\mu_2$は関節の回転粘性係数
- $g$は重力加速度

Lagrange法は以下のエネルギーで構成されるLaguraguanを一般化座標で偏微分して運動方程式を構成する。

- 運動エネルギー $W(t)$
- 位置エネルギー $U(t)$
- 消失エネルギー $D(t)$

今回の対象とするシステムのエネルギーは次のようなものがある。

|エネルギー|並進|回転|
|:---|:---:|:---:|
|運動| $\frac{1}{2} m \dot{x}^2(t)$ | $\frac{1}{2} I \dot{\theta}^2(t)$|
|重力| $m g h(t)$ | |
|粘性による消失| $\frac{1}{2} \mu \dot{x}^2(t)$| $\frac{1}{2} \mu_{\theta} \dot{\theta}^2(t)$|

## エネルギー項

### 運動エネルギー

運動エネルギーは重心$(x_1,y_1), (X_2, Y_2)$の運動と回転 $(I_2)$ の運動により以下で構成される。
$y_1$方向は拘束されて動作しないため、この項は現れない。

$$
W(t) = \frac{1}{2}m_1 (\dot{x}_1(t))^2 + \frac{1}{2}m_2 (\dot{X}_2(t))^2 + \frac{1}{2}m_2 (\dot{Y}_2(t))^2 + \frac{1}{2}I_2 (\dot{\theta}_2(t))^2 
$$

### 位置エネルギー

位置エネルギーは重心が高さ方向($y_0$)に変化する座標で現れる。今回は重心 $m_2$ が $Y_2$ の方向に変化するため以下となる。

$$
U(t) = m_2 g Y_2(t)
$$

### 消失エネルギー

粘性項を用いて以下で構成される。

$$
D(t) = \frac{1}{2}\mu_1 (\dot{x}_1(t))^2 + \frac{1}{2} \mu_2 (\dot{\theta}_2(t))^2  
$$

ここで、$(X_2, Y_2)$ を $(x_0, y_0)$ 座標で表すと

$$
\begin{aligned}
X_2(t) &= x_1(t) + l_2 S_{\theta_2} \\
Y_2(t) &= l_2 C_{\theta_2}
\end{aligned}
$$

よって、速度は以下となる。

$$
\begin{aligned}
\dot{X}_2(t) &= \dot{x}_1(t) + l_2 C_{\theta_2} \dot{\theta}_2(t)\\
\dot{Y}_2(t) &= - l_2 S_{\theta_2} \dot{\theta}_2(t)
\end{aligned}
$$


これより、運動エネルギー $W(t)$と位置エネルギー$U(t)$は以下のようになる。

$$
\begin{aligned}
W(t) &= \frac{1}{2}m_1 (\dot{x}_1(t))^2 + \frac{1}{2}m_2 \left(\dot{x}_1(t)^2 + l_2^2 {C_{\theta_2}}^2 (\dot{\theta}_2(t))^2 + 2 \dot{x}_1(t) l_2 C_{\theta_2} \dot{\theta}_2(t)  \right) + \frac{1}{2}m_2 \left(l_2^2 {S_{\theta_2}}^2 \dot{\theta_2}(t)^2 \right) + \frac{1}{2}I_2 (\dot{\theta}_2(t))^2 \\
&= \frac{1}{2}\left(m_1+m_2 \right) (\dot{x}_1(t))^2 + \frac{1}{2}m_2 l_2^2 {C_{\theta_2}}^2 (\dot{\theta}_2(t))^2 + m_2 \dot{x}_1(t) l_2 C_{\theta_2} \dot{\theta}_2(t)  + \frac{1}{2}m_2 \left(l_2^2 {S_{\theta_2}}^2 \dot{\theta_2}(t)^2 \right) + \frac{1}{2}I_2 (\dot{\theta}_2(t))^2 \\
&= \frac{1}{2}\left(m_1+m_2 \right) (\dot{x}_1(t))^2 + \frac{1}{2}m\left( m_2 l_2^2 + I_2 \right) (\dot{\theta}_2(t))^2 + m_2  l_2 C_{\theta_2} \dot{x}_1(t) \dot{\theta}_2(t) \\
U(t) &= m_2 g {l_2}^2 {C_{\theta_2}}^2
\end{aligned}
$$

## Lagraguan $L$ と一般化座標

$$
\begin{aligned}
L(t) &= W(t) - U(t) \\
&= \frac{1}{2}\left(m_1+m_2 \right) (\dot{x}_1(t))^2 + \frac{1}{2}m\left( m_2 l_2^2 + I_2 \right) (\dot{\theta}_2(t))^2 + m_2  l_2 C_{\theta_2} \dot{x}_1(t) \dot{\theta}_2(t) - m_2 g {l_2}^2 {C_{\theta_2}}^2
\end{aligned}
$$

$$
D(t) = \frac{1}{2}\mu_1 (\dot{x}_1(t))^2 + \frac{1}{2} \mu_2 (\dot{\theta}_2(t))^2  
$$

以上より、一般化座標 $\bm{q}$ を以下のように設定する。

$$
\bm{q} = [q_1, q_2] = [x_1(t), \theta_2(t)]
$$

また、$\bm{q}$ に対応する一般化力 $\bm{f}$ を以下のように設定する。

$$
\bm{f} = [f_1, f_2] = [f(t), 0]
$$

## 運動方程式の導出

Lagraguan L , 消失エネルギー D, 一般化座標 $\bm{q}$, 一般化力 $\bm{f}$ を用いて 運動方程式を次のように導出する。

$$
\frac{d}{dt}\left(\frac{\partial L}{\partial \dot{q}_i} \right) - \frac{\partial L}{\partial q_i} + \frac{\partial D}{\partial \dot{q}_i} = f_i ,\ quad (1=1,2)
$$
