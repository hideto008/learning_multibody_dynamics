## 1. Spatial Acceleration

### 1.1. 輸送定理

World座標系$\Sigma_0$と、剛体に固定された回転座標系$\Sigma_B$があるとする。

剛体は以下の図のように角速度$\boldsymbol{\omega}$で回転しているとする。

<p align="center">
<img src="./images/transport_theorem.png" style="width:30%;" />
</p>

$\Sigma_B$の原点$O_B$から、点$P$への相対位置ベクトルを$\boldsymbol{x}$とする。このベクトルを$\Sigma_0$で表すと以下となる。

$$
{^0}\boldsymbol{x} = {^0}R_B \ {^B}\boldsymbol{x}
$$

- ${^0}\boldsymbol{x}$ : $\Sigma_0$ で表したベクトル
- ${^B}\boldsymbol{x}$ : $\Sigma_B$ で表したベクトル
- ${^0}R{_B}$ : $\Sigma_B$ から $\Sigma_0$ への回転行列

上式を時間微分すると、積の公式より以下となる。

$$
{^0}\dot{\boldsymbol{x}} = {^0}\dot{R}_B \ {^B}{\boldsymbol{x}} + {^0}R{_B} \ {^B}\dot{\boldsymbol{x}}
$$

回転行列の時間微分は以下である。

$$
{^0}\dot{R}_B = [{^0}\boldsymbol{\omega}]^\land \ {^0}R_B
$$

代入すると、以下となる。

$$
{^0}\dot{\boldsymbol{x}} = [{^0}\boldsymbol{\omega}]^\land \ {^0}R_B \ {^B}{\boldsymbol{x}} + {^0}R{_B} \ {^B}\dot{\boldsymbol{x}}
$$

これより、次のように表せる。これを輸送定理と呼ぶ。

$$
\boxed{
{^0}\dot{\boldsymbol{x}} = {^0}R{_B} \ {^B}\dot{\boldsymbol{x}} + {^0}\boldsymbol{\omega} \times {^0}{\boldsymbol{x}} 
}
$$

つまり、世界座標系で見たベクトルの時間変化${^0}\dot{\boldsymbol{x}}$は次の２項の足し合わせとなる。

- ${^0}R{_B} \ {^B}\dot{\boldsymbol{x}}$ : 回転座標系内での変化を世界座標系への変換
- ${^0}\boldsymbol{\omega} \times {^0}{\boldsymbol{x}}$ : 座標系の回転による変化

上記のとおり、輸送定理には、ベクトルの原点$O_B$の並進速度は含まれない。なぜなら、輸送定理はベクトルの向きや大きさの時間変化を扱う定理であり、ベクトルの原点が移動しても、そのベクトル自体は変化しないからである。

例えば、剛体が回転せずに、右方向へ速度2m/sで移動しているとする。<br>
このとき、$\boldsymbol{\omega} = 0$ であり、さらに$\boldsymbol{x}$が剛体固定された相対位置ベクトルなら、${^B}\dot{\boldsymbol{x}}=0$ である。<br>
輸送定理より、以下となる。

$$
{^0}\dot{\boldsymbol{x}} = 0
$$

一方、点$P$の世界座標系$\Sigma_0$における位置を時間微分する場合、$\Sigma_B$の原点$O_B$の並進速度も考慮する必要がある。<br>
$\Sigma_0$の原点$O_0$から$\Sigma_B$の原点$O_B$までを位置ベクトル$\boldsymbol{r}_{O_O O_B}$ とする。これを用いて点$P$を世界座標系で示すと、次である。

$$
{^0}\boldsymbol{p}_P = {^0}\boldsymbol{r}_{O_O O_B} + {^0}R{_B} {^B}\boldsymbol{x}
$$

この時間微分は以下となる。

$$
{^0}\dot{\boldsymbol{p}}_P = {^0}\dot{\boldsymbol{r}}_{O_O O_B} + {^0}\dot{R}{_B} {^B}\boldsymbol{x} + {^0}R{_B} {^B}\dot{\boldsymbol{x}}
$$

回転行列の時間微分は以下である。

$$
{^0}\dot{R}_B = [{^0}\boldsymbol{\omega}]^\land \ {^0}R_B
$$

そして、${^0}\dot{\boldsymbol{p}}_P = {^0}{\boldsymbol{v}}_P, {^0}\dot{\boldsymbol{r}}_{O_O O_B} = {^0}{\boldsymbol{v}}_{O_B}$ とすると、以下となる。

$$
{^0}{\boldsymbol{v}}_P = {^0}{\boldsymbol{v}}_{O_B} + {^0}\boldsymbol{\omega} \times {^0}\boldsymbol{x} + {^0}R{_B} {^B}\dot{\boldsymbol{x}}
$$


### 1.2. Spaital Accelration

Spatial Motion Vector の時間微分が Spatial Acceleration である。

剛体に固定された座標系 $\Sigma_B$ の原点を$O_B$とする。

Spatial Motion Vectorは以下は、剛体の角速度${^B}\boldsymbol{\omega}$ と 座標系原点$O$の並進速度${^B}\boldsymbol{v}_{O_B}$を用いると以下となる。

$$
{^B}\boldsymbol{V}_{O_B} = \begin{bmatrix} {^B}\boldsymbol{\omega} \\ {^B}\boldsymbol{v}_{O_B} \end{bmatrix}
$$

Spatial Motion Vectorの両成分は $\Sigma_B$ で表されている。これを時間微分する。

#### 1.2.1. 角速度の時間微分

${^B}\boldsymbol{\omega}$ を$\Sigma_0$で表すと以下となる。

$$
{^0}\boldsymbol{\omega} = {^0}R{_B}{^B}\boldsymbol{\omega}
$$

この時間微分として輸送定理を適用すると、以下となる。

$$
{^0}\dot{\boldsymbol{\omega}} = {^0}R{_B}{^B}\dot{\boldsymbol{\omega}} + {^0}\boldsymbol{\omega} \times {^0}\boldsymbol{\omega}
$$

しかし、${^0}\boldsymbol{\omega} \times {^0}\boldsymbol{\omega} = 0$ となるため、以下となる。

$$
{^0}\dot{\boldsymbol{\omega}} = {^0}R{_B}{^B}\dot{\boldsymbol{\omega}}
$$

よって、角速度${^B}\boldsymbol{\omega}$の時間微分である角加速度を以下で表す。

$$
\boxed{
{^B}\dot{\boldsymbol{\omega}} = {^B}R{_0}{^0}\dot{\boldsymbol{\omega}} = {^B}\dot{\boldsymbol{\omega}} 
}
$$

#### 1.2.2. 並進速度の時間微分

${^B}\boldsymbol{v}_{O_B}$ を $\Sigma_0$ で表すと以下である。

$$
{^0}\boldsymbol{v}_{O_B} = {^0}R{_B}{^B}\boldsymbol{v}_{O_B}
$$

角速度と同じく、輸送定理を適用すると以下となる。

$$
{^0}\dot{\boldsymbol{v}}_{O_B} = {^0}R{_B}{^B}\dot{\boldsymbol{v}}_{O_B} + {^0}\boldsymbol{\omega} \times {^0}\boldsymbol{v}_{O_B}
$$

左辺は$\Sigma_0$で表した$\Sigma_B$の原点$O_B$の並進加速度であるため、以下で記述する。

$$
{^0}{\boldsymbol{a}}_{O_B} = {^0}\dot{\boldsymbol{v}}_{O_B}
$$

$$
{^0}{\boldsymbol{a}}_{O_B} = {^0}R{_B}{^B}\dot{\boldsymbol{v}}_{O_B} + {^0}\boldsymbol{\omega} \times {^0}\boldsymbol{v}_{O_B}
$$

両辺に${^B}R{_0}$を掛けて$\Sigma_B$へ座標変換すると、

$$
\begin{aligned}
{^B}R{_0}{^0}{\boldsymbol{a}}_{O_B} &= {^B}R{_0}{^0}R{_B}{^B}\dot{\boldsymbol{v}}_{O_B} + {^B}R{_0}{^0} \left(\boldsymbol{\omega} \times {^0}\boldsymbol{v}_{O_B} \right) \\
{^B}{\boldsymbol{a}}_{O_B} &= {^B}\dot{\boldsymbol{v}}_{O_B} + {^B}R{_0}{^0}\boldsymbol{\omega} \times {^B}R{_0}{^0}\boldsymbol{v}_{O_B} \\
{^B}{\boldsymbol{a}}_{O_B} &= {^B}\dot{\boldsymbol{v}}_{O_B} + {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_{O_B} \\
\end{aligned}
$$

よって、${^B}\boldsymbol{v}_{O_B}$ の時間微分は以下となる。

$$
\boxed{
{^B}\dot{\boldsymbol{v}}_{O_B} = {^B}{\boldsymbol{a}}_{O_B} - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_{O_B}
}
$$

### 1.2.3. Spaital Accelration の導出

Spatial Motion Vector の時間微分をまとめると以下となる。

$$
{^B}\dot{\boldsymbol{V}}_O = \begin{bmatrix} {^B}\dot{\boldsymbol{\omega}} \\ {^B}\dot{\boldsymbol{v}}_{O_B} \end{bmatrix} = 
\begin{bmatrix}
{^B}{\boldsymbol{\alpha}}  \\
{^B}{\boldsymbol{a}}_{O_B} - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_{O_B}
\end{bmatrix}
$$

これを Spatial Accelerationと呼ぶ。

$$
\boxed{
{^B}\boldsymbol{A}_{O_B} = \begin{bmatrix}
{^B}{\boldsymbol{\alpha}}  \\
{^B}{\boldsymbol{a}}_{O_B} - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_{O_B}
\end{bmatrix}
}
$$

### 1.2.4. Spatial Transform (基準点変更)

$\Sigma_B$の原点$O_B$の速度を重心$G$基準で示すと以下となる。

$$
{^B}\boldsymbol{v}_{O_B} = {^B}\boldsymbol{v}_G + {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{r}_{G O_B}
$$

上式の両辺を慣性座標系で時間微分し、得られた加速度を$\Sigma_B$で表す。物体がある角速度$\boldsymbol{\omega}$で回転しているとき、位置ベクトルの時間微分は$\boldsymbol{\omega}$との外積で計算できるため、以下となる。

$$
{^B}\boldsymbol{a}_{O_B} = {^B}\boldsymbol{a}_G + {^B}\dot{\boldsymbol{\omega}} \times {^B}\boldsymbol{r}_{G O_B} + {^B}\boldsymbol{\omega} \times ({^B}\boldsymbol{\omega} \times {^B}\boldsymbol{r}_{G O_B}) 
$$

これらをSpatial Accelertaionの並進加速度成分に代入する。

$$
\begin{aligned}
{^B}{\boldsymbol{a}}_{O_B} - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_{O_B} &= {^B}\boldsymbol{a}_G + {^B}\dot{\boldsymbol{\omega}} \times {^B}\boldsymbol{r}_{G O_B} \\ 
& \space + {^B}\boldsymbol{\omega} \times ({^B}\boldsymbol{\omega} \times {^B}\boldsymbol{r}_{G O_B}) \\
& \space - {^B}\boldsymbol{\omega} \times ({^B}\boldsymbol{v}_G + {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{r}_{G O_B} ) \\
&= {^B}\boldsymbol{a}_G + {^B}\dot{\boldsymbol{\omega}} \times {^B}\boldsymbol{r}_{G O_B} - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_G \\
&= {^B}\boldsymbol{a}_G - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_G - [{^B}\boldsymbol{r}_{G O_B}]^\land {^B}{\boldsymbol{\alpha}} 
\end{aligned}
$$

重心$G$におけるSpatial Accelerationを以下とすると、

$$
{^B}\boldsymbol{A}_{G} = \begin{bmatrix}
{^B}{\boldsymbol{\alpha}}  \\
{^B}{\boldsymbol{a}}_{G} - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_{G}
\end{bmatrix}
$$

重心$G$から$\Sigma_B$の原点$O_B$への基準点変更は以下となる。

$$
{^B}\boldsymbol{A}_{O_B} = \begin{bmatrix} \boldsymbol{I}_3 & 0 \\ - [{^B}\boldsymbol{r}_{G O_B}]^\land & \boldsymbol{I}_3 \end{bmatrix} {^B}\boldsymbol{A}_{G}
$$



### 1.2.5. Motion Vector, Force Vector との違い

Spatial Motion Vectorを$\Sigma_{B}$の原点$O_B$を示すと以下となる。

$$
{^B}\boldsymbol{V}{_{O_B}} = \begin{bmatrix} \boldsymbol{I}_3 & 0 \\ -[\boldsymbol{r}_{G O_B}]^\land & \boldsymbol{I} \end{bmatrix} {^B}\boldsymbol{V}{_{G}} , \quad {^B}\boldsymbol{V}{_{G}} =\begin{bmatrix} {^B}\boldsymbol{\omega} \\ {^B}\boldsymbol{v}_G \end{bmatrix}
$$

Spatial Force Vectorを$\Sigma_{B}$の原点$O_B$で示すと以下となる。

$$
{^B}\boldsymbol{f}{_{O_B}} = \begin{bmatrix} \boldsymbol{I}_3 & -[\boldsymbol{r}_{G O_B}]^\land \\ 0 & \boldsymbol{I} \end{bmatrix} {^B}\boldsymbol{f}{_{G}} , \quad {^B}\boldsymbol{f}{_{G}} = \begin{bmatrix} {^B}\boldsymbol{n}_G \\ {^B}\boldsymbol{F}_G\end{bmatrix}=\begin{bmatrix} {^B}\boldsymbol{I}_G {^B}\dot{\boldsymbol{\omega}}+ {^B}\boldsymbol{\omega} \times  ({^B}\boldsymbol{I}_G {^B}\boldsymbol{\omega}) \\ m{^B}\boldsymbol{a}_G \end{bmatrix}
$$

これより、${^B}\boldsymbol{V}{_{G}}, {^B}\boldsymbol{I}_G, m$ が分かれば、${^B}\boldsymbol{f}_G$と${^B}\boldsymbol{A}_G$が分かり。さらに${^B}\boldsymbol{r}_{G O_B}$が分かると、${^B}\boldsymbol{V}_{O_B}, {^B}\boldsymbol{f}_{O_B}$と${^B}\boldsymbol{A}_{O_B}$が分かる。

もちろんセンサなどにより直接計測すれば、${^B}\boldsymbol{V}{_{G}}, ${^B}\boldsymbol{f}{_{G}}, ${^B}\boldsymbol{A}{_{G}}$ を求めることはできる。
