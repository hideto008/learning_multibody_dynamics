# Spatial Vector

## 1. Spatial Motion Vector

二つの座標系$\Sigma_0, \Sigma_1$があり、$\Sigma_1$の原点は$\Sigma_0$の原点から${^0}p_1$の位置にあり、その姿勢は$\theta_1$となっている。

<img src="./images/position.png" style="width:30%;" />
</p>

$\Sigma_1$ 上にある点$P$の位置を$\Sigma_0$とした表すと次のようになる。$P$は$\Sigma_1$の原点から${^1}r_1$の位置にある。

$$
{^0}P = {^0}p_1 + {^0}R_1 {^1}r_1
$$

${^0}R_1$ は$\Sigma_0$における$\Sigma_1$の姿勢を示す回転行列である。

上式を時間微分し、点$P$の速度を求める。

$$
{^0}\dot{P} = {^0}\dot{p}_1 + {^0}\dot{R}_1 {^1}r_1 + {^0}R_1 {^1}\dot{r}_1
$$

このとき座標系 $\Sigma_1$は動くか、${^1}r_1$は動かないとすると、${^1}\dot{r}_1 = 0$となるため、

$$
{^0}\dot{P} = {^0}\dot{p}_1 + {^0}\dot{R}_1 {^1}r_1
$$

回転行列の時間微分は、角速度ベクトルを用いて ${^0}\dot{R}_1 = [{^0}\omega_{1}]^\land \ {^0}{R}_1$ と表せるため、

$$
{^0}\dot{P} = {^0}\dot{p}_1 + [{^0}\omega_{1}]^\land \ {^0}{R}_1 {^1}r_1 = {^0}\dot{p}_1 + [{^0}\omega_{1}]^\land {^0}r_1 
$$

${^0}\dot{P} = {^0}v_{P}$、${^0}\dot{p}_1 = {^0}v_O$、${^0}r_1 = {^0}r_{OP}$ と記述すると次となる。

$$
{^0}v_{P} = {^0}v_O + {^0}\omega_{1} \times {^0}r_{OP}
$$

- ${^0}v_O$ : $\Sigma_1$の原点$O$の並進運動
- ${^0}\omega_{1} \times {^0}r_{OP}$ : $\Sigma_1$の原点$O$まわりの回転による運動
