# Spatial Vector (空間ベクトル)

このNoteは以下の内容についての説明である。

1. Spatial Motion Vector (空間運動ベクトル)
2. Spatial Transform (空間変換)
3. Spatial Force Vector (空間力ベクトル)
4. Spatial Inertia (空間慣性行列)

## 1. Spatial Motion Vector (空間運動ベクトル)

二つの座標系$\Sigma_0, \Sigma_1$があり、$\Sigma_1$の原点$O_1$は$\Sigma_0$の原点$O_0$から${^0}\boldsymbol{p}_1$の位置にあり、その姿勢は$\theta_1$となっている。<br>
$\Sigma_1$は剛体上に定義された座標系であり、剛体とともに移動する。$\Sigma_0$はWorld座標とする。

<p align="center">
<img src="./images/position.png" style="width:30%;" />
</p>

$\Sigma_1$ 上にある点$P$の位置を$\Sigma_0$で表すと次のようになる。$P$は$\Sigma_1$の原点$O_1$から${^1}\boldsymbol{r}_1$の位置にある。

$$
{^0}\boldsymbol{p}_{P} = {^0}\boldsymbol{p}_1 + {^0}R_1 {^1}\boldsymbol{r}_1
$$

${^0}R_1$ は$\Sigma_0$における$\Sigma_1$の姿勢を示す回転行列である。

上式を時間微分し、点$P$の速度を求める。

$$
{^0}\dot{\boldsymbol{p}}_{P} = {^0}\dot{\boldsymbol{p}}_1 + {^0}\dot{R}_1 {^1}\boldsymbol{r}_1 + {^0}R_1 {^1}\dot{\boldsymbol{r}}_1
$$

点$P$は座標系 $\Sigma_1$に固定された剛体上の点であり、${^1}\boldsymbol{r}_1$は時間変化しない。すると${^1}\dot{\boldsymbol{r}}_1 = 0$となるため、

$$
{^0}\dot{\boldsymbol{p}}_{P} = {^0}\dot{\boldsymbol{p}}_1 + {^0}\dot{R}_1 {^1}\boldsymbol{r}_1
$$

回転行列の時間微分は、角速度ベクトルを用いて ${^0}\dot{R}_1 = [{^0}\boldsymbol{\omega}_{1}]^\land \ {^0}{R}_1$ と表せるため、

$$
{^0}\dot{\boldsymbol{p}}_P = {^0}\dot{\boldsymbol{p}}_1 + [{^0}\boldsymbol{\omega}_{1}]^\land \ {^0}{R}_1 {^1}\boldsymbol{r}_1 = {^0}\dot{\boldsymbol{p}}_1 + [{^0}\boldsymbol{\omega}_{1}]^\land {^0}\boldsymbol{r}_1 
$$

${^0}\dot{\boldsymbol{p}}_P = {^0}\boldsymbol{v}_{P}$、${^0}\dot{\boldsymbol{p}}_1 = {^0}\boldsymbol{v}_{O_1}$、${^0}\boldsymbol{r}_1 = {^0}\boldsymbol{r}_{O_1P}$ と記述すると次となる。

$$
{^0}\boldsymbol{v}_{P} = {^0}\boldsymbol{v}_{O_1} + {^0}\boldsymbol{\omega}_{1} \times {^0}\boldsymbol{r}_{O_1P}
$$

- ${^0}\boldsymbol{v}_{O_1}$ : $\Sigma_1$の原点$O_1$の並進運動
- ${^0}\boldsymbol{\omega}_{1} \times {^0}\boldsymbol{r}_{O_1P}$ : $\Sigma_1$の原点$O_1$まわりの回転による運動

<p align="center">
<img src="./images/velocity.png" style="width:30%;" />
</p>

剛体の角速度は剛体全の回転運動を表す量であるため、基準点に依存しない。よって点$O_1$と点$P$で同じ角速度を使用できる。

$$
{^0}\boldsymbol{\omega}_P = {^0}\boldsymbol{\omega}_1
$$

まとめると以下の行列形式となる。

$$
\begin{bmatrix}
{^0}\boldsymbol{\omega}_P \\ {^0}\boldsymbol{v}_{P}
\end{bmatrix} =
\begin{bmatrix}
{^0}\boldsymbol{\omega}_1 \\
{^0}\boldsymbol{v}_{O_1} + {^0}\boldsymbol{\omega}_{1} \times {^0}\boldsymbol{r}_{O_1P}
\end{bmatrix}
$$

外積は$A \times B = - B \times A$ であるため、次のように変換を行う。

$$
{^0}\boldsymbol{\omega}_{1} \times {^0}\boldsymbol{r}_{O_1P} = - {^0}\boldsymbol{r}_{O_1P} \times {^0}\boldsymbol{\omega}_{1} = - [{^0}\boldsymbol{r}_{O_1P}]^\land \ {^0}\boldsymbol{\omega}_{1}
$$

すると、剛体上の基準点を$O_1$から$P$に変更したときの角速度と並進速度の関係を次の行列形式で表すことが出来る。

$$
\begin{bmatrix}
{^0}\boldsymbol{\omega}_P \\ {^0}\boldsymbol{v}_{P}
\end{bmatrix} =
\begin{bmatrix}
\boldsymbol{I}_3 & 0 \\
- [{^0}\boldsymbol{r}_{O_1P}]^\land &  \boldsymbol{I}_3
\end{bmatrix}
\begin{bmatrix}
{^0}\boldsymbol{\omega}_1 \\ {^0}\boldsymbol{v}_{O_1}
\end{bmatrix}
$$


#### Local座標系での表現

上記表現はWorld座標系上での基準点の変更であった。

角速度と速度を$\boldsymbol{V}$とする。これをSpatial Motion Vector と呼ぶ。

$$
\boldsymbol{V} = \begin{bmatrix} \boldsymbol{\omega} \\ \boldsymbol{v} \end{bmatrix} \in \mathbf{R}^{6 \times 1}
$$

$\boldsymbol{V}$ を用いて以下のように基準点の変更を表すことが出来る。$\boldsymbol{X}_{P \leftarrow O_1}^{(0)}$は$(0)$で表現する座標系を示し、$P \leftarrow O_1$で基準点を$O_1$から$P$へ変更することを示している。

$$
{^0}\boldsymbol{V}{_P} = \boldsymbol{X}_{P \leftarrow O_1}^{(0)} {^0}\boldsymbol{V}{_{O_1}}
$$

ABAでは各リンクのローカル座標系で演算するため、${^1}\boldsymbol{V}{_P}$ を求める。

##### 角速度

角速度は回転行列を用いて次の座標変換となる。

$$
{^1}\boldsymbol{\omega}_{1} = ({^0}R_{1})^T {^0}\boldsymbol{\omega}_{1} ,\quad {^1}\boldsymbol{\omega}_{P} = ({^0}R_{1})^T {^0}\boldsymbol{\omega}_{P} 
$$

##### 速度

$\Sigma_0$による表現は以下である。

$$
{^0}\boldsymbol{v}_{P} = {^0}\boldsymbol{v}_{O_1} + {^0}\boldsymbol{\omega}_{1} \times {^0}\boldsymbol{r}_{O_1P}
$$

上式の両辺に$({^0}R{_1})^T$を掛けると、以下となる。

$$
\begin{aligned}
({^0}R{_1})^T \ {^0}\boldsymbol{v}_{P} &= ({^0}R{_1})^T \ {^0}\boldsymbol{v}_{O_1} + ({^0}R{_1})^T \left({^0}\boldsymbol{\omega}_{1} \times {^0}\boldsymbol{r}_{O_1P} \right) \\
&= ({^0}R{_1})^T \ {^0}\boldsymbol{v}_{O_1} +  \left(({^0}R{_1})^T \ {^0}\boldsymbol{\omega}_{1} \right)  \times \left( ({^0}R{_1})^T \ {^0}\boldsymbol{r}_{O_1P} \right)  \\
{^1}\boldsymbol{v}_{P} &= {^1}\boldsymbol{v}_{O_1} + {^1}\boldsymbol{\omega}_{1} \times {^1}\boldsymbol{r}_{O_1P} \\
&= {^1}\boldsymbol{v}_{O_1} - {^1}\boldsymbol{r}_{O_1P} \times {^1}\boldsymbol{\omega}_{1}
\end{aligned}
$$

すると、Local座標において、剛体上の基準点を$O_1$から$P$に変更したときの角速度と並進速度の関係を次の行列形式で表すことが出来る。

$$
\begin{bmatrix}
{^1}\boldsymbol{\omega}_P \\ {^1}\boldsymbol{v}_{P}
\end{bmatrix} =
\begin{bmatrix}
\boldsymbol{I}_3 & 0 \\
- [{^1}\boldsymbol{r}_{O_1P}]^\land &  \boldsymbol{I}_3
\end{bmatrix}
\begin{bmatrix}
{^1}\boldsymbol{\omega}_1 \\ {^1}\boldsymbol{v}_{O_1}
\end{bmatrix}
$$

Spatial Motion Vectorを用いると次のように記述できる。

$$
{^1}\boldsymbol{V}{_P} = \boldsymbol{X}_{P \leftarrow O_1}^{(1)} {^1}\boldsymbol{V}{_{O_1}}
$$


## 2. Spatial Transform (空間変換)

World座標$\Sigma_0$からLocal座標$\Sigma_1$への座標変換を行うと、

$$
{^1}\boldsymbol{\omega}_{1} = {^1}R_0 \ {^0}\boldsymbol{\omega}_{1} \\
{^1}\boldsymbol{v}_{O_1} = {^1}R_0 \ {^0}\boldsymbol{v}_{O_1}
$$

より、座標系原点のSpatial Motion Vectorの座標変換は以下となる。

$$
\begin{bmatrix}
{^1}\boldsymbol{\omega}_{1} \\ {^1}\boldsymbol{v}_{O_1}
\end{bmatrix} = 
\begin{bmatrix}
{^1}R_0 & 0 \\
0 & {^1}R_0
\end{bmatrix}
\begin{bmatrix}
{^0}\boldsymbol{\omega}_{1} \\ {^0}\boldsymbol{v}_{O_1}
\end{bmatrix}  \quad \rightarrow \quad
{^1}\boldsymbol{V}{_{O_1}} = 
\begin{bmatrix}
{^1}R_0 & 0 \\
0 & {^1}R_0
\end{bmatrix}
{^0}\boldsymbol{V}{_{O_1}} 
$$

基準点の変更は以下であったので、上式を代入すると

$$
\begin{aligned}
{^1}\boldsymbol{V}{_P} &=
\begin{bmatrix}
\boldsymbol{I}_3 & 0 \\
- [{^1}\boldsymbol{r}_{O_1P}]^\land &  \boldsymbol{I}_3
\end{bmatrix}
{^1}\boldsymbol{V}{_{O_1}} =
\begin{bmatrix}
\boldsymbol{I}_3 & 0 \\
- [{^1}\boldsymbol{r}_{O_1P}]^\land &  \boldsymbol{I}_3
\end{bmatrix}
\begin{bmatrix}
{^1}R_0 & 0 \\
0 & {^1}R_0
\end{bmatrix}
{^0}\boldsymbol{V}{_{O_1}} \\
&= \begin{bmatrix}
{^1}R_0 & 0 \\
- [{^1}\boldsymbol{r}_{O_1P}]^\land \ {^1}R_0 &  {^1}R_0
\end{bmatrix}
{^0}\boldsymbol{V}{_{O_1}}
\end{aligned}
$$

このSpatial Transform は剛体の原点$O_1$におけるSpatial Motion Vectorを、表現座標系を$\Sigma_0$から$\Sigma_1$に変更するとともに、基準点を$O_1$から$P$に変更するものである。



## 3. Spatial Force Vector (空間力ベクトル)

剛体に作用する力とモーメントはそれぞれ以下のように表現できる。

- 並進運動

$$
\boldsymbol{F} = m \boldsymbol{a}_G
$$

- 回転運動

$$
\boldsymbol{n}_G = \boldsymbol{I}_G \dot{\boldsymbol{\omega}} + \boldsymbol{\omega} \times (\boldsymbol{I}_G \boldsymbol{\omega})
$$

表現する座標系が統一されていれば2式はそのまま使用することが出来る。

Spatial Force Vectorは、力とモーメントを1つの6次元ベクトルとして表現する。

$$
\boldsymbol{f}_O = \begin{bmatrix} \boldsymbol{n}_O \\ \boldsymbol{F} \end{bmatrix} \in \mathbf{R}^{6 \times 1}
$$

- $\boldsymbol{n}_O$ : 剛体の点$O$周りのモーメント
- $\boldsymbol{F}$ : 剛体に作用する合力

### 3.1. モーメントの基準点変更

剛体上に点$P$と点$Q$があり、点$Q$に力$\boldsymbol{F}$が作用しているとする。

<p align="center">
<img src="./images/force.png" style="width:30%;" />
</p>



点$O_1$まわりのモーメントは次である。

$$
\boldsymbol{n}_{O_1} = \boldsymbol{r}_{O_1 Q} \times \boldsymbol{F}
$$

点$P$まわりのモーメントは次で表せる。

$$
\boldsymbol{n}_{P} = \boldsymbol{r}_{P Q} \times \boldsymbol{F}
$$

位置ベクトル$\boldsymbol{r}_{O_1 Q}$は以下の関係がある。

$$
\boldsymbol{r}_{O_1 Q} = \boldsymbol{r}_{O_1 P} + \boldsymbol{r}_{P Q} 
$$

代入すると、

$$
\begin{aligned}
\boldsymbol{n}_{O_1} &= (\boldsymbol{r}_{O_1 P} + \boldsymbol{r}_{P Q} ) \times \boldsymbol{F} = \boldsymbol{r}_{O_1 P} \times \boldsymbol{F} + \boldsymbol{r}_{P Q} \times \boldsymbol{F} \\
&= \boldsymbol{r}_{O_1 P} \times \boldsymbol{F} + \boldsymbol{n}_{P}
\end{aligned}
$$

これより、モーメントの基準点変更の式は以下となる。

$$
\boldsymbol{n}_{P} = \boldsymbol{n}_{O_1} -   \boldsymbol{r}_{O_1 P} \times \boldsymbol{F} 
$$

剛体に偶力モーメント$\tau$が加わっている場合、そのモーメントはどの基準でも同じ値になるため以下の2式となる。

$$
\boldsymbol{N}_{O_1} = \boldsymbol{r}_{O_1 Q} \times \boldsymbol{F} + \tau
$$

$$
\boldsymbol{N}_{P} = \boldsymbol{r}_{P Q} \times \boldsymbol{F} + \tau
$$

位置ベクトル$\boldsymbol{r}_{O_1 Q}$の関係を用いると、

$$
\boldsymbol{r}_{O_1 Q} = \boldsymbol{r}_{O_1 P} + \boldsymbol{r}_{P Q} 
$$

より

$$
\begin{aligned}
\boldsymbol{N}_{O_1} &= \left(\boldsymbol{r}_{O_1 P} + \boldsymbol{r}_{P Q} \right)  \times \boldsymbol{F} + \tau \\
&= \boldsymbol{r}_{O_1 P} \times \boldsymbol{F} + \boldsymbol{r}_{P Q} \times \boldsymbol{F} + \tau \\
&= \boldsymbol{r}_{O_1 P} \times \boldsymbol{F} + \boldsymbol{N}_{P}
\end{aligned}
$$

よって、偶力モーメントが加わっていても、モーメントの基準点変更の式は同じ形式となる。

$$
\boldsymbol{N}_{P} = \boldsymbol{N}_{O_1} -  \boldsymbol{r}_{O_1 P} \times \boldsymbol{F}
$$





### 3.2 Spatial Force Vectorの行列形式

モーメントの基準点変更の式の全てベクトルをWorld座標系$\Sigma_0$で表すと以下となる。

$$
{^0}\boldsymbol{n}_{P} = {^0}\boldsymbol{n}_{O_1} -   {^0}\boldsymbol{r}_{O_1 P} \times {^0}\boldsymbol{F} 
$$

合力$\boldsymbol{F}$は基準点を変更しても変化しないため、以下となる。

$$
{^0}\boldsymbol{F}_P = {^0}\boldsymbol{F}_{O_1}
$$

モーメントと合力の二つの式をまとめると以下となる。

$$
\begin{bmatrix}
{^0}\boldsymbol{n}_{P} \\ {^0}\boldsymbol{F}_P
\end{bmatrix} =
\begin{bmatrix}
{^0}\boldsymbol{n}_{O_1} - [{^0}\boldsymbol{r}_{O_1 P}]^\land \ {^0}\boldsymbol{F} \\
{^0}\boldsymbol{F}_{O_1}
\end{bmatrix} = 
\begin{bmatrix}
\boldsymbol{I}_3 & - [{^0}\boldsymbol{r}_{O_1 P}]^\land \\
0 & \boldsymbol{I}_3
\end{bmatrix}
\begin{bmatrix}
{^0}\boldsymbol{n}_{O_1} \\ {^0}\boldsymbol{F}_{O_1}
\end{bmatrix}
$$

Spatial Force Vectorを用いると、次で記述できる。

$$
{^0}\boldsymbol{f}_P =
X_{F, P \leftarrow O_1}^{(0)}
{^0}\boldsymbol{f}_{O_1}
$$


### 3.3 Spatial Motion Vectorとの関連

Motion Vectorの基準点の変更式は、以下であった。

$$
\begin{bmatrix}
\boldsymbol{\omega} \\
\boldsymbol{v}_P
\end{bmatrix} =
\begin{bmatrix}
\boldsymbol{I}_3 & 0 \\
-[\boldsymbol{r}]^\land & \boldsymbol{I}_3
\end{bmatrix}
\begin{bmatrix}
\boldsymbol{\omega} \\
\boldsymbol{v}_{O_1}
\end{bmatrix}
$$

Force Vectorの基準点の変更式は以下である。

$$
\begin{bmatrix}
\boldsymbol{n}_P \\
\boldsymbol{F}
\end{bmatrix} =
\begin{bmatrix}
\boldsymbol{I}_3 & -[\boldsymbol{r}]^\land \\
0  & \boldsymbol{I}_3
\end{bmatrix}
\begin{bmatrix}
\boldsymbol{n}_{O_1} \\
\boldsymbol{F} \\
\end{bmatrix}
$$

ここで $\boldsymbol{r} = \boldsymbol{r}_{O_1 P}$ である。

Motion Vectorの変換行列を$X$として表す。

$$
X = \begin{bmatrix}
\boldsymbol{I}_3 & 0 \\
-[\boldsymbol{r}]^\land & \boldsymbol{I}_3
\end{bmatrix}
$$

$X$の逆行列は以下となる。

$$
X^{-1} = \begin{bmatrix}
\boldsymbol{I}_3 & 0 \\
[\boldsymbol{r}]^\land & \boldsymbol{I}_3
\end{bmatrix}
$$

この逆行列を転置する。外積行列は歪対称行列であるため、$([\boldsymbol{r}]^\land)^T = - [\boldsymbol{r}]^\land$ となるため、

$$
X^{-T} = \begin{bmatrix}
\boldsymbol{I}_3 & -[\boldsymbol{r}]^\land \\
0 & \boldsymbol{I}_3
\end{bmatrix}
$$

Force Vectorの変換行列と一致する。

Spatial Motion VectorとSpatial Force Vectorの内積は、剛体に作用する力の仕事率を表す。

$$
\boldsymbol{f}^T \boldsymbol{V} = \boldsymbol{n}^T \boldsymbol{\omega} + \boldsymbol{F}^T \boldsymbol{v}
$$

仕事率は基準点を変更しても変化しない。

$$
\boldsymbol{f}_P^T \boldsymbol{V}_P = \boldsymbol{f}_{O_1}^T \boldsymbol{V}_{O_1}
$$

Motion Vectorが $\boldsymbol{V}_P = X \boldsymbol{V}_{O_1}$ と変換されるなら、Force Vectorは $\boldsymbol{f}_P = X^{-T} \boldsymbol{f}_{O_1}$ である必要がある。これがForce Vecotorの変換行列がMotion Vectorの変換行列の逆行列の転置となる理由である。

$$
\boldsymbol{f}_P^T \boldsymbol{V}_P = \left(X^{-T}\boldsymbol{f}_{O_1}\right)^T X \boldsymbol{V}_{O_1} = \boldsymbol{f}_{O_1}^T X^{-1} X \boldsymbol{V}_{O_1} = \boldsymbol{f}_{O_1}^T \boldsymbol{V}_{O_1}
$$
$$

### 3.4. Local座標系での表現とSpatial Transform

基準点$O_1$の座標系をWorld座標からLocal座標へ回転行列を用いて変換すると、モーメントと力は以下となる。

$$
{^1} \boldsymbol{n}_{O_1} = {^1}R_0 \ {^0} \boldsymbol{n}_{O_1} \\
{^1} \boldsymbol{F}_{O_1} = {^1}R_0 \ {^0} \boldsymbol{F}_{O_1} \\
$$

これをSpatial Force Vector で行列形式で表すと、以下となる。

$$
\begin{bmatrix}
{^1} \boldsymbol{n}_{O_1} \\ {^1} \boldsymbol{F}_{O_1} 
\end{bmatrix} =
\begin{bmatrix}
{^1}R_0 & 0 \\
0 & {^1}R_0
\end{bmatrix}
\begin{bmatrix}
{^0} \boldsymbol{n}_{O_1} \\ {^0} \boldsymbol{F}_{O_1} 
\end{bmatrix} \quad \rightarrow \quad
{^1}\boldsymbol{f}_{O_1} =
\begin{bmatrix}
{^1}R_0 & 0 \\
0 & {^1}R_0
\end{bmatrix}
{^0}\boldsymbol{f}_{O_1}
$$

Local座標系で表現したSpatial Force Vectorの基準点変更は以下である。

$$
{^1}\boldsymbol{f}_{P} = 
\begin{bmatrix}
\boldsymbol{I}_3 & - [{^1}\boldsymbol{r}_{O_1 P}]^\land \\
0 & \boldsymbol{I}_3
\end{bmatrix}
{^1}\boldsymbol{f}_{O_1}
$$

これに上式を適用すると、表現をWorld座標系からLocal座標系へ変更し、基準点を$O_1$から$P$へ変換する式となる。これがSpatial Force Vectorにおける Spatial Transform である。

$$
\begin{aligned}
{^1}\boldsymbol{f}_{P} &= 
\begin{bmatrix}
\boldsymbol{I}_3 & - [{^1}\boldsymbol{r}_{O_1 P}]^\land \\
0 & \boldsymbol{I}_3
\end{bmatrix}
\begin{bmatrix}
{^1}R_0 & 0 \\
0 & {^1}R_0
\end{bmatrix}
{^0}\boldsymbol{f}_{O_1} \\
&= \begin{bmatrix}
{^1}R_0 & - [{^1}\boldsymbol{r}_{O_1 P}]^\land \ {^1}R_0 \\
0 & {^1}R_0
\end{bmatrix}
{^0}\boldsymbol{f}_{O_1} \\
\end{aligned}
$$


## 4. Spatial Inertia (空間慣性行列)

### 4.1. Spatial Momentum (空間運動量)

通常の並進運動量は物体の質量と重心速度により以下となる。

$$
\boldsymbol{p} = m \boldsymbol{v}_G
$$

重心まわりの角運動量は以下となる。

$$
\boldsymbol{h}_G = \boldsymbol{I}_G \boldsymbol{\omega}
$$

剛体を特定の点を固定せずに、空中で剛体が並進と回転をしながら運動していることを考える。

すると、剛体の任意の基準点$O$まわりの角運動量$\boldsymbol{h}_O$は、点$O$から剛体内の微小質量要素$dm$までの位置ベクトルと、その質量要素の速度の外積を剛体全体にわたって質量積分することで求められる。

$$
\boldsymbol{h}_O = \int_B \boldsymbol{r}_{OP} \times \boldsymbol{v}_P \ dm
$$

角速度 $\boldsymbol{\omega}$ は剛体のどの点で評価しても変化しないため、剛体の重心$G$を使い点$P$の位置と速度を表現する。

$$
\boldsymbol{r}_{OP} = \boldsymbol{r}_{OG} + \boldsymbol{r}_{GP}, \quad \boldsymbol{v}_P = \boldsymbol{v}_G + \boldsymbol{\omega} \times \boldsymbol{r}_{GP}
$$

<p align="center">
<img src="./images/spatial_momentun_at_O.png" style="width:25%;" />
</p>

これらを代入すると、以下となる。

$$
\boldsymbol{h}_O = \int_B (\boldsymbol{r}_{OG} + \boldsymbol{r}_{GP}) \times (\boldsymbol{v}_G + \boldsymbol{\omega} \times \boldsymbol{r}_{GP} )\ dm
$$

展開すると、各項は次のようになる。

- 第1項 

    重心の並進による角運動量の定義であるため、以下となる。

    $$
    \int \boldsymbol{r}_{OG} \times \boldsymbol{v}_G \ dm = \boldsymbol{r}_{OG} \times m \boldsymbol{v}_G 
    $$

- 第2項

    位置ベクトル$\boldsymbol{r}_{OG}$と角速度$\boldsymbol{\omega}$は質量要素$dm$によらず一定とみなせるため、積分記号の外側にくくり出すことが出来る。

    $$
    \int \boldsymbol{r}_{OG} \times (\boldsymbol{\omega} \times \boldsymbol{r}_{GP}) \ dm = \boldsymbol{r}_{OG} \times (\boldsymbol{\omega} \times \int \boldsymbol{r}_{GP} \ dm) 
    $$

    重心の定義より、重心周りの位置ベクトルの積分は常にゼロとなるので、第2項はゼロとなる。

    $$
    \int \boldsymbol{r}_{GP} \ dm = 0
    $$

- 第3項

    点$G$の速度$\boldsymbol{v}_G$は質量要素$dm$によらず一定とみなせるため、積分記号の外側にくくり出すことが出来る。

    $$
    \int \boldsymbol{r}_{GP} \times \boldsymbol{v}_G dm = \left(\int \boldsymbol{r}_{GP} \ dm \right)\times \boldsymbol{v}_G  
    $$

    これも重心の定義より、第3項はゼロとなる。

- 第4項

    重心周りの角運動量の定義式([note](../../3D-EoM/Basic/Inertial-Matrix.ipynb))であるため、次となる。

    $$
    \int \boldsymbol{r}_{GP} \times (\boldsymbol{\omega} \times \boldsymbol{r}_{GP}) \ dm = \boldsymbol{I}_G \boldsymbol{\omega}
    $$

よって、剛体の任意の基準点$O$まわりによる角運動量$\boldsymbol{h}_O$を評価すると、次となる。

$$
\boldsymbol{h}_O = \boldsymbol{I}_G \boldsymbol{\omega} + \boldsymbol{r}_{OG} \times m \boldsymbol{v}_G 
$$

任意の点$O$周りの角運動量$\boldsymbol{h}_O$と並進運動量$\boldsymbol{p}$により以下の6次元ベクトルを定義する。これを Spatial Momentum と呼ぶ。

$$
\boldsymbol{H}_O = \begin{bmatrix} \boldsymbol{h}_O \\ \boldsymbol{p}\end{bmatrix} \in \mathbf{R}^{6 \times 1}
$$


### 4.2. Spatial Inertia

剛体の特定の点を固定せずに、空中で剛体が並進を回転を行いながら運動していることを考える。

剛体の重心$G$の速度を剛体上の点$O$により表現すると以下となる。

$$
\boldsymbol{v}_G = \boldsymbol{v}_O + \boldsymbol{\omega} \times \boldsymbol{r}_{OG}
$$

4.1. で示した点$O$における角運動量$\boldsymbol{h}_O$は以下である。これに$\boldsymbol{v}_G$ を代入して整理する。

$$
\begin{aligned}
\boldsymbol{h}_O &= \boldsymbol{I}_G \boldsymbol{\omega} + \boldsymbol{r}_{OG} \times m \boldsymbol{v}_G \\
&= \boldsymbol{I}_G \boldsymbol{\omega} + \boldsymbol{r}_{OG} \times m \left( \boldsymbol{v}_O + \boldsymbol{\omega} \times \boldsymbol{r}_{OG} \right) \\
&= m \boldsymbol{r}_{OG} \times \boldsymbol{v}_O + \boldsymbol{I}_G \boldsymbol{\omega} - m \boldsymbol{r}_{OG} \times (\boldsymbol{r}_{OG} \times  \boldsymbol{\omega}) \\
&= m [\boldsymbol{r}_{OG}]^\land \boldsymbol{v}_O + \boldsymbol{I}_G \boldsymbol{\omega} - m [\boldsymbol{r}_{OG}]^\land [\boldsymbol{r}_{OG}]^\land \boldsymbol{\omega} \\
&= m [\boldsymbol{r}_{OG}]^\land \boldsymbol{v}_O + \left(\boldsymbol{I}_G - m [\boldsymbol{r}_{OG}]^\land [\boldsymbol{r}_{OG}]^\land \right) \boldsymbol{\omega} \\
\end{aligned}
$$

並進運動量の式に$\boldsymbol{v}_G$を代入して整理する。

$$
\begin{aligned}
\boldsymbol{p} &= m \boldsymbol{v}_G \\
&= m \boldsymbol{v}_O + m \boldsymbol{\omega} \times \boldsymbol{r}_{OG} \\
&= m \boldsymbol{v}_O - m [\boldsymbol{r}_{OG}]^\land \boldsymbol{\omega}\\
\end{aligned}
$$


点$O$における Spatial Moment Vector は以下であり

$$
\boldsymbol{V}_O = \begin{bmatrix} \boldsymbol{\omega} \\ \boldsymbol{v}_O\end{bmatrix}
$$

点$O$における Spatial Momentum は以下である。

$$
\boldsymbol{H}_O = \begin{bmatrix} \boldsymbol{h}_O \\ \boldsymbol{p} \end{bmatrix}
$$

上式をまとめると、以下の行列形式となる。

$$
\boldsymbol{H}_O =
\begin{bmatrix}
\boldsymbol{I}_G - m [\boldsymbol{r}_{OG}]^\land [\boldsymbol{r}_{OG}]^\land & m [\boldsymbol{r}_{OG}]^\land \\
- m [\boldsymbol{r}_{OG}]^\land & m \boldsymbol{I}_3 
\end{bmatrix}
\boldsymbol{V}_O
$$

上式は運動量 $p=mv$ 、 角運動量 $L= \boldsymbol{I}_G \omega$ と同じ形式となっている。よって、$\boldsymbol{V}_O$の係数をSpatial Inertia $\mathcal{I}_O$ と呼ぶ。

$$
\mathcal{I}_O = 
\begin{bmatrix}
\boldsymbol{I}_G - m [\boldsymbol{r}_{OG}]^\land [\boldsymbol{r}_{OG}]^\land & m [\boldsymbol{r}_{OG}]^\land \\
- m [\boldsymbol{r}_{OG}]^\land & m \boldsymbol{I}_3 
\end{bmatrix} \in \mathbf{R}^{6 \times 6}
$$


### 4.3. Spatial Inertia の各ブロックの意味

- 左上 : $\boldsymbol{I}_G - m [\boldsymbol{r}_{OG}]^\land [\boldsymbol{r}_{OG}]^\land$

以下の式は重心$G$から離れた点$O$まわりの慣性テンソルを示す、平行軸の定理である。

$$
\boldsymbol{I}_O = \boldsymbol{I}_G - m [\boldsymbol{r}_{OG}]^\land [\boldsymbol{r}_{OG}]^\land 
$$

点$O$の並進速度$\boldsymbol{v}_O$がゼロなら、点$O$の角運動量$\boldsymbol{h}_O$は以下のように点$O$まわりの角運動量となる。

$$
\boldsymbol{h}_O = \boldsymbol{I}_O \boldsymbol{\omega}
$$

- 右上 : $m [\boldsymbol{r}_{OG}]^\land $

これは、並進速度が角運動量に与える影響を表している。

点$O$まわりに角運動量$\boldsymbol{h}_O$は以下である。

$$
\boldsymbol{h}_O = m [\boldsymbol{r}_{OG}]^\land \boldsymbol{v}_O + \boldsymbol{I}_O  \boldsymbol{\omega}
$$

このとき角速度 $\boldsymbol{\omega}$ がゼロとなっても$\boldsymbol{h}_O$は並進速度による角運動量が生じる。

$$
\boldsymbol{h}_O = m [\boldsymbol{r}_{OG}]^\land \boldsymbol{v}_O
$$

これは、剛体が空中をある速度で運動しているとき、剛体は回転していなくとも、評価する点を重心$G$から点$O$にすると、点$O$から見た重心の並進運動によって角運動量が生じることを意味している。

- 左下 : $- m [\boldsymbol{r}_{OG}]^\land$

これは、角速度が並進運動量に与える影響を表している。

点$O$における並進運動量 $\boldsymbol{p}$ は以下である。

$$
\boldsymbol{p} = m \boldsymbol{v}_O - m [\boldsymbol{r}_{OG}]^\land \boldsymbol{\omega}\\
$$

このとき、点$O$の並進速度 $\boldsymbol{v}_O$ がゼロとなっていても、$\boldsymbol{p}$は角速度による並進運動量が生じる。

$$
\boldsymbol{p} = - m [\boldsymbol{r}_{OG}]^\land \boldsymbol{\omega} = m \boldsymbol{\omega} \times \boldsymbol{r}_{OG}
$$

これは、点$O$の並進速度がゼロであっても、剛体が点$O$周りに回転すると、重心$G$に接線速度が生じる。その結果、並進運動量が生じることを意味している。

- 右下 : $m \boldsymbol{I}_3$

並進速度による並進運動量を示している。

点$O$における並進運動量 $\boldsymbol{p}$ は以下である。

$$
\boldsymbol{p} = m \boldsymbol{v}_O - m [\boldsymbol{r}_{OG}]^\land \boldsymbol{\omega}\\
$$

角速度$\boldsymbol{\omega}$がゼロであれば、並進運動量は以下となり、通常の並進運動量そのものを表している。

$$
\boldsymbol{p} = m \boldsymbol{v}_O
$$

### 4.4. Spatial Inertiaの対称性

慣性テンソルはその定義より対称行列である。Spatial Inertia $\mathcal{I}_O$の対称性を確かめる。

$$
\mathcal{I}_O = 
\begin{bmatrix}
\boldsymbol{I}_G - m [\boldsymbol{r}_{OG}]^\land [\boldsymbol{r}_{OG}]^\land & m [\boldsymbol{r}_{OG}]^\land \\
- m [\boldsymbol{r}_{OG}]^\land & m \boldsymbol{I}_3 
\end{bmatrix} \in \mathbf{R}^{6 \times 6}
$$

$S = [\boldsymbol{r}_{OG}]^\land$ とすると、

$$
\mathcal{I}_O = 
\begin{bmatrix}
\boldsymbol{I}_G - m S^2 & m S \\
- m S & m \boldsymbol{I}_3 
\end{bmatrix}
$$

上式を転置すると、各ブロックは $3\times3$の行列であるため、

$$
\mathcal{I}_O^T = 
\begin{bmatrix}
\boldsymbol{I}_G^T - m (S^2)^T & -m S^T \\
m S^T & m \boldsymbol{I}_3 
\end{bmatrix}
$$

$\boldsymbol{I}_G$ の慣性テンソルは対称行列であるため、

$$
\boldsymbol{I}_G^T = \boldsymbol{I}_G
$$

$S$は歪対称行列であるため、

$$
S^T = -S, \quad (S^2)^T = (S^T)^2 = (-S)^2 = S^2
$$

となる。よって Spatial Inertia は対称行列である。

$$
\mathcal{I}_O^T = 
\begin{bmatrix}
\boldsymbol{I}_G - m (S^2) & m S \\
- m S & m \boldsymbol{I}_3 
\end{bmatrix} = \mathcal{I}_O 
$$

#### Spatial Inertia と運動エネルギーの関係

剛体の運動エネルギーは以下である。

$$
T = \frac{1}{2}m \boldsymbol{v}_G^T \boldsymbol{v}_G + \frac{1}{2} \boldsymbol{\omega}^T \boldsymbol{I}_G \boldsymbol{\omega}
$$

上節では、Spatial Momentum を Spatial Inertia と Spatial Moment Vectorで以下のように示されることを導いた。

$$
\boldsymbol{H}_O = \begin{bmatrix} \boldsymbol{h}_O \\ \boldsymbol{p} \end{bmatrix} = \mathcal{I}_O \boldsymbol{V}_O
$$

運動エネルギーは速度と運動量の積となるため、

$$
\begin{aligned}
\boldsymbol{V}_O^T \boldsymbol{H}_O &= \begin{bmatrix} \boldsymbol{\omega}^T & \boldsymbol{v}_O^T \end{bmatrix} \begin{bmatrix} \boldsymbol{h}_O \\ \boldsymbol{p} \end{bmatrix} \\
&= \boldsymbol{\omega}^T \boldsymbol{h}_O + \boldsymbol{v}_O^T\boldsymbol{p} \\
&= \boldsymbol{\omega}^T (\boldsymbol{I}_G \boldsymbol{\omega} + \boldsymbol{r}_{OG} \times m \boldsymbol{v}_G ) + m \boldsymbol{v}_O^T \boldsymbol{v}_G \\
&= \boldsymbol{\omega}^T \boldsymbol{I}_G \boldsymbol{\omega} + \boldsymbol{\omega}^T (\boldsymbol{r}_{OG} \times m \boldsymbol{v}_G ) + m \boldsymbol{v}_O^T \boldsymbol{v}_G \\

\end{aligned}
$$

スカラー三重積より以下となるので、

$$
\boldsymbol{\omega}^T (\boldsymbol{r}_{OG} \times m \boldsymbol{v}_G ) = m (\boldsymbol{\omega} \times \boldsymbol{r}_{OG})^T \boldsymbol{v}_G
$$

$$
\begin{aligned}
\boldsymbol{V}_O^T \boldsymbol{H}_O &= \boldsymbol{\omega}^T \boldsymbol{I}_G \boldsymbol{\omega} + m (\boldsymbol{\omega} \times \boldsymbol{r}_{OG})^T \boldsymbol{v}_G + m \boldsymbol{v}_O^T \boldsymbol{v}_G \\
&= \boldsymbol{\omega}^T \boldsymbol{I}_G \boldsymbol{\omega} + m (\boldsymbol{v}_O + \boldsymbol{\omega} \times \boldsymbol{r}_{OG})^T \boldsymbol{v}_G\\
&= \boldsymbol{\omega}^T \boldsymbol{I}_G \boldsymbol{\omega} + m \boldsymbol{v}_G^T \boldsymbol{v}_G\\
\end{aligned} \\
$$

これより、対称行列であるSpatial Inertia 剛体の運動エネルギーを Spatial Motion Vectorの二次形式として表現できる。

$$
T = \frac{1}{2} \boldsymbol{V}_O^T \boldsymbol{H}_O = \frac{1}{2} \boldsymbol{V}_O^T \mathcal{I}_O \boldsymbol{V}_O
$$
