# Spatial Vector

## 1. Spatial Motion Vector

二つの座標系$\Sigma_0, \Sigma_1$があり、$\Sigma_1$の原点$O_1$は$\Sigma_0$の原点$O_0$から${^0}p_1$の位置にあり、その姿勢は$\theta_1$となっている。<br>
$\Sigma_1$は剛体上に定義された座標系であり、剛体とともに移動する。$\Sigma_0$はWorld座標とする。

</p>

$\Sigma_1$ 上にある点$P$の位置を$\Sigma_0$で表すと次のようになる。$P$は$\Sigma_1$の原点$O_1$から${^1}r_1$の位置にある。

$$
{^0}p_{P} = {^0}p_1 + {^0}R_1 {^1}r_1
$$

${^0}R_1$ は$\Sigma_0$における$\Sigma_1$の姿勢を示す回転行列である。

上式を時間微分し、点$P$の速度を求める。

$$
{^0}\dot{p}_{P} = {^0}\dot{p}_1 + {^0}\dot{R}_1 {^1}r_1 + {^0}R_1 {^1}\dot{r}_1
$$

点$P$は座標系 $\Sigma_1$に固定された剛体上の点であり、${^1}r_1$は時間変化しない。すると${^1}\dot{r}_1 = 0$となるため、

$$
{^0}\dot{p}_{P} = {^0}\dot{p}_1 + {^0}\dot{R}_1 {^1}r_1
$$

回転行列の時間微分は、角速度ベクトルを用いて ${^0}\dot{R}_1 = [{^0}\omega_{1}]^\land \ {^0}{R}_1$ と表せるため、

$$
{^0}\dot{p}_P = {^0}\dot{p}_1 + [{^0}\omega_{1}]^\land \ {^0}{R}_1 {^1}r_1 = {^0}\dot{p}_1 + [{^0}\omega_{1}]^\land {^0}r_1 
$$

${^0}\dot{p}_P = {^0}v_{P}$、${^0}\dot{p}_1 = {^0}v_{O_1}$、${^0}r_1 = {^0}r_{O_1P}$ と記述すると次となる。

$$
{^0}v_{P} = {^0}v_{O_1} + {^0}\omega_{1} \times {^0}r_{O_1P}
$$

- ${^0}v_{O_1}$ : $\Sigma_1$の原点$O_1$の並進運動
- ${^0}\omega_{1} \times {^0}r_{O_1P}$ : $\Sigma_1$の原点$O_1$まわりの回転による運動

<img src="./images/velocity.png" style="width:30%;" />
</p>

剛体の角速度は剛体全の回転運動を表す量であるため、基準点に依存しない。よって点$O_1$と点$P$で同じ角速度を使用できる。

$$
{^0}\omega_P = {^0}\omega_1
$$

まとめると以下の行列形式となる。

$$
\begin{bmatrix}
{^0}\omega_P \\ {^0}v_{P}
\end{bmatrix} =
\begin{bmatrix}
{^0}\omega_1 \\
{^0}v_{O_1} + {^0}\omega_{1} \times {^0}r_{O_1P}
\end{bmatrix}
$$

外積は$A \times B = - B \times A$ であるため、次のように変換を行う。

$$
{^0}\omega_{1} \times {^0}r_{O_1P} = - {^0}r_{O_1P} \times {^0}\omega_{1} = - [{^0}r_{O_1P}]^\land \ {^0}\omega_{1}
$$

すると、剛体上の基準点を$O$から$P$に変更したときの角速度と並進速度の関係を次の行列形式で表すことが出来る。

$$
\begin{bmatrix}
{^0}\omega_P \\ {^0}v_{P}
\end{bmatrix} =
\begin{bmatrix}
\boldsymbol{I}_3 & 0 \\
- [{^0}r_{O_1P}]^\land &  \boldsymbol{I}_3
\end{bmatrix}
\begin{bmatrix}
{^0}\omega_1 \\ {^0}v_{O_1}
\end{bmatrix}
$$


#### Local座標系での表現

上記表現はWorld座標系上での基準点の変更であった。

角速度と速度を$\boldsymbol{V}$とする。これをSpatial Motion Vector と呼ぶ。

$$
\boldsymbol{V} = \begin{bmatrix} \omega \\ v \end{bmatrix} \in \mathbf{R}^{6 \times 1}
$$

$\boldsymbol{V}$ を用いて以下のように基準点の変更を表すことが出来る。$\boldsymbol{X}_{P \leftarrow O_1}^{(0)}$は$(0)$で表現する座標系を示し、$P \leftarrow O_1$で$O_1$で表現した基準点を$P$へ変更することを示している。

$$
{^0}\boldsymbol{V}{_P} = \boldsymbol{X}_{P \leftarrow O_1}^{(0)} {^0}\boldsymbol{V}{_O_1}
$$

ABAでは各リンクのローカル座標系で演算するため、${^1}\boldsymbol{V}{_P}$ を求める。

##### 角速度

角速度は回転行列を用いて次の座標変換となる。

$$
{^1}\omega{_1} = ({^0}R{_1})^T {^0}\omega{_1} ,\quad {^1}\omega{_P} = ({^0}R{_1})^T {^0}\omega{_P} 
$$

##### 速度

$\Sigma_0$による表現は以下である。

$$
{^0}v_{P} = {^0}v_O + {^0}\omega_{1} \times {^0}r_{O_1P}
$$

上式の両辺に$({^0}R{_1})^T$を掛けると、以下となる。

$$
\begin{aligned}
({^0}R{_1})^T \ {^0}v_{P} &= ({^0}R{_1})^T \ {^0}v_{O_1} + ({^0}R{_1})^T \left({^0}\omega_{1} \times {^0}r_{O_1P} \right) \\
&= ({^0}R{_1})^T \ {^0}v_{O_1} +  \left(({^0}R{_1})^T \ {^0}\omega_{1} \right)  \times \left( ({^0}R{_1})^T \ {^0}r_{O_1P} \right)  \\
{^1}v_{P} &= {^1}v_{O_1} + {^1}\omega_{1} \times {^1}r_{O_1P} \\
&= {^1}v_{O_1} - {^1}r_{O_1P} \times {^1}\omega_{1}
\end{aligned}
$$

すると、Local座標において、剛体上の基準点を$O_1$から$P$に変更したときの角速度と並進速度の関係を次の行列形式で表すことが出来る。

$$
\begin{bmatrix}
{^1}\omega_P \\ {^1}v_{P}
\end{bmatrix} =
\begin{bmatrix}
\boldsymbol{I}_3 & 0 \\
- [{^1}r_{O_1P}]^\land &  \boldsymbol{I}_3
\end{bmatrix}
\begin{bmatrix}
{^1}\omega_1 \\ {^1}v_{O_1}
\end{bmatrix}
$$

Spatial Motion Vectorを用いると次のように記述できる。

$$
{^1}\boldsymbol{V}{_P} = \boldsymbol{X}_{P \leftarrow O}^{(1)} {^1}\boldsymbol{V}{_O}
$$


## 2. Spatial Transform

World座標$\Sigma_0$からLocal座標$\Sigma_1$への座標変換を行うと、

$$
{^1}\omega_{1} = {^1}R_0 \ {^0}\omega{_1} \\
{^1}v_{O_1} = {^1}R_0 \ {^0}v_{O_1}
$$

より、座標系原点のSpatial Motion Vectorの座標変換は以下となる。

$$
\begin{bmatrix}
{^1}\omega_{1} \\ {^1}v_{O_1}
\end{bmatrix} = 
\begin{bmatrix}
{^1}R_0 & 0 \\
0 & {^1}R_0
\end{bmatrix}
\begin{bmatrix}
{^0}\omega_{1} \\ {^0}v_{O_1}
\end{bmatrix}  \quad \rightarrow \quad
{^1}\boldsymbol{V}{_{O_1}} = 
\begin{bmatrix}
{^1}R_0 & 0 \\
0 & {^1}R_0
\end{bmatrix}
{^0}\boldsymbol{V}{_{O_1}} 
$$

基準点の変更は以下であったので、上式を代入すると

$$
\begin{aligned}
{^1}\boldsymbol{V}{_P} &=
\begin{bmatrix}
\boldsymbol{I}_3 & 0 \\
- [{^1}r_{O_1P}]^\land &  \boldsymbol{I}_3
\end{bmatrix}
{^1}\boldsymbol{V}{_{O_1}} =
\begin{bmatrix}
\boldsymbol{I}_3 & 0 \\
- [{^1}r_{O_1P}]^\land &  \boldsymbol{I}_3
\end{bmatrix}
\begin{bmatrix}
{^1}R_0 & 0 \\
0 & {^1}R_0
\end{bmatrix}
{^0}\boldsymbol{V}{_{O_1}} \\
&= \begin{bmatrix}
{^1}R_0 & 0 \\
- [{^1}r_{O_1P}]^\land \ {^1}R_0 &  {^1}R_0
\end{bmatrix}
{^0}\boldsymbol{V}{_{O_1}}
\end{aligned}
$$
