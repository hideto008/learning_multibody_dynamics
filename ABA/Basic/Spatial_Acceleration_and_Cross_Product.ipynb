1. Spatial Acceleration (空間加速度)

## 1. Spatial Acceleration

### 1.1. 輸送定理

World座標系$\Sigma_0$と、剛体に固定された回転座標系$\Sigma_B$があるとする。

剛体は以下の図のように角速度$\boldsymbol{\omega}$で回転しているとする。

<p align="center">
<img src="./images/transport_theorem.png" style="width:30%;" />
</p>

$\Sigma_B$の原点$O_B$から、点$P$への相対位置ベクトルを$\boldsymbol{x}$とする。このベクトルを$\Sigma_0$で表すと以下となる。

$$
{^0}\boldsymbol{x} = {^0}R_B \ {^B}\boldsymbol{x}
$$

- ${^0}\boldsymbol{x}$ : $\Sigma_0$ で表したベクトル
- ${^B}\boldsymbol{x}$ : $\Sigma_B$ で表したベクトル
- ${^0}R{_B}$ : $\Sigma_B$ から $\Sigma_0$ への回転行列

上式を時間微分すると、積の公式より以下となる。

$$
{^0}\dot{\boldsymbol{x}} = {^0}\dot{R}_B \ {^B}{\boldsymbol{x}} + {^0}R{_B} \ {^B}\dot{\boldsymbol{x}}
$$

回転行列の時間微分は以下である。

$$
{^0}\dot{R}_B = [{^0}\boldsymbol{\omega}_{B/0}]^\land \ {^0}R_B
$$

代入すると、以下となる。

$$
{^0}\dot{\boldsymbol{x}} = [{^0}\boldsymbol{\omega}_{B/0}]^\land \ {^0}R_B \ {^B}{\boldsymbol{x}} + {^0}R{_B} \ {^B}\dot{\boldsymbol{x}}
$$

これより、次のように表せる。これを輸送定理と呼ぶ。

$$
\boxed{
{^0}\dot{\boldsymbol{x}} = {^0}R{_B} \ {^B}\dot{\boldsymbol{x}} + {^0}\boldsymbol{\omega}_{B/0} \times {^0}{\boldsymbol{x}} 
}
$$

つまり、世界座標系で見たベクトルの時間変化${^0}\dot{\boldsymbol{x}}$は次の２項の足し合わせとなる。

- ${^0}R{_B} \ {^B}\dot{\boldsymbol{x}}$ : 回転座標系内での変化を世界座標系への変換
- ${^0}\boldsymbol{\omega}_{B/0} \times {^0}{\boldsymbol{x}}$ : $\Sigma_B$座標系が世界座標系に対して回転することによる変化

上記のとおり、輸送定理には、ベクトルの原点$O_B$の並進速度は含まれない。なぜなら、輸送定理はベクトルの向きや大きさの時間変化を扱う定理であり、ベクトルの原点が移動しても、そのベクトル自体は変化しないからである。

例えば、剛体が回転せずに、右方向へ速度2m/sで移動しているとする。<br>
このとき、$\boldsymbol{\omega} = 0$ であり、さらに$\boldsymbol{x}$が剛体固定された相対位置ベクトルなら、${^B}\dot{\boldsymbol{x}}=0$ である。<br>
輸送定理より、以下となる。

$$
{^0}\dot{\boldsymbol{x}} = 0
$$

一方、点$P$の世界座標系$\Sigma_0$における位置を時間微分する場合、$\Sigma_B$の原点$O_B$の並進速度も考慮する必要がある。<br>
$\Sigma_0$の原点$O_0$から$\Sigma_B$の原点$O_B$までを位置ベクトル$\boldsymbol{r}_{O_O O_B}$ とする。これを用いて点$P$を世界座標系で示すと、次である。

$$
{^0}\boldsymbol{p}_P = {^0}\boldsymbol{r}_{O_O O_B} + {^0}R{_B} {^B}\boldsymbol{x}
$$

この時間微分は以下となる。

$$
{^0}\dot{\boldsymbol{p}}_P = {^0}\dot{\boldsymbol{r}}_{O_O O_B} + {^0}\dot{R}{_B} {^B}\boldsymbol{x} + {^0}R{_B} {^B}\dot{\boldsymbol{x}}
$$

回転行列の時間微分は以下である。

$$
{^0}\dot{R}_B = [{^0}\boldsymbol{\omega}]^\land \ {^0}R_B
$$

そして、${^0}\dot{\boldsymbol{p}}_P = {^0}{\boldsymbol{v}}_P, {^0}\dot{\boldsymbol{r}}_{O_O O_B} = {^0}{\boldsymbol{v}}_{O_B}$ とすると、以下となる。

$$
{^0}{\boldsymbol{v}}_P = {^0}{\boldsymbol{v}}_{O_B} + {^0}\boldsymbol{\omega} \times {^0}\boldsymbol{x} + {^0}R{_B} {^B}\dot{\boldsymbol{x}}
$$


### 1.2. Spaital Accelration

Spatial Motion Vector の時間微分が Spatial Acceleration である。

剛体に固定された座標系 $\Sigma_B$ の原点を$O_B$とする。

Spatial Motion Vectorは、剛体の角速度${^B}\boldsymbol{\omega}$ と 座標系原点$O$の並進速度${^B}\boldsymbol{v}_{O_B}$を用いると以下となる。

$$
{^B}\boldsymbol{V}_{O_B} = \begin{bmatrix} {^B}\boldsymbol{\omega} \\ {^B}\boldsymbol{v}_{O_B} \end{bmatrix}
$$

Spatial Motion Vectorの両成分は $\Sigma_B$ で表されている。これを時間微分する。

#### 1.2.1. 角速度の時間微分

${^B}\boldsymbol{\omega}$ を$\Sigma_0$で表すと以下となる。

$$
{^0}\boldsymbol{\omega} = {^0}R{_B}{^B}\boldsymbol{\omega}
$$

この時間微分として輸送定理を適用すると、以下となる。

$$
{^0}\dot{\boldsymbol{\omega}} = {^0}R{_B}{^B}\dot{\boldsymbol{\omega}} + {^0}\boldsymbol{\omega} \times {^0}\boldsymbol{\omega}
$$

しかし、${^0}\boldsymbol{\omega} \times {^0}\boldsymbol{\omega} = 0$ となるため、以下となる。

$$
{^0}\dot{\boldsymbol{\omega}} = {^0}R{_B}{^B}\dot{\boldsymbol{\omega}}
$$

よって、角速度${^B}\boldsymbol{\omega}$の時間微分である角加速度を以下で表す。

$$
\boxed{
{^B}{\boldsymbol{\alpha}} = {^B}R{_0}{^0}\dot{\boldsymbol{\omega}} = {^B}\dot{\boldsymbol{\omega}} 
}
$$

#### 1.2.2. 並進速度の時間微分

${^B}\boldsymbol{v}_{O_B}$ を $\Sigma_0$ で表すと以下である。

$$
{^0}\boldsymbol{v}_{O_B} = {^0}R{_B}{^B}\boldsymbol{v}_{O_B}
$$

角速度と同じく、輸送定理を適用すると以下となる。

$$
{^0}\dot{\boldsymbol{v}}_{O_B} = {^0}R{_B}{^B}\dot{\boldsymbol{v}}_{O_B} + {^0}\boldsymbol{\omega} \times {^0}\boldsymbol{v}_{O_B}
$$

左辺は$\Sigma_0$で表した$\Sigma_B$の原点$O_B$の並進加速度であるため、以下で記述する。

$$
{^0}{\boldsymbol{a}}_{O_B} = {^0}\dot{\boldsymbol{v}}_{O_B}
$$

$$
{^0}{\boldsymbol{a}}_{O_B} = {^0}R{_B}{^B}\dot{\boldsymbol{v}}_{O_B} + {^0}\boldsymbol{\omega} \times {^0}\boldsymbol{v}_{O_B}
$$

両辺に${^B}R{_0}$を掛けて$\Sigma_B$へ座標変換すると、

$$
\begin{aligned}
{^B}R{_0}{^0}{\boldsymbol{a}}_{O_B} &= {^B}R{_0}{^0}R{_B}{^B}\dot{\boldsymbol{v}}_{O_B} + {^B}R{_0}{^0} \left(\boldsymbol{\omega} \times {^0}\boldsymbol{v}_{O_B} \right) \\
{^B}{\boldsymbol{a}}_{O_B} &= {^B}\dot{\boldsymbol{v}}_{O_B} + {^B}R{_0}{^0}\boldsymbol{\omega} \times {^B}R{_0}{^0}\boldsymbol{v}_{O_B} \\
{^B}{\boldsymbol{a}}_{O_B} &= {^B}\dot{\boldsymbol{v}}_{O_B} + {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_{O_B} \\
\end{aligned}
$$

よって、${^B}\boldsymbol{v}_{O_B}$ の時間微分は以下となる。

$$
\boxed{
{^B}\dot{\boldsymbol{v}}_{O_B} = {^B}{\boldsymbol{a}}_{O_B} - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_{O_B}
}
$$

### 1.2.3. Spaital Accelration の導出

Spatial Motion Vector の時間微分をまとめると以下となる。

$$
{^B}\dot{\boldsymbol{V}}_O = \begin{bmatrix} {^B}\dot{\boldsymbol{\omega}} \\ {^B}\dot{\boldsymbol{v}}_{O_B} \end{bmatrix} = 
\begin{bmatrix}
{^B}{\boldsymbol{\alpha}}  \\
{^B}{\boldsymbol{a}}_{O_B} - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_{O_B}
\end{bmatrix}
$$

これを Spatial Accelerationと呼ぶ。

$$
\boxed{
{^B}\boldsymbol{A}_{O_B} = \begin{bmatrix}
{^B}{\boldsymbol{\alpha}}  \\
{^B}{\boldsymbol{a}}_{O_B} - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_{O_B}
\end{bmatrix}
}
$$

### 1.2.4. Spatial Transform (基準点変更)

$\Sigma_B$の原点$O_B$の速度を重心$G$基準で示すと以下となる。

$$
{^B}\boldsymbol{v}_{O_B} = {^B}\boldsymbol{v}_G + {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{r}_{G O_B}
$$

上式の両辺を慣性座標系で時間微分し、得られた加速度を$\Sigma_B$で表す。物体がある角速度$\boldsymbol{\omega}$で回転しているとき、位置ベクトルの時間微分は$\boldsymbol{\omega}$との外積で計算できるため、以下となる。

$$
{^B}\boldsymbol{a}_{O_B} = {^B}\boldsymbol{a}_G + {^B}\dot{\boldsymbol{\omega}} \times {^B}\boldsymbol{r}_{G O_B} + {^B}\boldsymbol{\omega} \times ({^B}\boldsymbol{\omega} \times {^B}\boldsymbol{r}_{G O_B}) 
$$

これらをSpatial Accelertaionの並進加速度成分に代入する。

$$
\begin{aligned}
{^B}{\boldsymbol{a}}_{O_B} - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_{O_B} &= {^B}\boldsymbol{a}_G + {^B}\dot{\boldsymbol{\omega}} \times {^B}\boldsymbol{r}_{G O_B} \\ 
& \space + {^B}\boldsymbol{\omega} \times ({^B}\boldsymbol{\omega} \times {^B}\boldsymbol{r}_{G O_B}) \\
& \space - {^B}\boldsymbol{\omega} \times ({^B}\boldsymbol{v}_G + {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{r}_{G O_B} ) \\
&= {^B}\boldsymbol{a}_G + {^B}\dot{\boldsymbol{\omega}} \times {^B}\boldsymbol{r}_{G O_B} - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_G \\
&= {^B}\boldsymbol{a}_G - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_G - [{^B}\boldsymbol{r}_{G O_B}]^\land {^B}{\boldsymbol{\alpha}} 
\end{aligned}
$$

重心$G$におけるSpatial Accelerationを以下とすると、

$$
{^B}\boldsymbol{A}_{G} = \begin{bmatrix}
{^B}{\boldsymbol{\alpha}}  \\
{^B}{\boldsymbol{a}}_{G} - {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{v}_{G}
\end{bmatrix}
$$

重心$G$から$\Sigma_B$の原点$O_B$への基準点変更は以下となる。

$$
{^B}\boldsymbol{A}_{O_B} = \begin{bmatrix} \boldsymbol{I}_3 & 0 \\ - [{^B}\boldsymbol{r}_{G O_B}]^\land & \boldsymbol{I}_3 \end{bmatrix} {^B}\boldsymbol{A}_{G}
$$



### 1.2.5. Motion Vector, Force Vector との違い

Spatial Motion Vectorを$\Sigma_{B}$の原点$O_B$を示すと以下となる。

$$
{^B}\boldsymbol{V}{_{O_B}} = \begin{bmatrix} \boldsymbol{I}_3 & 0 \\ -[\boldsymbol{r}_{G O_B}]^\land & \boldsymbol{I} \end{bmatrix} {^B}\boldsymbol{V}{_{G}} , \quad {^B}\boldsymbol{V}{_{G}} =\begin{bmatrix} {^B}\boldsymbol{\omega} \\ {^B}\boldsymbol{v}_G \end{bmatrix}
$$

Spatial Force Vectorを$\Sigma_{B}$の原点$O_B$で示すと以下となる。

$$
{^B}\boldsymbol{f}{_{O_B}} = \begin{bmatrix} \boldsymbol{I}_3 & -[\boldsymbol{r}_{G O_B}]^\land \\ 0 & \boldsymbol{I} \end{bmatrix} {^B}\boldsymbol{f}{_{G}} , \quad {^B}\boldsymbol{f}{_{G}} = \begin{bmatrix} {^B}\boldsymbol{n}_G \\ {^B}\boldsymbol{F}_G\end{bmatrix}=\begin{bmatrix} {^B}\boldsymbol{I}_G {^B}\dot{\boldsymbol{\omega}}+ {^B}\boldsymbol{\omega} \times  ({^B}\boldsymbol{I}_G {^B}\boldsymbol{\omega}) \\ m{^B}\boldsymbol{a}_G \end{bmatrix}
$$

これより、${^B}\boldsymbol{V}{_{G}}$とその時間変化、${^B}\boldsymbol{I}_G, m$ が分かれば、${^B}\boldsymbol{A}_G$と${^B}\boldsymbol{f}_G$を求めることができる。さらに${^B}\boldsymbol{r}_{G O_B}$が分かると、${^B}\boldsymbol{V}_{O_B}, {^B}\boldsymbol{f}_{O_B}$と${^B}\boldsymbol{A}_{O_B}$を求めることができる。


## 2. Spatial Cross Product

二つの Spatial Motion Vectorを考える。<br> 具体的には、2Linkのアームシステムがあり、Link1とLink2はそれぞれの関節によって運動する。Link2の運動はLink1から伝わる運動と、Link1に対するLink2自身の相対運動を合成したものである。これらの運動をSpatiam Motion Vectorで表し、速度の合成と時間微分を考える。

Link2の座標系を$\Sigma_2$とし、Link2のLink1に対する相対運動のSpatial Motion Vectorを、$\Sigma_2$で表すと、

$$
{^2}\boldsymbol{V}_{2/1, P} = \begin{bmatrix}
{^2} \boldsymbol{\omega}_{2/1} \\
{^2} \boldsymbol{v}_{2/1, P}
\end{bmatrix}
$$

Link1の運動を$\Sigma_2$で表したものを以下とする。

$$
{^2}\boldsymbol{V}_{1,P} = \begin{bmatrix}
{^2} \boldsymbol{\omega}_{1} \\
{^2} \boldsymbol{v}_{1,P}
\end{bmatrix}
$$

この両方のMotion Vectorの基準点は同じ場所である。例えば、Link1とLink2の節点$P$とする。

<img src="./images/cross_vector_2linkarm.png" style="width:35%;" />
</p>

これらを用いて$\Sigma_2$におけるLink2の速度は二つのMotion Vectorの合成となる。

$$
{^2}\boldsymbol{V}_{2,P} = {^2}\boldsymbol{V}_{1,P} + {^2}\boldsymbol{V}_{2/1,P}
$$

この時間微分はSpatial Accelertionとなる。

$$
{^2}\dot{\boldsymbol{V}}_{2,P} = {^2}{\boldsymbol{A}}_{2,P} = \frac{d}{dt}\left({^2}\boldsymbol{V}_{1,P}\right) + \frac{d}{dt}\left({^2}\boldsymbol{V}_{2/1,P}\right)
$$

### 2.1. Link1のSpatial Motion Vectorの時間微分について

$\frac{d}{dt}\left({^2}\boldsymbol{V}_{1,P}\right)$ について計算を行う。

$$
\frac{d}{dt}\left({^2}\boldsymbol{V}_{1,P}\right) = \frac{d}{dt}\begin{bmatrix}
{^2} \boldsymbol{\omega}_{1} \\
{^2} \boldsymbol{v}_{1,P}
\end{bmatrix}
$$

#### 2.1.1. 角速度${^2} \boldsymbol{\omega}_{1}$の時間微分

Link1の角速度${^1}\boldsymbol{\omega}_{1}$を$\Sigma_2$ で表すと以下である。

$$
{^2}\boldsymbol{\omega}_{1} = {^2}R{_1} {^1}\boldsymbol{\omega}_{1}
$$

時間微分を行うと、

$$
\frac{d}{dt}\left({^2}\boldsymbol{\omega}_{1}\right) = {^2}\dot{R}{_1} {^1}\boldsymbol{\omega}_{1} + {^2}{R}{_1} {^1}\dot{\boldsymbol{\omega}}_{1}
$$

${^2}{R}{_1}$は$\Sigma_1$と$\Sigma_2$の相対回転行列である。まず、

$$
{^1}\dot{R}{_2} = {^1}{R}{_2} [{^2}\boldsymbol{\omega}{_{2/1}}]^\land 
$$

上式を転置して以下となる。

$$
{^2}\dot{R}{_1} = - [{^2}\boldsymbol{\omega}{_{2/1}}]^\land {^2}{R}{_1}  
$$

${^1}\dot{\boldsymbol{\omega}}_{1}={^1}{\boldsymbol{\alpha}}_{1}$ であるため、${^1}\boldsymbol{\omega}_{1}$ の時間微分は以下となる。

$$
\begin{aligned}
\frac{d}{dt}\left({^2}\boldsymbol{\omega}_{1}\right) &= - [{^2}\boldsymbol{\omega}{_{2/1}}]^\land {^2}{R}{_1}{^1}\boldsymbol{\omega}_{1} + {^2}{R}{_1} {^1}{\boldsymbol{\alpha}}_{1} \\
&= - [{^2}\boldsymbol{\omega}{_{2/1}}]^\land {^2}\boldsymbol{\omega}_{1} + {^2}{\boldsymbol{\alpha}}_{1} \\
\end{aligned}
$$


#### 2.1.3. 並進速度${^2}\boldsymbol{v}_{1,P}$の時間微分

節点$P$の速度はLink1とLink2で同じになる。ここで ${^2}\boldsymbol{v}_{2,P}$ はLink2の$\Sigma_0$における速度を$\Sigma_2$で表したものである。実際にLink2の点$P$の位置はLink1により動かされるため、相対速度${^2}\boldsymbol{v}_{2/1,P}$とは異なりゼロではない。

$$
{^2}\boldsymbol{v}_{1,P} = {^2}\boldsymbol{v}_{2,P} = {^2}\boldsymbol{v}_P
$$

世界座標系$\Sigma_0$で表すと以下となる。

$$
{^0}\boldsymbol{v}_P = {^0}R{_2} {^2}\boldsymbol{v}_P
$$

輸送定理より、${^0}\boldsymbol{\omega}_2$を用いて以下となる。

$$
{^0}\dot{\boldsymbol{v}}_P = {^0}R{_2} {^2}\dot{\boldsymbol{v}}_P + {^0}\boldsymbol{\omega}_2 \times {^0}{\boldsymbol{v}}_P
$$

両辺に${^2}R{_0}$を掛けて整理すると、${^2}\boldsymbol{v}_P$の時間微分が求まる。

$$
\frac{d}{dt}\left({^2}\boldsymbol{v}_P\right) = {^2}\boldsymbol{a}_P - {^2}\boldsymbol{\omega}_2 \times {^2}\boldsymbol{v}_P
$$

### 2.2 Link1のSpatial Accelerationとの差

Link1の基準点$P$のSpatial Accelerationを$\Sigma_2$で表すと、1.2.3 節の定義より以下となる。これは${^2}\boldsymbol{V}_{1,P}$を$\Sigma_2$で時間微分したものではない。

$$
{^2}\boldsymbol{A}_{1,P} = \begin{bmatrix}
{^2}\boldsymbol{\alpha}_1 \\
{^2}\boldsymbol{a}_{P} - {^2}\boldsymbol{\omega}_{1} \times {^2}\boldsymbol{v}_{P}
\end{bmatrix}
$$

Link1の基準点$P$のSpatial Motion Vectorを$\Sigma_2$で時間微分したものは以下であった。

$$
\frac{d}{dt}\left({^2}\boldsymbol{V}_{1,P} \right) = 
\frac{d}{dt}\begin{bmatrix}
{^2} \boldsymbol{\omega}_{1} \\
{^2} \boldsymbol{v}_{1,P}
\end{bmatrix} = 
\begin{bmatrix}
{^2}{\boldsymbol{\alpha}}_{1} - {^2}\boldsymbol{\omega}{_{2/1}} \times {^2}\boldsymbol{\omega}_{1} \\
{^2}\boldsymbol{a}_{P} - {^2}\boldsymbol{\omega}_2 \times {^2}\boldsymbol{v}_P
\end{bmatrix}
$$


両式の差を計算すると次となる。

$$
\begin{aligned}
\frac{d}{dt}\left({^2}\boldsymbol{V}_{1,P} \right) - {^2}\boldsymbol{A}_{1,P} &= \begin{bmatrix}
- {^2}\boldsymbol{\omega}{_{2/1}} \times {^2}\boldsymbol{\omega}_{1} \\
- {^2}\boldsymbol{\omega}_2 \times {^2}\boldsymbol{v}_P + {^2}\boldsymbol{\omega}_{1} \times {^2}\boldsymbol{v}_{P}
\end{bmatrix} \\
&= \begin{bmatrix}
- {^2}\boldsymbol{\omega}{_{2/1}} \times {^2}\boldsymbol{\omega}_{1} \\
- \left( {^2}\boldsymbol{\omega}_2 - {^2}\boldsymbol{\omega}_{1} \right) \times {^2}\boldsymbol{v}_P 
\end{bmatrix}
\end{aligned}
$$

角速度ベクトルは以下のように合成することができるため、

$$
{^2}\boldsymbol{\omega}_2 = {^2}\boldsymbol{\omega}_1 + {^2}\boldsymbol{\omega}_{2/1} \quad \rightarrow \quad {^2}\boldsymbol{\omega}_{2/1} = {^2}\boldsymbol{\omega}_2 - {^2}\boldsymbol{\omega}_1
$$

よって、以下となる。

$$
\frac{d}{dt}\left({^2}\boldsymbol{V}_{1,P} \right) - {^2}\boldsymbol{A}_{1,P} = \begin{bmatrix}
- {^2}\boldsymbol{\omega}{_{2/1}} \times {^2}\boldsymbol{\omega}_{1} \\
- {^2}\boldsymbol{\omega}_{2/1} \times {^2}\boldsymbol{v}_P 
\end{bmatrix} = - \begin{bmatrix}
[{^2}\boldsymbol{\omega}{_{2/1}}]^\land & 0 \\
0 & [{^2}\boldsymbol{\omega}_{2/1}]^\land  
\end{bmatrix}
\begin{bmatrix}
 {^2}\boldsymbol{\omega}_{1} \\ {^2}\boldsymbol{v}_P
\end{bmatrix} 
$$

### 2.3 Spatial Cross Product (並進速度がない場合)

${^2}\boldsymbol{V}_{2/1,P}$の運動について考える。<br>
関節$P$は回転関節である。すると、相対運動である$\boldsymbol{v}_{2/1, P}$は、$\boldsymbol{\omega}_{2/1}$で回転していたとしても、速度がゼロとなる。

$$
{^2}\boldsymbol{v}_{2/1, P} = 0
$$

したがって以下となる。

$$
{^2}\boldsymbol{V}_{2/1,P} = \begin{bmatrix}
{^2} \boldsymbol{\omega}_{2/1} \\
0
\end{bmatrix}
$$

この${^2}\boldsymbol{V}_{2/1,P}$の要素を用いて、Spetial Corss Productを以下で定義する。

$$
\text{crm}({^2}\boldsymbol{V}_{2/1,P}) = \begin{bmatrix}
[{^2}\boldsymbol{\omega}{_{2/1}}]^\land & 0 \\
0 & [{^2}\boldsymbol{\omega}_{2/1}]^\land
\end{bmatrix}
$$

すると、${^2}\boldsymbol{V}_{1,P}$の$\Sigma_2$での時間微分は以下で表せる。

$$
\frac{d}{dt}\left({^2}\boldsymbol{V}_{1,P} \right)= {^2}\boldsymbol{A}_{1,P} - \text{crm}({^2}\boldsymbol{V}_{2/1,P}){^2}\boldsymbol{V}_{1,P} 
$$

### 2.4 Link2 の Spatial Acceleration

Link2 の Spatial Accelerationは以下であった。

$$
{^2}{\boldsymbol{A}}_{2,P} = \frac{d}{dt}\left({^2}\boldsymbol{V}_{1,P}\right) + \frac{d}{dt}\left({^2}\boldsymbol{V}_{2/1,P}\right)
$$

上式にさきほど導出した${^2}\boldsymbol{V}_{1,P}$の$\Sigma_2$での時間微分を代入すると、以下となる。

$$
{^2}{\boldsymbol{A}}_{2,P} = {^2}\boldsymbol{A}_{1,P} - \text{crm}({^2}\boldsymbol{V}_{2/1,P}){^2}\boldsymbol{V}_{1,P} + \frac{d}{dt}\left({^2}\boldsymbol{V}_{2/1,P}\right)
$$

Spatial Cross Product は通常のベクトルの外積の性質である$a \times b = - b \times a$ と同じような、反対称性がある。

$$
\text{crm}(\boldsymbol{V})\boldsymbol{U} = - \text{crm}(\boldsymbol{U})\boldsymbol{V}
$$ 

これにより上式は

$$
{^2}{\boldsymbol{A}}_{2,P} = {^2}\boldsymbol{A}_{1,P} + \text{crm}({^2}\boldsymbol{V}_{1,P}){^2}\boldsymbol{V}_{2/1,P} + \frac{d}{dt}\left({^2}\boldsymbol{V}_{2/1,P}\right)
$$

以下の式にSpatial Corss Product Operator を適用すると、線形性より以下となる。

$$
{^2}\boldsymbol{V}_{2,P} = {^2}\boldsymbol{V}_{1,P} + {^2}\boldsymbol{V}_{2/1,P}
\quad \rightarrow \quad\text{crm}({^2}\boldsymbol{V}_{2,P}) = \text{crm}({^2}\boldsymbol{V}_{1,P}) + \text{crm}({^2}\boldsymbol{V}_{2/1,P})
$$

上式の両辺に${^2}\boldsymbol{V}_{2/1,P}$を右側から掛け、$\text{crm}(U)U =0 $の性質より、以下となる。

$$
\text{crm}({^2}\boldsymbol{V}_{2,P}){^2}\boldsymbol{V}_{2/1,P} = \text{crm}({^2}\boldsymbol{V}_{1,P}){^2}\boldsymbol{V}_{2/1,P}
$$

よって、次の関係を導くことが出来る。

$$
{^2}{\boldsymbol{A}}_{2,P} = {^2}\boldsymbol{A}_{1,P} + \frac{d}{dt}\left({^2}\boldsymbol{V}_{2/1,P}\right) + \text{crm}({^2}\boldsymbol{V}_{2,P}){^2}\boldsymbol{V}_{2/1,P} 
$$



### 2.5 一般的のSpatial Corss Product

前節では2Likkアームの回転関節を例にSpatial Cross Productを導いた。このとき、関節点を基準とする相対並進速度はゼロであったため、Spatial Cross Productには回転運動による項のみが現れた。<br>
この節では、相対運動が回転と並進の両方を含む場合を考え、一般的なSpatial Corss Product を導出する。

今回の考えるのは、次の図ような二つのBodyが節点$J_B$で接続され、Body AもBody Bも並進しながらそれぞれの角速度が生じている状況において、Body Bの点$P$を基準点とする状況である。

</p>

親座標系をBody A の$\Sigma_A$、子座標系をBody B の$\Sigma_B$とし、親Aから子Bへ伝わる運動と、子B自身の親Aからの相対運動を、同じ基準点$P$、同じ座標系$\Sigma_B$で表す。

#### 2.5.1. Spatial Motion Vectors

親の運動を表す Spatial Motion Vector $\boldsymbol{V}_A$ と子の親からの相対運動を表す Spatial Motion Vector $\boldsymbol{V}_{B/A}$ があり、それぞれ $\Sigma_B$で表され、基準点を同一の点$P$とする。そのため、特に明示しない場合、座標系は$\Sigma_B$とし、基準点は$P$であるとする。

$$
\boldsymbol{V}_A = \begin{bmatrix} \boldsymbol{\omega}_A \\ \boldsymbol{v}_A \end{bmatrix} , \quad 
\boldsymbol{V}_{B/A} = \begin{bmatrix} \boldsymbol{\omega}_{B/A} \\ \boldsymbol{v}_{B/A} \end{bmatrix}
$$

子の速度$\boldsymbol{V}_B$は以下のように二つのSpatial Motion Vectorを足した合成速度である。

$$
\boldsymbol{V}_B = \boldsymbol{V}_A + \boldsymbol{V}_{B/A} 
$$

上記を時間微分すると以下となる。

$$
\frac{d}{dt}(\boldsymbol{V}_B) = \frac{d}{dt}(\boldsymbol{V}_A) + \frac{d}{dt}(\boldsymbol{V}_{B/A}) 
$$

#### 2.5.2 親のSpatial Motion Vectorの時間微分

親のSpatial Motion Vector $\boldsymbol{V}_A$は回転と並進する子の座標 $\Sigma_B$ で表されている。そして基準点は子のある点$P$であるため、子の運動の影響が生じる。

##### 2.5.2.1 角速度 $\boldsymbol{\omega}_A$ の時間微分

角速度は次のように表される。

$$
\boldsymbol{\omega}_A = {^B}R{_A} {^A}\boldsymbol{\omega}_A
$$

これを時間微分すると、以下となる。

$$
\dot{\boldsymbol{\omega}}_A = {^B}\dot{R}{_A} {^A}{\boldsymbol{\omega}}_A + {^B}{R}{_A} {^A}\dot{\boldsymbol{\omega}}_A
$$

${^A}R{_B}$は親座標から見た子座標の相対的な回転である。$\Sigma_B$で表現された子の相対角速度$\boldsymbol{\omega}_{B/A}$は回転行列により以下のように計算される。

$$
[\boldsymbol{\omega}_{B/A}]^\land =  {^A}R{_B}^T {^A}\dot{R}{_B}
$$

上式の両辺を転置した後、右側から${^B}R{_A}$を掛けると、${^B}\dot{R}{_A}$が求まる。 

$$
- [\boldsymbol{\omega}_{B/A}]^\land = {^B}\dot{R}{_A} {^A}R{_B} \quad \rightarrow {^B}\dot{R}{_A} = - [\boldsymbol{\omega}_{B/A}]^\land {^B}R{_A}
$$

よって、$\boldsymbol{\omega}_A$の$\Sigma_B$による時間微分は以下となる。

$$
\begin{aligned}
\dot{\boldsymbol{\omega}}_A &=  {^B}{R}{_A} {^A}\dot{\boldsymbol{\omega}}_A - [\boldsymbol{\omega}_{B/A}]^\land {^B}R{_A} {^A}{\boldsymbol{\omega}}_A \\
&={\boldsymbol{\alpha}}_A - [\boldsymbol{\omega}_{B/A}]^\land {\boldsymbol{\omega}}_A
\end{aligned}
$$

##### 2.5.2.2 並進速度 $\boldsymbol{v}_A$ の時間微分

$\boldsymbol{v}_A$ 親Aの慣性座標系に対する運動を、子Bの基準点$P$で評価し、座標系$\Sigma_B$で表した並進速度成分である。模式図で描くと次の状況である。

<p align="center">
<img src="./images/parent_motion_at_child.png" style="width:30%;" />
</p>

角速度$\boldsymbol{\omega}_A$により、点$P$では$\boldsymbol{\omega}_A \times \boldsymbol{r}_{O_A P}$ が生じる。これに$\Sigma_A$の原点速度$\boldsymbol{v}_{O_A}$が加わることで、$\boldsymbol{v}_A$は以下となる。

$$
\boldsymbol{v}_A = \boldsymbol{v}_{O_A} + \boldsymbol{\omega}_{A} \times \boldsymbol{r}_{O_A P}
$$

これは、点$P$の実際の並進速度ではなく、点$P$を基準点としたSpatial Motion Vector $\boldsymbol{V}_A$ の並進速度である。上式を時間微分を行う。 

$$
\dot{\boldsymbol{v}}_A = \dot{\boldsymbol{v}}_{O_A} + \dot{\boldsymbol{\omega}}_{A} \times \boldsymbol{r}_{O_A P} + \boldsymbol{\omega}_{A} \times \dot{\boldsymbol{r}}_{O_A P}
$$

それぞれの項を計算する。

##### $\dot{\boldsymbol{v}}_{O_A}$

$\boldsymbol{v}_{O_A}$を世界座標で表すと以下である。

$$
{^0}\boldsymbol{v}_{O_A} = {^0}R{_B} {^B}\boldsymbol{v}_{O_A}
$$

輸送定理を適用すると、${^0}\boldsymbol{\omega}_{B/0}$を用いて以下となる。

$$
{^0}\dot{\boldsymbol{v}}_{O_A} = {^0}R{_B} {^B}\dot{\boldsymbol{v}}_{O_A} + {^0}\boldsymbol{\omega}_{B/0} \times {^0}\boldsymbol{v}_{O_A}
$$

左辺は${^0}{\boldsymbol{a}}_{O_A}$となる。両辺に${^B}R{_0}$を掛け、整理すると${^B}\boldsymbol{v}_{O_A}$の時間微分が求まる。

$$
{^B}\dot{\boldsymbol{v}}_{O_A} = {^B}{\boldsymbol{a}}_{O_A} - {^B}\boldsymbol{\omega}_{B/0} \times {^B}\boldsymbol{v}_{O_A}
$$

これは本節に表記に合わせると、以下のように書ける。

$$
\dot{\boldsymbol{v}}_{O_A} = {\boldsymbol{a}}_{O_A} - \boldsymbol{\omega}_{B/0} \times \boldsymbol{v}_{O_A}
$$

##### $\dot{\boldsymbol{\omega}}_{A}$

これは前節の角速度の時間微分となる。

$$
\dot{\boldsymbol{\omega}}_{A} = {\boldsymbol{\alpha}}_A - [\boldsymbol{\omega}_{B/A}]^\land {\boldsymbol{\omega}}_A
$$

##### $\dot{\boldsymbol{r}}_{O_A P}$

$\boldsymbol{r}_{O_A P}$ を世界座標$\Sigma_0$で表すと以下となる。

$$
{^0}\boldsymbol{r}_{O_A P} = {^0}R{_B} {^B}\boldsymbol{r}_{O_A P}
$$

輸送定理を適用すると、${^0}\boldsymbol{\omega}_{B/0}$を用いて以下となる。

$$
{^0}\dot{\boldsymbol{r}}_{O_A P} = {^0}R{_B} {^B}\dot{\boldsymbol{r}}_{O_A P} + {^0}\boldsymbol{\omega}_{B/0} \times {^0}\boldsymbol{r}_{O_A P}
$$

両辺に${^B}R{_0}$に掛けると$\Sigma_B$で表される。

$$
{^B}R{_0}{^0}\dot{\boldsymbol{r}}_{O_A P} = {^B}\dot{\boldsymbol{r}}_{O_A P} + {^B}\boldsymbol{\omega}_{B/0} \times {^B}\boldsymbol{r}_{O_A P}
$$

左辺の${^B}R{_0}{^0}\dot{\boldsymbol{r}}_{O_A P}$は位置ベクトルの慣性系における時間微分を$\Sigma_B$で表したものである。<br>
${^B}R{_0}{^0}\dot{\boldsymbol{r}}_{O_A P}$は$\Sigma_A$の原点$O_A$から点$P$までの距離の時間微分であるため、$O_A$からの点$P$の相対速度となる。

$$
{^B}R{_0}{^0}\dot{\boldsymbol{r}}_{O_A P} = {^B}\boldsymbol{v}_P - {^B}\boldsymbol{v}_{O_A}
$$

点$P$の実際の並進速度はBody A と Body Bの並進速度の合成であるため以下となる。

$$
{^B}\boldsymbol{v}_P = {^B}\boldsymbol{v}_A + {^B}\boldsymbol{v}_{B/A}
$$

$\boldsymbol{v}_A$は以下であった。

$$
{^B}\boldsymbol{v}_A = {^B}\boldsymbol{v}_{O_A} + {^B}\boldsymbol{\omega}_{A} \times {^B}\boldsymbol{r}_{O_A P}
$$

こららを代入していくと、

$$
\begin{array}{c}
{^B}\boldsymbol{v}_P - {^B}\boldsymbol{v}_{O_A} = {^B}\dot{\boldsymbol{r}}_{O_A P} + {^B}\boldsymbol{\omega}_{B/0} \times {^B}\boldsymbol{r}_{O_A P} \\ \downarrow \\
{^B}\boldsymbol{v}_A + {^B}\boldsymbol{v}_{B/A} - {^B}\boldsymbol{v}_{O_A} = {^B}\dot{\boldsymbol{r}}_{O_A P} + {^B}\boldsymbol{\omega}_{B/0} \times {^B}\boldsymbol{r}_{O_A P} \\ \downarrow \\
\cancel{{^B}\boldsymbol{v}_{O_A}} + {^B}\boldsymbol{\omega}_{A} \times {^B}\boldsymbol{r}_{O_A P}+ {^B}\boldsymbol{v}_{B/A} - \cancel{{^B}\boldsymbol{v}_{O_A}} = {^B}\dot{\boldsymbol{r}}_{O_A P} + {^B}\boldsymbol{\omega}_{B/0} \times {^B}\boldsymbol{r}_{O_A P} \\ \downarrow \\
{^B}\dot{\boldsymbol{r}}_{O_A P} = {^B}\boldsymbol{v}_{B/A} + {^B}\boldsymbol{\omega}_{A} \times {^B}\boldsymbol{r}_{O_A P} - {^B}\boldsymbol{\omega}_{B/0} \times {^B}\boldsymbol{r}_{O_A P} \\ \downarrow \\
{^B}\dot{\boldsymbol{r}}_{O_A P} = {^B}\boldsymbol{v}_{B/A} + ({^B}\boldsymbol{\omega}_{A} - {^B}\boldsymbol{\omega}_{B/0}) \times {^B}\boldsymbol{r}_{O_A P}
\end{array}
$$
 
角速度は合成角速度となるので、

$$
{^B}\boldsymbol{\omega}_{B/0} = {^B}\boldsymbol{\omega}_{A} + {^B}\boldsymbol{\omega}_{B/A} \quad \rightarrow \quad {^B}\boldsymbol{\omega}_{B/A} =  {^B}\boldsymbol{\omega}_{B/0} - {^B}\boldsymbol{\omega}_{A}
$$

よって、以下となる。

$$
{^B}\dot{\boldsymbol{r}}_{O_A P} = {^B}\boldsymbol{v}_{B/A} - {^B}\boldsymbol{\omega}_{B/A} \times {^B}\boldsymbol{r}_{O_A P}
$$

本節の表記に合わせると、以下となる。

$$
\dot{\boldsymbol{r}}_{O_A P} = \boldsymbol{v}_{B/A} - \boldsymbol{\omega}_{B/A} \times \boldsymbol{r}_{O_A P}
$$

##### 並進速度の時間微分

$$
\begin{aligned}
\dot{\boldsymbol{v}}_A &=  {\boldsymbol{a}}_{O_A} - \boldsymbol{\omega}_{B/0} \times \boldsymbol{v}_{O_A} + ({\boldsymbol{\alpha}}_A - [\boldsymbol{\omega}_{B/A}]^\land {\boldsymbol{\omega}}_A) \times \boldsymbol{r}_{O_A P} + \boldsymbol{\omega}_{A} \times ( \boldsymbol{v}_{B/A} - \boldsymbol{\omega}_{B/A} \times \boldsymbol{r}_{O_A P}) \\
&= {\boldsymbol{a}}_{O_A} - \boldsymbol{\omega}_{B/0} \times \boldsymbol{v}_{O_A} + {\boldsymbol{\alpha}}_A \times \boldsymbol{r}_{O_A P} + ({\boldsymbol{\omega}}_A \times \boldsymbol{\omega}_{B/A}) \times \boldsymbol{r}_{O_A P} + \boldsymbol{\omega}_{A} \times \boldsymbol{v}_{B/A} - \boldsymbol{\omega}_{A} \times (\boldsymbol{\omega}_{B/A} \times \boldsymbol{r}_{O_A P})
\end{aligned}
$$

上式に$\boldsymbol{\omega}_{B/0} = \boldsymbol{\omega}_{A} + \boldsymbol{\omega}_{B/A}$ を代入すると、

$$
\begin{aligned}
\dot{\boldsymbol{v}}_A
&= {\boldsymbol{a}}_{O_A} - (\boldsymbol{\omega}_{A} + \boldsymbol{\omega}_{B/A}) \times \boldsymbol{v}_{O_A} + {\boldsymbol{\alpha}}_A \times \boldsymbol{r}_{O_A P} + ({\boldsymbol{\omega}}_A \times \boldsymbol{\omega}_{B/A}) \times \boldsymbol{r}_{O_A P} + \boldsymbol{\omega}_{A} \times \boldsymbol{v}_{B/A} - \boldsymbol{\omega}_{A} \times (\boldsymbol{\omega}_{B/A} \times \boldsymbol{r}_{O_A P})
\end{aligned}
$$

ベクトル三重積の公式 $A \times (B \times C) = (A^T C)B - (A^T B)C, (A \times B) \times C = (A^T C)B - (B^T C)A$ を適用すると、次のようになる。

$$
({\boldsymbol{\omega}}_A \times \boldsymbol{\omega}_{B/A}) \times \boldsymbol{r}_{O_A P} = ({\boldsymbol{\omega}}_A^T\boldsymbol{r}_{O_A P})\boldsymbol{\omega}_{B/A} - (\boldsymbol{\omega}_{B/A}^T\boldsymbol{r}_{O_A P}) {\boldsymbol{\omega}}_A
$$

$$
\boldsymbol{\omega}_{A} \times (\boldsymbol{\omega}_{B/A} \times \boldsymbol{r}_{O_A P}) = (\boldsymbol{\omega}_{A}^T \boldsymbol{r}_{O_A P})\boldsymbol{\omega}_{B/A} - (\boldsymbol{\omega}_{A}^T\boldsymbol{\omega}_{B/A})\boldsymbol{r}_{O_A P}
$$

これにより、ベクトル三重積の項は以下となる。

$$
- (\boldsymbol{\omega}_{B/A}^T\boldsymbol{r}_{O_A P}) {\boldsymbol{\omega}}_A + (\boldsymbol{\omega}_{A}^T\boldsymbol{\omega}_{B/A})\boldsymbol{r}_{O_A P}
$$

内積は順番を入れ替えても同じであり、

$$
 (\boldsymbol{\omega}_{B/A}^T \boldsymbol{\omega}_{A})\boldsymbol{r}_{O_A P} - (\boldsymbol{\omega}_{B/A}^T\boldsymbol{r}_{O_A P}) {\boldsymbol{\omega}}_A 
$$

ベクトル三重積のABCに以下と当てはめると、

- $\boldsymbol{\omega}_{B/A}$ : A
- $\boldsymbol{\omega}_{A}$ : C
- $\boldsymbol{r}_{O_A P}$ : B

次のように整理できる。

$$
\boldsymbol{\omega}_{B/A} \times (\boldsymbol{r}_{O_A P} \times \boldsymbol{\omega}_{A})
$$

これより、

$$
\begin{aligned}
\dot{\boldsymbol{v}}_A
&= {\boldsymbol{a}}_{O_A} - (\boldsymbol{\omega}_{A} + \boldsymbol{\omega}_{B/A}) \times \boldsymbol{v}_{O_A} + {\boldsymbol{\alpha}}_A \times \boldsymbol{r}_{O_A P} + \boldsymbol{\omega}_{A} \times \boldsymbol{v}_{B/A} +\boldsymbol{\omega}_{B/A} \times (\boldsymbol{r}_{O_A P} \times \boldsymbol{\omega}_{A})\\
&= {\boldsymbol{a}}_{O_A} - \boldsymbol{\omega}_{A} \times \boldsymbol{v}_{O_A}  + {\boldsymbol{\alpha}}_A \times \boldsymbol{r}_{O_A P} + \boldsymbol{\omega}_{A} \times \boldsymbol{v}_{B/A} - \boldsymbol{\omega}_{B/A} \times \boldsymbol{v}_{O_A} - \boldsymbol{\omega}_{B/A} \times (\boldsymbol{\omega}_{A} \times \boldsymbol{r}_{O_A P} ) \\
&= {\boldsymbol{a}}_{O_A} - \boldsymbol{\omega}_{A} \times \boldsymbol{v}_{O_A}  + {\boldsymbol{\alpha}}_A \times \boldsymbol{r}_{O_A P} + \boldsymbol{\omega}_{A} \times \boldsymbol{v}_{B/A} - \boldsymbol{\omega}_{B/A} \times (\boldsymbol{v}_{O_A} + \boldsymbol{\omega}_{A} \times \boldsymbol{r}_{O_A P} )\\
\end{aligned}
$$

$\boldsymbol{v}_A = \boldsymbol{v}_{O_A} + \boldsymbol{\omega}_{A} \times \boldsymbol{r}_{O_A P}$ より、親Aの並進速度の時間微分は以下となる。

$$
\begin{aligned}
\dot{\boldsymbol{v}}_A
&={\boldsymbol{a}}_{O_A} - \boldsymbol{\omega}_{A} \times \boldsymbol{v}_{O_A}  + {\boldsymbol{\alpha}}_A \times \boldsymbol{r}_{O_A P} -  \boldsymbol{v}_{B/A} \times \boldsymbol{\omega}_{A} - \boldsymbol{\omega}_{B/A} \times \boldsymbol{v}_{A}
\end{aligned}
$$


### 2.5.3 親のSpatial Accelerationとの差

親AのSpatial Accelration を$\Sigma_B$座標で基準点$P$で表すと以下である。

$$
\boldsymbol{A}_A = \begin{bmatrix}
\boldsymbol{\alpha}_A \\
\boldsymbol{a}_A - \boldsymbol{\omega}_A \times \boldsymbol{v}_A
\end{bmatrix}
$$

ここで$\boldsymbol{v}_A$は親AのMotion Vectorを点$P$で評価したものあったので、$\boldsymbol{a}_A$も親Aの加速度を点$P$で評価する必要がある。Spatial Accelerationは親A自身の加速度を表すものである。したがって、子Bの相対運動を含めずに、親Aに固定された仮想点$Q$の加速度を用いる。

親Aに固定された仮想的な点$Q$を考える。このとき、点$Q$の加速度は$\Sigma_0$で示した以下の位置ベクトルの2回時間微分となる。

- 位置

    $$
    {^0}\boldsymbol{r}_Q = {^0}\boldsymbol{r}_{O_A} + {^0}R{_A} {^A} \boldsymbol{r}_{O_A Q}
    $$

- 速度

    $$
    \begin{aligned}
    {^0}\boldsymbol{v}_Q &= {^0}\dot{\boldsymbol{r}}_{O_A} + {^0}\dot{R}{_A} {^A}\boldsymbol{r}_{O_A Q} + {^0}R{_A} {^A}\dot{\boldsymbol{r}}_{O_A Q} 
    \end{aligned}
    $$

    ${^0}\dot{R}{_A} = [{^0}\boldsymbol{\omega}_A]^\land {^0}R{_A}$、${^A}\dot{\boldsymbol{r}}_{O_A Q} =0$ より

    $$
    \begin{aligned}
    {^0}\boldsymbol{v}_Q &= {^0}{\boldsymbol{v}}_{O_A} + [{^0}\boldsymbol{\omega}_A]^\land {^0}\boldsymbol{r}_{O_A Q} 
    \end{aligned}
    $$

- 加速度

    $$
    \begin{aligned}
    {^0}\boldsymbol{a}_Q &= {^0}\dot{\boldsymbol{v}}_{O_A} + {^0}\dot{\boldsymbol{\omega}}_A \times {^0}\boldsymbol{r}_{O_A Q} + {^0}\boldsymbol{\omega}_A \times {^0}\dot{\boldsymbol{r}}_{O_A Q} 
    \end{aligned}
    $$

    ${^0}\dot{\boldsymbol{r}}_{O_A Q} = {^0}\dot{R}{_A} {^A}\boldsymbol{r}_{O_A Q} = {^0}\boldsymbol{\omega}_A \times {^0}\boldsymbol{r}_{O_A Q}$より

    $$
    \begin{aligned}
    {^0}\boldsymbol{a}_Q &= {^0}{\boldsymbol{a}}_{O_A} + {^0}{\boldsymbol{\alpha}}_A \times {^0}\boldsymbol{r}_{O_A Q} + {^0}\boldsymbol{\omega}_A \times ({^0}\boldsymbol{\omega}_A \times {^0}\boldsymbol{r}_{O_A Q}) 
    \end{aligned}
    $$


この仮想点$Q$がある瞬間に子B上の点$P$と同じ位置にあるとする。このとき、点$Q$の加速度を親Aの加速度場の点$P$における値$\boldsymbol{a}_A$として用いる。ただし、点$P$自身の実際の加速度とは異なる。

$$
{^0}\boldsymbol{a}_A = {^0}\boldsymbol{a}_{O_A} + {^0}{\boldsymbol{\alpha}}_{A} \times {^0} \boldsymbol{r}_{O_A P} + {^0}\boldsymbol{\omega}_{A}  \times ( {^0}\boldsymbol{\omega}_{A} \times {^0} {\boldsymbol{r}}_{O_A P})
$$

上式を$\Sigma_B$で表すと以下となる。

$$
\boldsymbol{a}_A = \boldsymbol{a}_{O_A} + {\boldsymbol{\alpha}}_{A} \times \boldsymbol{r}_{O_A P} + \boldsymbol{\omega}_{A}  \times ( \boldsymbol{\omega}_{A} \times {\boldsymbol{r}}_{O_A P})
$$


これより、$\boldsymbol{v}_A = \boldsymbol{v}_{O_A} + \boldsymbol{\omega}_{A} \times \boldsymbol{r}_{O_A P}$を用いて、

$$
\begin{aligned}
\boldsymbol{a}_A - \boldsymbol{\omega}_A \times \boldsymbol{v}_A &= \boldsymbol{a}_{O_A} + {\boldsymbol{\alpha}}_{A} \times \boldsymbol{r}_{O_A P} + \boldsymbol{\omega}_{A}  \times ( \boldsymbol{\omega}_{A} \times  {\boldsymbol{r}}_{O_A P})- \boldsymbol{\omega}_A \times (\boldsymbol{v}_{O_A} + \boldsymbol{\omega}_{A} \times \boldsymbol{r}_{O_A P}) \\
&= \boldsymbol{a}_{O_A} + {\boldsymbol{\alpha}}_{A} \times \boldsymbol{r}_{O_A P} - \boldsymbol{\omega}_A \times \boldsymbol{v}_{O_A}
\end{aligned}
$$ 

よって、Spatial Acclerationは以下となる。

$$
\boldsymbol{A}_A = \begin{bmatrix}
\boldsymbol{\alpha}_A \\
\boldsymbol{a}_{O_A} + {\boldsymbol{\alpha}}_{A} \times \boldsymbol{r}_{O_A P} - \boldsymbol{\omega}_A \times \boldsymbol{v}_{O_A}
\end{bmatrix}
$$


親AのSpatial Motion Vectorを$\Sigma_B$で時間微分したものは以下であった。

$$
\frac{d}{dt}(\boldsymbol{V}_A) = \frac{d}{dt}\begin{bmatrix} \boldsymbol{\omega}_A \\ \boldsymbol{v}_A \end{bmatrix} = 
\begin{bmatrix}
{\boldsymbol{\alpha}}_A - [\boldsymbol{\omega}_{B/A}]^\land {\boldsymbol{\omega}}_A \\
{\boldsymbol{a}}_{O_A} - \boldsymbol{\omega}_{A} \times \boldsymbol{v}_{O_A}  + {\boldsymbol{\alpha}}_A \times \boldsymbol{r}_{O_A P} -  \boldsymbol{v}_{B/A} \times \boldsymbol{\omega}_{A} - \boldsymbol{\omega}_{B/A} \times \boldsymbol{v}_{A}
\end{bmatrix}
$$

両式の差を計算すると次となる。

$$
\frac{d}{dt}(\boldsymbol{V}_A) - \boldsymbol{A}_A = \begin{bmatrix}
- [\boldsymbol{\omega}_{B/A}]^\land {\boldsymbol{\omega}}_A \\
-  [\boldsymbol{v}_{B/A}]^\land \boldsymbol{\omega}_{A} - [\boldsymbol{\omega}_{B/A}]^\land \boldsymbol{v}_{A}
\end{bmatrix} = - \begin{bmatrix}
[\boldsymbol{\omega}_{B/A}]^\land & 0 \\
[\boldsymbol{v}_{B/A}]^\land & [\boldsymbol{\omega}_{B/A}]^\land
\end{bmatrix}
\begin{bmatrix} {\boldsymbol{\omega}}_A \\ \boldsymbol{v}_{A} \end{bmatrix}
$$

#### 2.5.4 Spatial Corss Product

親Aからの相対運動である $\boldsymbol{V}_{B/A}$ は以下であった。

$$
\boxed{
\boldsymbol{V}_{B/A} = \begin{bmatrix} \boldsymbol{\omega}_{B/A} \\ \boldsymbol{v}_{B/A} \end{bmatrix}
}
$$

$\boldsymbol{V}_{B/A}$ の要素を用いて、Spatial Cross Product を以下で定義する。

$$
\boxed{
\text{crm}(\boldsymbol{V}_{B/A}) = \begin{bmatrix}
[\boldsymbol{\omega}_{B/A}]^\land & 0 \\
[\boldsymbol{v}_{B/A}]^\land & [\boldsymbol{\omega}_{B/A}]^\land
\end{bmatrix}
}
$$

すると、$\boldsymbol{V}_A$ の $\Sigma_B$ での時間微分は以下で表せる。

$$
\frac{d}{dt}(\boldsymbol{V}_A) = \boldsymbol{A}_A - \text{crm}(\boldsymbol{V}_{B/A})\boldsymbol{V}_A
$$

#### 2.5.5 子BのSpatial Acceleration

子の速度$\boldsymbol{V}_B$は以下のように二つのSpatial Motion Vectorを足した合成速度である。

$$
\boldsymbol{V}_B = \boldsymbol{V}_A + \boldsymbol{V}_{B/A} 
$$

上記を時間微分すると以下となる。

$$
\frac{d}{dt}(\boldsymbol{V}_B) = \boldsymbol{A}_B = \frac{d}{dt}(\boldsymbol{V}_A) + \frac{d}{dt}(\boldsymbol{V}_{B/A}) 
$$

2.5.4. で導出した$\boldsymbol{V}_A$ の $\Sigma_B$ での時間微分を代入すると以下となる。

$$
\boldsymbol{A}_B = \boldsymbol{A}_A - \text{crm}(\boldsymbol{V}_{B/A})\boldsymbol{V}_A + \frac{d}{dt}(\boldsymbol{V}_{B/A}) 
$$

Spatial Cross Product の反対称性 $\text{crm}(V)U = -\text{crm}(U)V$ より、以下となる。  

$$
\boldsymbol{A}_B = \boldsymbol{A}_A + \text{crm}(\boldsymbol{V}_{A})\boldsymbol{V}_{B/A} + \frac{d}{dt}(\boldsymbol{V}_{B/A}) 
$$

以下の式にSpatial Cross Product Operator を適用すると、線形性より以下となる。

$$
\boldsymbol{V}_B = \boldsymbol{V}_A + \boldsymbol{V}_{B/A} \rightarrow \text{crm}(\boldsymbol{V}_B) = \text{crm}(\boldsymbol{V}_A) + \text{crm}(\boldsymbol{V}_{B/A})
$$

上式の両辺に　$\boldsymbol{V}_{B/A}$ を右側から掛け、$\text{crm}(U)U=0$の性質より、以下となる。

$$
\text{crm}(\boldsymbol{V}_B)\boldsymbol{V}_{B/A} = \text{crm}(\boldsymbol{V}_A)\boldsymbol{V}_{B/A}
$$

よって、次の関係を導くことが出来る。

$$
\boxed{
\boldsymbol{A}_B = \boldsymbol{A}_A + \frac{d}{dt}(\boldsymbol{V}_{B/A}) + \text{crm}(\boldsymbol{V}_B)\boldsymbol{V}_{B/A}
}
$$


## 3. Spatial Cross Product for Force

### 3.1. 定義
crmはSpatial Motion Vector に作用する演算子(Operator)である。

$$
\text{crm}(\boldsymbol{V}) = \begin{bmatrix}
[\boldsymbol{\omega}]^\land & 0 \\
[\boldsymbol{v}]^\land & [\boldsymbol{\omega}]^\land
\end{bmatrix}
$$

一方、crfはSpatial Force Vector に作用する演算子として、以下で定義される。

$$
\text{crf}(V) = - \text{crm}(V)^T , \quad \text{crf}(V) = \begin{bmatrix}
[\boldsymbol{\omega}]^\land & [\boldsymbol{v}]^\land  \\
0 & [\boldsymbol{\omega}]^\land
\end{bmatrix}
$$

### 3.2. Motion Vector と Force Vector の関係

一般の仕事率(Power)は単位時間当たりの仕事で $F x /t = F v$ と計算できる。Spation Vector も同様に内積を用いて仕事率を計算できる。

$$
P = \boldsymbol{f}^T \boldsymbol{V}
$$

Motion Vector $\boldsymbol{V}$ が座標変換によって $\boldsymbol{V}^{\prime}$ へ変化したとする。

$$
\boldsymbol{V}^{\prime} = X \boldsymbol{V}
$$

仕事率は座標系によって変化しないため、以下の関係となる。

$$
{\boldsymbol{f}^{\prime}}^T \boldsymbol{V}^{\prime} = \boldsymbol{f}^T \boldsymbol{V}
$$

これより、Spatial Force Vector の変換行列は Spatial Motion Vector の変換行列の逆行列の転置となる。

$$
({X^{-T}\boldsymbol{f}})^T (X\boldsymbol{V}) = \boldsymbol{f}^T \boldsymbol{V} , \quad \boldsymbol{f}^{\prime} = {X^{-T}\boldsymbol{f}}
$$

### 3.3. 微小変換

#### 3.3.1 通常の3次元回転行列の場合

$z$軸周りの回転行列は以下である。

$$
R(\theta) = \begin{bmatrix}
\cos(\theta) & - \sin(\theta) & 0 \\
\sin(\theta) & \cos(\theta) & 0 \\
0 & 0 & 1 
\end{bmatrix}
$$

$\theta = \varepsilon$ と非常に小さな回転を与えた場合、三角関数は以下となる。

$$
\sin(\varepsilon) \approx \varepsilon , \quad \cos(\varepsilon) \approx 1
$$

よって、回転行列は次のようになる。

$$
R(\varepsilon) \approx \begin{bmatrix}
1 & - \varepsilon & 0 \\
\varepsilon & 1 & 0 \\
0 & 0 & 1 
\end{bmatrix}
$$

上式を分解すると、以下となる。

$$
R(\varepsilon) \approx 
\boldsymbol{I} + 
\varepsilon\begin{bmatrix}
0 & - 1 & 0 \\
1 & 0 & 0 \\
0 & 0 & 0 
\end{bmatrix}
$$

$e_z = [0, 0, 1]^T = [e_{zx}, e_{zy}, e_{zz}]^T$ であり、この歪対称行列は以下である。

$$
[e_z]^\land = \begin{bmatrix}
0 & -e_{zz} & e_{zy} \\
e_{zz} & 0 & -e_{zx} \\
-e_{zy} & e_{zx} & 0 
\end{bmatrix} = 
\begin{bmatrix}
0 & -1 & 0 \\
1 & 0 & 0 \\
0 & 0 & 0
\end{bmatrix}
$$

よって、ある変換の微小変換は以下のように記述することが出来る。

$$
R(\varepsilon) = \boldsymbol{I} + \varepsilon [e_z]^\land + O(\varepsilon^2)
$$

#### 3.3.2 回転行列の微小時間変化

回転行列の時間微分は次のように角速度の歪対称行列との積となる。

$$
\dot{R}(t) = [\boldsymbol{\omega}(t)]^\land R(t)
$$

角速度そのものは変換行列ではないが、角速度は回転行列の時間変化量を決めている。

微小時間$\Delta t$ では以下の回転量変化となる。

$$
R(t + \Delta t) \approx (\boldsymbol{I} + \Delta t [\boldsymbol{\omega}]^\land)R(t)
$$

よって、以下が微小な回転変換を表す。

$$
\boldsymbol{I} + \Delta t [\boldsymbol{\omega}]^\land
$$

R(0) = $\boldsymbol{I}$であるとき、微小変換量は以下となる。

$$
R(\Delta t) \approx \boldsymbol{I} + \Delta t [\boldsymbol{\omega}]^\land
$$

#### 3.3.3 Spatial Motion Vectorの座標変換行列

$$
\boldsymbol{V}^{\prime} = X \boldsymbol{V}
$$

以上の座標変換行列$X$は表現座標系の変換と基準点を変換するものとして、以下である。

$$
X = \begin{bmatrix}
R & 0 \\
-[\boldsymbol{r}]^\land R & R
\end{bmatrix}
$$

ここで$X(0) = \boldsymbol{I}_6$ である微小変化量を考えると、変換行列のそれぞれの項を以下のように置く。

$$
\begin{aligned}
R(\varepsilon) &= \boldsymbol{I}_3 + \varepsilon [\boldsymbol{\omega}]^T + O(\varepsilon^2) \\
\boldsymbol{r}(\varepsilon) &= \varepsilon \boldsymbol{u} + O(\varepsilon^2)
\end{aligned}
$$

これを変換行列に代入すると、以下となる。

$$
X(\varepsilon) = \begin{bmatrix}
\boldsymbol{I}_3 + \varepsilon [\boldsymbol{\omega}]^T & 0 \\
- \varepsilon [\boldsymbol{u}]^\land (\boldsymbol{I}_3 + \varepsilon [\boldsymbol{\omega}]^T) & \boldsymbol{I}_3 + \varepsilon [\boldsymbol{\omega}]^T
\end{bmatrix} + O(\varepsilon^2)
$$

$\varepsilon$の2次以上の項を$O(\varepsilon)$とすると、Xの微小変換は以下となる。

$$
X(\varepsilon) = \boldsymbol{I}_6 + 
\varepsilon
\begin{bmatrix}
 [\boldsymbol{\omega}]^T & 0 \\
- [\boldsymbol{u}]^\land & [\boldsymbol{\omega}]^T
\end{bmatrix} + O(\varepsilon^2)
$$

### 3.4 crm との関連

crm はSpatial Motion Vector $\boldsymbol{V}$ の要素を用いて以下であった。

$$
\text{crm}(\boldsymbol{V}) = \begin{bmatrix}
[\boldsymbol{\omega}]^\land & 0 \\
[\boldsymbol{v}]^\land & [\boldsymbol{\omega}]^\land
\end{bmatrix}
$$

上記の微小変換の並進成分の速度について

$$
\boldsymbol{u} = -\boldsymbol{v}
$$

となる向きの変換を考えれば、以下となる。

$$
X(\varepsilon) = \boldsymbol{I}_6 + 
\varepsilon
\begin{bmatrix}
 [\boldsymbol{\omega}]^T & 0 \\
[\boldsymbol{v}]^\land & [\boldsymbol{\omega}]^T
\end{bmatrix} + O(\varepsilon^2)
$$

よって、Spatial Motion Vector側の微小変換はcrmを用いて以下で記述できる。

$$
X(\varepsilon) = \boldsymbol{I}_6 + \varepsilon \ \text{crm}(\boldsymbol{V}) + O(\varepsilon^2)
$$

### 3.4. crfの導出

3.2 では Force Vectorの変換を次のように導いた。

$$
\boldsymbol{f}^\prime = X^{-T} \boldsymbol{f}
$$

これより、 $X^{-T}$の微小変換を $X(\varepsilon)$から求める。

$$
A = \boldsymbol{I}_6 - \varepsilon \ \text{crm}(\boldsymbol{V}) + O(\varepsilon^2)
$$

とすると、以下となるので、$A = X(\varepsilon)^{-1}$である。

$$
\begin{aligned}
X(\varepsilon) A &= 
(\boldsymbol{I}_6 + \varepsilon \ \text{crm}(\boldsymbol{V}) + O(\varepsilon^2)) 
(\boldsymbol{I}_6 - \varepsilon \ \text{crm}(\boldsymbol{V}) + O(\varepsilon^2)) \\ &=
(\boldsymbol{I}_6 + \varepsilon \ \text{crm}(\boldsymbol{V}))(\boldsymbol{I}_6 - \varepsilon \ \text{crm}(\boldsymbol{V})) + O(\varepsilon^2) \\ &=
\boldsymbol{I}_6 - \varepsilon^2 \text{crm}^2 + O(\varepsilon^2)  \\
&= \boldsymbol{I}_6 + O(\varepsilon^2)  \\
\end{aligned}
$$

$X(\varepsilon)^{-1}$ の転置を行うと以下となる。

$$
X(\varepsilon)^{-T} = \boldsymbol{I}_6 - \varepsilon \ \text{crm}(\boldsymbol{V})^T + O(\varepsilon^2)
$$

crf($\boldsymbol{V}$) = - crm $(\boldsymbol{V})^T$ と定義すると、Spatial Force Vector 側の微小変換は以下のように表すことが出来る。

$$
X(\varepsilon)^{-T} = \boldsymbol{I}_6 + \varepsilon \ \text{crf}(\boldsymbol{V})^T + O(\varepsilon^2)
$$

これより、Spatial Force Vector に関してのCross Product は以下となる。

$$
\text{crf}(\boldsymbol{V}) = \begin{bmatrix}
[\boldsymbol{\omega}]^\land & [\boldsymbol{v}]^\land \\
0 & [\boldsymbol{\omega}]^\land
\end{bmatrix}
$$

この $\text{crf}(\boldsymbol{V})$ は Spatial Force Vector に対して作用するため、

$$
\boldsymbol{f} = \begin{bmatrix} \boldsymbol{n} \\ \boldsymbol{F} \end{bmatrix}
$$

とすれば以下となる。

$$
\text{crf}(\boldsymbol{V})\boldsymbol{f} = \begin{bmatrix}
[\boldsymbol{\omega}]^\land & [\boldsymbol{v}]^\land \\
0 & [\boldsymbol{\omega}]^\land
\end{bmatrix}
\begin{bmatrix} \boldsymbol{n} \\ \boldsymbol{F} \end{bmatrix} =
\begin{bmatrix}
[\boldsymbol{\omega}]^\land \ \boldsymbol{n} + [\boldsymbol{v}]^\land \ \boldsymbol{F} \\
[\boldsymbol{\omega}]^\land \ \boldsymbol{F}
\end{bmatrix}
$$

一行目はモーメント成分の輸送項、二行目は力成分の輸送項である。

$\boldsymbol{\omega}$で回転する座標系にあるベクトル$\boldsymbol{a}$の時間微分は輸送定理として以下であった。

$$
{^0}\dot{\boldsymbol{a}} = {^0}R_1 {^1}\boldsymbol{a} + {^0}\boldsymbol{\omega} \times {^0}\boldsymbol{a}
$$

輸送定理の2項目は、座標系が回転することによって生じる変化であり、これを輸送項と呼ぶ。
