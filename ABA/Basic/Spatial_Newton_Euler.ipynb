# Spatial Newton Euler Method (空間ニュートン・オイラー方程式)

Spatial Newton Euler Methodは以下の方程式式である。

$$
\boldsymbol{f} = \boldsymbol{I} \boldsymbol{a} + \text{crf}(\boldsymbol{V}) \boldsymbol{I} \boldsymbol{V}
$$

- $\boldsymbol{f}$ : Spatial Force Vector
- $\boldsymbol{I}$ : Spatial Inertia
- $\boldsymbol{a}$ : Spatial Acceleration
- $\boldsymbol{V}$ : Spatial Motion Vector
- $\text{crf}(\boldsymbol{V})$ : Spatial Force Cross Product

この式は、剛体に作用する合力・合モーメントを表す運動方程式である。

本Noteでは従来の方程式からSpatial Newton Euler Method の導出を行う。


## 1. 従来のNewton-Euler 方程式

剛体の重心$G$を基準点とする。

- 並進運動

    $$
    \boldsymbol{F} = m \boldsymbol{a}_G
    $$

    - $m$: 剛体の質量
    - $\boldsymbol{a}_G$ : 重心の加速度ベクトル

- [回転運動](../../3D-EoM/Basic/euler-equation-of-motion.ipynb)

    $$
    \boldsymbol{n}_G = \boldsymbol{I}_G \dot{\boldsymbol{\omega}} + \boldsymbol{\omega} \times (\boldsymbol{I}_G\boldsymbol{\omega})
    $$

    - $\boldsymbol{I}_G$ : 重心まわりの慣性テンソル
    - $\dot{\boldsymbol{\omega}}$ : 剛体の角加速度ベクトル
    - ${\boldsymbol{\omega}}$ : 剛体の角速度ベクトル

上式をまとめることで、Spatial Newton-Euler 方程式を構築することが出来る。


## 2. Newton-Euler 方程式と運動量の時間微分

重心の並進の運動量$\boldsymbol{p}$は質量と速度の積で求まる。そして、運動量の時間微分が並進の運動方程式となる。

$$
\boldsymbol{p} = m \boldsymbol{v}_G \rightarrow \boldsymbol{F} = \frac{d \ \boldsymbol{p}}{dt}
$$

重心まわりの角運動量は慣性テンソルと角速度の積で求まる。そして、角運動量の時間微分が回転の運動方程式となる。

$$
\boldsymbol{L}_G = \boldsymbol{I}_G \boldsymbol{\omega} \rightarrow \boldsymbol{n}_G = \frac{d \ \boldsymbol{L}_G}{dt}
$$

### 2.1 角運動量の座標変換と輸送定理

剛体座標系$\Sigma_B$の角運動量が回転行列を用いて以下のように世界座標$\Sigma_W$で表されるとする。

$$
{^W}\boldsymbol{L}_G = {^W}R{_B} {^B}\boldsymbol{L}_G 
$$ 

上式を時間微分すると、積の微分則より、以下となる。

$$
{^W}\dot{\boldsymbol{L}}_G = {^W}\dot{R}{_B} {^B}\boldsymbol{L}_G + {^W}R{_B} {^B}\dot{\boldsymbol{L}}_G 
$$ 

回転行列の時間微分は角速度を剛体座標系で表すと以下となる。

$$
{^W}\dot{R}{_B} = {^W}{R}{_B} [{^B}\boldsymbol{\omega}_{B/W}]^\land
$$

上式を代入すると、角運動量の世界座標系における時間微分は以下となる。これが角運動に対する輸送定理である。

$$
\begin{aligned}
{^W}\dot{\boldsymbol{L}}_G &= {^W}{R}{_B} [{^B}\boldsymbol{\omega}_{B/W}]^\land {^B}\boldsymbol{L}_G + {^W}R{_B} {^B}\dot{\boldsymbol{L}}_G \\
&= {^W}{R}{_B} \left({^B}\dot{\boldsymbol{L}}_G + {^B}\boldsymbol{\omega}_{B/W} \times {^B}\boldsymbol{L}_G \right)\\
\end{aligned}
$$ 

上式の左右に${^B}{R}{_W}$を掛けると、以下のようになる。

$$
{^B}{R}{_W} {^W}\dot{\boldsymbol{L}}_G = {^B}\dot{\boldsymbol{L}}_G + {^B}\boldsymbol{\omega}_{B/W} \times {^B}\boldsymbol{L}_G
$$


### 2.2 Newton-Euler 方程式との接続

剛体座標系$\Sigma_B$では、重心まわりの慣性テンソルは一定である。したがって、${^B}\boldsymbol{L}_G$の時間微分は次となる。

$$
{^B}\boldsymbol{L}_G = {^B}\boldsymbol{I}_G {^B}\boldsymbol{\omega}_{B/W} \rightarrow {^B}\dot{\boldsymbol{L}}_G = {^B}\boldsymbol{I}_G {^B}\dot{\boldsymbol{\omega}}_{B/W}
$$ 

先ほどの式に代入すると以下となる。

$$
{^W}\dot{\boldsymbol{L}}_G = {^W}{R}{_B} \left( {^B}\boldsymbol{I}_G {^B}\dot{\boldsymbol{\omega}}_{B/W} + {^B}\boldsymbol{\omega}_{B/W} \times ({^B}\boldsymbol{I}_G {^B}\boldsymbol{\omega}_{B/W}) \right)
$$

剛体座標系で表すために上式の両辺に${^B}{R}{_W}$をかける。$\dot{\boldsymbol{L}}_G = \boldsymbol{n}_G$であるため、剛体座標系におけるNewton-Euler方程式となる。

$$
{^B}{\boldsymbol{n}}_G = {^B}\boldsymbol{I}_G {^B}\dot{\boldsymbol{\omega}}_{B/W} + {^B}\boldsymbol{\omega}_{B/W} \times ({^B}\boldsymbol{I}_G {^B}\boldsymbol{\omega}_{B/W})
$$


## 3. Spatiam Momentum とその時間微分

通常のNewton-Euler方程式は、角運動量の時間微分と輸送定理を接続することにより導出できた。Spatial Vectorについても同様に運動量であるSpatial Momentum の時間微分と輸送定理を接続を行い、Newton-Euler方程式を導出する。

[Spatial VectorのNote](./spatial_vector.ipynb)の4.1 節より、Spatial Momentum は 並進運動量と角運動量をまとめた以下のものである。

$$
\boldsymbol{H}_G = \begin{bmatrix} \boldsymbol{L}_G \\ \boldsymbol{p} \end{bmatrix} 
= \begin{bmatrix} \boldsymbol{I}_G \boldsymbol{\omega} \\ m \boldsymbol{v}_G \end{bmatrix}
$$

剛体の任意の基準点$O$まわりの角運動量は次の式である。

$$
\begin{aligned}
\boldsymbol{L}_O &= \boldsymbol{I}_G \boldsymbol{\omega} + \boldsymbol{r}_{O G} \times (m \boldsymbol{v}_G ) \\
&= \boldsymbol{L}_G + \boldsymbol{r}_{O G} \times \boldsymbol{p}
\end{aligned}
$$

並進運動量はどの評価点でも変化しないため、基準点$O$周りの Spatial Momentum は以下となる。

$$
\boldsymbol{H}_O = \begin{bmatrix} \boldsymbol{L}_O \\ \boldsymbol{p} \end{bmatrix} 
= \begin{bmatrix} \boldsymbol{L}_G + \boldsymbol{r}_{O G} \times \boldsymbol{p} \\ \boldsymbol{p}  \end{bmatrix}
$$

上式の世界座標系で時間微分を行う。

### 3.1. 世界座標での時間微分

角運動量$\boldsymbol{L}_O$の時間微分は以下である。

$$
\frac{d \ \boldsymbol{L}_O}{dt} = \frac{d \ \boldsymbol{L}_G}{dt} + \frac{d \ \boldsymbol{r}_{O G}}{dt} \times \boldsymbol{p} + \boldsymbol{r}_{O G} \times \frac{d \ \boldsymbol{p}}{dt}
$$

角運動量の時間微分はモーメントになり、並進運動量の時間微分は力になる。

$$
\frac{d \ \boldsymbol{L}_G}{dt} = \boldsymbol{n}_G , \quad \frac{d \ \boldsymbol{p}}{dt} = \boldsymbol{F}
$$

$\boldsymbol{r}_{O G}$ は剛体座標系の原点$O_B$を用いて以下である。

$$
\boldsymbol{r}_{O G} = \boldsymbol{r}_{O_B G} - \boldsymbol{r}_{O_B O}
$$

よって、$\boldsymbol{r}_{O G}$ の時間微分は以下である。

$$
\frac{d \ \boldsymbol{r}_{O G}}{dt} = \boldsymbol{v}_{G} - \boldsymbol{v}_{O}
$$

これより、以下となる。

$$
\begin{aligned}
\frac{d \ \boldsymbol{L}_O}{dt} &= \boldsymbol{n}_G + (\boldsymbol{v}_{G} - \boldsymbol{v}_{O}) \times \boldsymbol{p} + \boldsymbol{r}_{O G} \times \boldsymbol{F} \\
&= \boldsymbol{n}_G + m \boldsymbol{v}_{G} \times \boldsymbol{v}_G - \boldsymbol{v}_{O} \times \boldsymbol{p} + \boldsymbol{r}_{O G} \times \boldsymbol{F} \\
&= \boldsymbol{n}_G  - \boldsymbol{v}_{O} \times \boldsymbol{p} + \boldsymbol{r}_{O G} \times \boldsymbol{F}
\end{aligned}
$$

$\boldsymbol{n}_O = \boldsymbol{n}_G + \boldsymbol{r}_{O G} \times \boldsymbol{F}$ であるため上式は以下となる。

$$
\frac{d \ \boldsymbol{L}_O}{dt} = \boldsymbol{n}_O -  \boldsymbol{v}_{O} \times \boldsymbol{p}
$$

よって、点$O$周りのSpatial Momentum の世界座標での時間微分は以下となる。

$$
\frac{d}{dt} \boldsymbol{H}_O = \begin{bmatrix} \dfrac{d \ \boldsymbol{L}_O}{dt} \\[8pt] \dfrac{d \ \boldsymbol{p}}{dt} \end{bmatrix} 
= \begin{bmatrix} \boldsymbol{n}_O -  \boldsymbol{v}_{O} \times \boldsymbol{p} \\ \boldsymbol{F}  \end{bmatrix}
$$


### 3.2 剛体座標への変換

#### 3.2.1 角運動量の時間微分

3.1の表記は世界座標で表したものであるので、変換行列${^B}R_{W}$ を両辺に掛け、剛体座標系$\Sigma_B$への変換を行う。

$$
{^B}R_{W} \ \frac{d \ \boldsymbol{L}_O}{dt} = {^B}R_{W} \boldsymbol{n}_O -  ({^B}R_{W} \ \boldsymbol{v}_{O}) \times ({^B}R_{W} \ \boldsymbol{p})
$$

上式のベクトルを剛体座標系で表記すると以下となる。

$$
{^B}R_{W} \frac{d \ \boldsymbol{L}_O}{dt} = {^B}\boldsymbol{n}_O - {^B}\boldsymbol{v}_{O} \times {^B}\boldsymbol{p}
$$

2.1節で導出した輸送定理を上式の左辺に適用すると、以下となる。

$$
{^B}R_{W} \frac{d \ \boldsymbol{L}_O}{dt} = {^B} \dot{\boldsymbol{L}}_O + {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{L}_O
$$

上式を左辺へ代入するといかとなる。

$$
{^B} \dot{\boldsymbol{L}}_O + {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{L}_O =  {^B}\boldsymbol{n}_O - {^B}\boldsymbol{v}_{O} \times {^B}\boldsymbol{p}
$$

ここから、剛体座標系の点$O$を基準点としたモーメントの剛体座標系における表記は次のようになる。

$$
\boxed{
{^B}\boldsymbol{n}_O = {^B} \dot{\boldsymbol{L}}_O + {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{L}_O + {^B}\boldsymbol{v}_{O} \times {^B}\boldsymbol{p}
}
$$

#### 3.2.2 並進運動量の時間微分

変換行列${^B}R_{W}$ を両辺に掛け、剛体座標系$\Sigma_B$への変換を行う。

$$
{^B}R_{W} \ \dfrac{d \ \boldsymbol{p}}{dt} = {^B}R_{W} \ F
$$

ベクトルを剛体座標系で示すと以下である。

$$
{^B}R_{W} \ \dfrac{d \ \boldsymbol{p}}{dt} = {^B}F
$$

左辺に輸送定理を適用する。

$$
{^B}R_{W} \ \dfrac{d \ \boldsymbol{p}}{dt} = {^B}\dot{\boldsymbol{p}} + {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{p}
$$

これより、剛体に作用する力の剛体座標系における表記は次のようになる。

$$
\boxed{
{^B}F = {^B}\dot{\boldsymbol{p}} + {^B}\boldsymbol{\omega} \times {^B}\boldsymbol{p}
}
$$

#### 3.2.3 剛体座標系におけるモーメントと力

結果をまとめると、以下となる。

$$
\begin{bmatrix}
{^B}\boldsymbol{n}_O \\ {^B}F
\end{bmatrix} =
\begin{bmatrix} {^B} \dot{\boldsymbol{L}}_O \\ {^B}\dot{\boldsymbol{p}}
\end{bmatrix} +
\begin{bmatrix}
{^B}\boldsymbol{\omega} \times {^B}\boldsymbol{L}_O + {^B}\boldsymbol{v}_{O} \times {^B}\boldsymbol{p} \\
{^B}\boldsymbol{\omega} \times {^B}\boldsymbol{p}
\end{bmatrix}
$$


### 4. Spatial Newton-Euler方程式の導出

剛体座標系で表した評価点$O$でのSpatial Momentumを以下とする。

$$
{^B}\boldsymbol{H}_O = \begin{bmatrix} {^B}\boldsymbol{L}_O \\ {^B}\boldsymbol{p} \end{bmatrix}
$$

剛体座標系で表した評価点$O$でのSpatial Force Vectorを以下とする。

$$
{^B}\boldsymbol{f}_O = \begin{bmatrix} {^B}\boldsymbol{n}_O \\ {^B}F \end{bmatrix}
$$

これを用いて、3.2.3 で導出した式は以下となる。

$$
{^B}\boldsymbol{f}_O = {^B}\dot{\boldsymbol{H}}_O + \begin{bmatrix}
[{^B}\boldsymbol{\omega}]^\land & [{^B}\boldsymbol{v}_{O}]^\land  \\
[{^B}\boldsymbol{\omega}]^\land & 0
\end{bmatrix}
{^B}\boldsymbol{H}_O
$$

剛体座標系で表した評価点$$のSpatial Motion Vectorを以下とする。

$$
{^B}\boldsymbol{V}_O = \begin{bmatrix} {^B}\boldsymbol{\omega} \\ {^B}\boldsymbol{v}{_O} \end{bmatrix}
$$

[このNote](./Spatial_Acceleration_and_Cross_Product.ipynb)の3.5節で導出した crf を用いると上式の行列を crf で記述できる。

$$
\text{crf}({^B}\boldsymbol{V}_O) = \begin{bmatrix}
[{^B}\boldsymbol{\omega}]^\land & [{^B}\boldsymbol{v}_{O}]^\land  \\
0 & [{^B}\boldsymbol{\omega}]^\land
\end{bmatrix}
$$

よって、以下のSpatial Newton-Euler方程式となる。

$$
{^B}\boldsymbol{f}_O = {^B}\dot{\boldsymbol{H}}_O + \text{crf}({^B}\boldsymbol{V}_O){^B}\boldsymbol{H}_O
$$

#### 4.1 Spaital Acceleration と Spatial Inertiaの導入

[このNote](./spatial_vector.ipynb)の4.2節でSpatial Inartia 、Spatial Momentum、Spatial Motion Vectorの関係を示した。この表記は座標系によらず一定である。

$$
\boldsymbol{H}_O = \mathcal{I}_O \boldsymbol{V}_O
$$

剛体座標系で表現されており、基準点$O$が剛体に固定され質量分布が時間変化しない剛体であれば、Spaitl Inertia は時間変化しない。よって、剛体座標系におけるSpatial Momentun の時間微分は以下となる。

$$
{^B}\dot{\boldsymbol{H}}_O = {^B}\mathcal{I}_O {^B}\dot{\boldsymbol{V}}_O
$$

固定座標系におけるSpatial Motion Vectorの時間微分は、Spatial Acceleration になる。

$$
{^B}\dot{\boldsymbol{V}}_O = {^B}{\boldsymbol{A}}_O
$$

これより、以下のSpatial Newton-Euler方程式が導出される。

$$
{^B}\boldsymbol{f}_O = {^B}\mathcal{I}_O {^B}{\boldsymbol{A}}_O + \text{crf}({^B}\boldsymbol{V}_O){^B}\mathcal{I}_O {^B}\boldsymbol{V}_O
$$

この式は剛体に固定された任意の点$O$を基準点としたNewton-Euler方程式である。

#### 4.2 式の内容

左上の座標系記号$B$を省略し、上式を展開して考える。

- 第1項

$$
\mathcal{I}_O = \begin{bmatrix}
\boldsymbol{I}_G - m \boldsymbol{S}^2 & m \boldsymbol{S} \\
- m \boldsymbol{S} & m \boldsymbol{I}_3 
\end{bmatrix} \in \mathbf{R}^{6 \times 6} , \quad (\boldsymbol{r} = \boldsymbol{r}_{OG}, \ \boldsymbol{S} = [\boldsymbol{r}]^\land )
$$

$$
\boldsymbol{A}_O = \begin{bmatrix} {\boldsymbol{\alpha}} \\
{\boldsymbol{a}}_O - {\boldsymbol{\omega}} \times {\boldsymbol{v}}_O 
\end{bmatrix}
$$

より

$$
\mathcal{I}_O {\boldsymbol{A}}_O 
= \begin{bmatrix}
(\boldsymbol{I}_G - m \boldsymbol{S}^2){\boldsymbol{\alpha}} +  m \boldsymbol{S} ({\boldsymbol{a}}_O - {\boldsymbol{\omega}} \times {\boldsymbol{v}}_O 
) \\
- m \boldsymbol{S}{\boldsymbol{\alpha}} + m({\boldsymbol{a}}_O - {\boldsymbol{\omega}} \times {\boldsymbol{v}}_O 
)
\end{bmatrix}
$$

- 第2項

$$
\boldsymbol{V}_O = \begin{bmatrix} \boldsymbol{\omega} \\ \boldsymbol{v}{_O} \end{bmatrix}
$$

より、

$$
\mathcal{I}_O \boldsymbol{V}_O = 
\begin{bmatrix}
(\boldsymbol{I}_G - m \boldsymbol{S}^2) \boldsymbol{\omega} +  m \boldsymbol{S} \boldsymbol{v}{_O}\\
- m \boldsymbol{S} \boldsymbol{\omega} +  m \boldsymbol{v}{_O}
\end{bmatrix}
$$

$$
\text{crf}(\boldsymbol{V}_O) = \begin{bmatrix}
[\boldsymbol{\omega}]^\land & [\boldsymbol{v}_{O}]^\land  \\
0 & [\boldsymbol{\omega}]^\land
\end{bmatrix}
$$

より、

$$
\text{crf}(\boldsymbol{V}_O)\mathcal{I}_O \boldsymbol{V}_O = \begin{bmatrix}
[\boldsymbol{\omega}]^\land \{ (\boldsymbol{I}_G - m \boldsymbol{S}^2) \boldsymbol{\omega} +  m \boldsymbol{S} \boldsymbol{v}{_O} \} + [\boldsymbol{v}_{O}]^\land \{ - m \boldsymbol{S} \boldsymbol{\omega} +  m \boldsymbol{v}{_O} \} \\
[\boldsymbol{\omega}]^\land \{ - m \boldsymbol{S} \boldsymbol{\omega} +  m \boldsymbol{v}{_O} \}
\end{bmatrix}
$$

上記のような計算となる。

点$O$が重心となる場合、上式から$\boldsymbol{S}=0$とした計算になる。
