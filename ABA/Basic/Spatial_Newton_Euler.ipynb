# Spatial Newton Euler Method (空間ニュートン・オイラー方程式)

Spatial Newton Euler Methodは以下の方程式式である。

$$
\boldsymbol{f} = \boldsymbol{I} \boldsymbol{a} + \text{crf}(\boldsymbol{V}) \boldsymbol{I} \boldsymbol{V}
$$

- $\boldsymbol{f}$ : Spatial Force Vector
- $\boldsymbol{I}$ : Spatial Inertia
- $\boldsymbol{a}$ : Spatial Acceleration
- $\boldsymbol{V}$ : Spatial Motion Vector
- $\text{crf}(\boldsymbol{V})$ : Spatial Force Cross Product

この式は、剛体に作用する合力・合モーメントを表す運動方程式である。

本Noteでは従来の方程式からSpatial Newton Euler Method の導出を行う。


## 1. 従来のNewton-Euler 方程式

剛体の重心$G$を基準点とする。

- 並進運動

    $$
    \boldsymbol{F} = m \boldsymbol{a}_G
    $$

    - $m$: 剛体の質量
    - $\boldsymbol{a}_G$ : 重心の加速度ベクトル

- [回転運動](../../3D-EoM/Basic/euler-equation-of-motion.ipynb)

    $$
    \boldsymbol{n}_G = \boldsymbol{I}_G \dot{\boldsymbol{\omega}} + \boldsymbol{\omega} \times (\boldsymbol{I}_G\boldsymbol{\omega})
    $$

    - $\boldsymbol{I}_G$ : 重心まわりの慣性テンソル
    - $\dot{\boldsymbol{\omega}}$ : 剛体の角加速度ベクトル
    - ${\boldsymbol{\omega}}$ : 剛体の角速度ベクトル

上式をまとめることで、Spatial Newton-Euler 方程式を構築することが出来る。

## 2. 重心を原点すると座標系

最初は簡単にするため、剛体の座標系の原点$O$に重心$G$を一致させた状態で考える。

全てのベクトルは世界座標系と平行な軸で表現するとする。

### 2.1 Spatial Motion Vector

重心$G$を基準点とするとSpatial Motion Vectorは以下である。

$$
\boldsymbol{V}_G = \begin{bmatrix} \boldsymbol{\omega} \\ \boldsymbol{v}_G \end{bmatrix} \in \mathbf{R}^{6 \times 1}
$$

- $\boldsymbol{\omega}$ : 剛体の角速度
- $\boldsymbol{v}_G$ : 重心$G$の並進速度

### 2.2 Spatial Inertia 

点$O$における[Spatial Inertia](./spatial_vector.ipynb)は以下であった。

$$
\mathcal{I}_O = 
\begin{bmatrix}
\boldsymbol{I}_G - m [\boldsymbol{r}_{OG}]^\land [\boldsymbol{r}_{OG}]^\land & m [\boldsymbol{r}_{OG}]^\land \\
- m [\boldsymbol{r}_{OG}]^\land & m \boldsymbol{I}_3 
\end{bmatrix} \in \mathbf{R}^{6 \times 6}
$$

重心$G$を評価点とすると、上式の$\boldsymbol{r}_{OG}$ は $\boldsymbol{r}_{GG}=0$ となる。よって重心$G$におけるとSpatial Inertiaは以下である。

$$
\mathcal{I}_G = 
\begin{bmatrix}
\boldsymbol{I}_G  & 0 \\
0 & m \boldsymbol{I}_3 
\end{bmatrix}
$$

### 2.3 Spatial Acceleration

重心$G$を評価点とする[Spatial Acceleration](./Spatial_Acceleration_and_Cross_Product.ipynb) は以下となる。

$$
\boldsymbol{A}_G = \begin{bmatrix} {\boldsymbol{\alpha}} \\
{\boldsymbol{a}}_G - {\boldsymbol{\alpha}} \times {\boldsymbol{v}}_G \end{bmatrix}
$$

- $\dot{\boldsymbol{\omega}}$ : 剛体の角加速度
- ${\boldsymbol{a}}_G$ : 重心$G$の加速度

### 2.4 Spatial Force Cross Product

$\boldsymbol{V}_G$におけるcrfは以下である。

$$
\text{crf}(\boldsymbol{V}_G) = \begin{bmatrix}
[\boldsymbol{\omega}]^\land & [\boldsymbol{v}_G]^\land \\
0 & [\boldsymbol{\omega}]^\land
\end{bmatrix}
$$
